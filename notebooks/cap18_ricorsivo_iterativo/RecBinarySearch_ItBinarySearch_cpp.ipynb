## Ricorsione in coda: RecBinarySearch e ItBinarySearch

Una chiamata ricorsiva è *in coda* se è l'ultima operazione e il suo risultato è restituito così com'è: si può sostituire con l'aggiornamento dei parametri e un salto all'inizio, cioè con un ciclo (condizione del ciclo = negazione del caso base; chiamata in coda = assegnamento simultaneo dei parametri). Il risultato è lo stesso, il numero di passi ha lo stesso ordine di grandezza e la memoria extra scende da $\Theta(\text{profondità})$ a $\Theta(1)$.

**RecBS**$(a,l,r,m)$ / **ItBS**$(a,l,r,m)$ — *Input*: un array ordinato, due indici $l$, $r$ e un valore $m$. *Output*: un indice di $m$ in $a[l..r]$ oppure $-1$. `ItBS(a, 1, n, m)` è la ricerca binaria iterativa del capitolo «Ricerca». Nello pseudocodice del capitolo si chiamano `RecBinarySearch` e `ItBinarySearch`.

Sull'array $a=[3,9,11,12,12,16,19]$ del capitolo si cercano $m=12,10,3,19,100$: per ogni valore l'indice restituito (da 0), le chiamate di `RecBS` e le iterazioni di `ItBS`.

```python
%%writefile rec_it_binary_search.cpp
#include <iostream>
#include <vector>
#include <string>
#include <algorithm>
#include <cmath>
#include <cstdio>
#include <chrono>
#include <climits>
#include <functional>
#include <array>
using namespace std;

int rbsCalls = 0;        // chiamate di RecBS
int bsIters = 0;         // iterazioni di ItBS

// RecBS(a, l, r, m)
int RecBS(int *a, int l, int r, int m)
{
    rbsCalls++;
    if (l > r)
    {
        return -1;
    }
    int mid = l + (r - l) / 2;
    if (a[mid] == m)
    {
        return mid;
    }
    if (a[mid] < m)
    {
        return RecBS(a, mid + 1, r, m);     // chiamata in coda
    }
    else
    {
        return RecBS(a, l, mid - 1, m);     // chiamata in coda
    }
}

// ItBS(a, l, r, m)
int ItBS(int *a, int l, int r, int m)
{
    while (l <= r)                          // negazione del caso base l > r
    {
        bsIters++;
        int mid = l + (r - l) / 2;
        if (a[mid] == m)
        {
            return mid;                     // il secondo caso base resta nel ciclo
        }
        if (a[mid] < m)
        {
            l = mid + 1;
        }
        else
        {
            r = mid - 1;
        }
    }
    return -1;
}

string Fmt(const int *a, int n)
{
    string s = "[";
    for (int i = 0; i < n; i++)
    {
        if (i > 0) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

#ifndef NOMAIN
int main()
{
    int a[] = {3, 9, 11, 12, 12, 16, 19};
    int n = 7;
    printf("a = %s   n = %d\n", Fmt(a, n).c_str(), n);
    printf("%4s %6s %6s %6s %11s\n", "m", "RecBS", "calls", "ItBS", "iterations");
    int ms[] = {12, 10, 3, 19, 100};
    for (int m : ms)
    {
        rbsCalls = 0;
        int r1 = RecBS(a, 0, n - 1, m);
        bsIters = 0;
        int r2 = ItBS(a, 0, n - 1, m);
        printf("%4d %6d %6d %6d %11d   %s\n", m, r1, rbsCalls, r2, bsIters, r1 == r2 ? "OK" : "ERROR");
    }
    return 0;
}
#endif
```

**Prova tu.** Cerca altri valori, presenti o assenti, in un altro array ordinato: confronta le chiamate di `RecBS` con le iterazioni di `ItBS`.

```python
%%writefile prova_rec_it_binary_search.cpp
#define NOMAIN
#include "rec_it_binary_search.cpp"

int main()
{
    // Cerca altri valori, presenti o assenti: confronta le chiamate di RecBS con le iterazioni di ItBS.
    int b[] = {2, 5, 8, 11, 14, 17, 20, 23};
    int n = 8;
    int ms[] = {14, 1, 30};
    for (int m : ms)
    {
        rbsCalls = 0;
        int r1 = RecBS(b, 0, n - 1, m);
        bsIters = 0;
        int r2 = ItBS(b, 0, n - 1, m);
        printf("%d %d %d %d %d\n", m, r1, rbsCalls, r2, bsIters);
    }
    return 0;
}
```
