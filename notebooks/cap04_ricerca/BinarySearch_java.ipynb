## BinarySearch

**Input:** un array $a[1..n]$ di $n$ numeri naturali **ordinato in ordine non decrescente** e un valore cercato $m$. **Output:** l'indice di $m$ in $a$, oppure $-1$ se $m$ non c'è.

L'algoritmo mantiene l'intervallo di ricerca $a[left..right]$ e a ogni iterazione confronta $m$ con l'elemento centrale $a[mid]$, dimezzando l'intervallo: esegue al più $\lfloor \log_2 n \rfloor + 1$ iterazioni. Se $m$ compare più volte restituisce l'indice di una delle sue occorrenze, non necessariamente la prima.

```python
%%writefile BinarySearch.java
import java.util.Arrays;

public class BinarySearch {

    // Restituisce l'indice di m in a[0..n-1] (ordinato), oppure -1 se m non c'è
    static int binarySearch(int[] a, int n, int m, boolean trace) {
        int left = 0;                   // estremo sinistro (1 nello pseudocodice)
        int right = n - 1;              // estremo destro (n nello pseudocodice)
        int iteration = 0;              // solo per la traccia

        while (left <= right) {
            iteration++;
            int mid = left + (right - left) / 2;    // indice centrale

            if (a[mid] == m) {
                if (trace)
                    System.out.printf("%10d %5d %5d %4d  a[%d] = %d = m => Trovato! Restituisce %d\n", iteration, left, right, mid, mid, a[mid], mid);
                return mid;             // elemento trovato
            }

            if (a[mid] < m) {
                if (trace)
                    System.out.printf("%10d %5d %5d %4d  a[%d] = %d < %d => left = %d + 1 = %d\n", iteration, left, right, mid, mid, a[mid], m, mid, mid + 1);
                left = mid + 1;         // cerca nella metà destra
            } else {
                if (trace)
                    System.out.printf("%10d %5d %5d %4d  a[%d] = %d > %d => right = %d - 1 = %d\n", iteration, left, right, mid, mid, a[mid], m, mid, mid - 1);
                right = mid - 1;        // cerca nella metà sinistra
            }
        }
        if (trace)
            System.out.printf("%10s %5d %5d %4s  left > right => Non trovato! Restituisce -1\n", "-", left, right, "-");
        return -1;                      // elemento non trovato
    }

    static void run(int[] a, int m) {
        int n = a.length;
        System.out.printf("Ricerca: m = %d, a = %s, n = %d\n", m, Arrays.toString(a), n);
        System.out.printf("%10s %5s %5s %4s  %s\n", "iterazione", "left", "right", "mid", "azione");
        int result = binarySearch(a, n, m, true);
        System.out.printf("Restituisce: %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 1) {          // «Prova tu»: primo numero = m, gli altri = array ordinato
            int[] v = new int[args.length - 1];
            for (int k = 1; k < args.length; k++) v[k - 1] = Integer.parseInt(args[k]);
            run(v, Integer.parseInt(args[0]));
            return;
        }
        int[] a = {3, 9, 11, 12, 12, 16, 19};
        run(a, 12);
        run(a, 10);
        run(a, 3);
        run(a, 19);
        run(new int[]{1, 2, 2, 2, 2, 2, 3}, 2);     // m compare più volte: restituisce una sua occorrenza
    }
}
```
