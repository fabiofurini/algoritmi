## LinearSearch

**Input:** un array $a[1..n]$ di $n$ numeri naturali e un valore cercato $m$. **Output:** l'indice di $m$ in $a$, oppure $-1$ se $m$ non c'è.

L'algoritmo confronta $m$ con gli elementi dell'array uno dopo l'altro e si ferma al primo uguale a $m$: l'indice restituito è il più piccolo $k$ con $a[k]=m$. Nel codice gli indici partono da 0, quindi anche l'indice restituito è quello dello pseudocodice meno 1.

```python
def linear_search(a, n, m, trace=False):
    i = 0                               # i = 1 nello pseudocodice
    while i < n:                        # i <= n nello pseudocodice
        if a[i] == m:
            if trace:
                print("%10d %3d  %-16s  %s" % (i + 1, i, "a[%d] = %d = m" % (i, a[i]), "Trovato! Restituisce %d" % i))
            return i                    # elemento trovato in posizione i
        if trace:
            print("%10d %3d  %-16s  %s" % (i + 1, i, "a[%d] = %d != m" % (i, a[i]), "continua"))
        i = i + 1
    if trace:
        print("%10s %3d  %-16s  %s" % ("-", i, "-", "i = n: il ciclo termina, non trovato! Restituisce -1"))
    return -1                           # elemento non trovato
```

### Esecuzione sugli esempi del capitolo

```python
def run(a, m):
    n = len(a)
    print("Ricerca: m = %d, a = %s, n = %d" % (m, a, n))
    print("%10s %3s  %-16s  %s" % ("iterazione", "i", "confronto", "azione"))
    result = linear_search(a, n, m, trace=True)
    print("Restituisce: %d" % result)
    print()


a = [4, 2, 7, 1, 5, 9, 3]
run(a, 5)
run(a, 6)
run([1, 2, 2, 2, 2, 2, 3], 2)    # m compare più volte: restituisce il primo indice
```
