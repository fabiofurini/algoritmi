## BinarySearch

**Input:** un array $a[1..n]$ di $n$ numeri naturali **ordinato in ordine non decrescente** e un valore cercato $m$. **Output:** l'indice di $m$ in $a$, oppure $-1$ se $m$ non c'è.

L'algoritmo mantiene l'intervallo di ricerca $a[left..right]$ e a ogni iterazione confronta $m$ con l'elemento centrale $a[mid]$, dimezzando l'intervallo: esegue al più $\lfloor \log_2 n \rfloor + 1$ iterazioni. Se $m$ compare più volte restituisce l'indice di una delle sue occorrenze, non necessariamente la prima.

```python
%%writefile binary_search.cpp
#include <cstdarg>
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

// Funzioni di servizio per stampare la traccia
string Str(int n, const int *a)     // l'array come "[4, 2, 7]"
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Restituisce l'indice di m in a[0..n-1] (ordinato), oppure -1 se m non c'è
int BinarySearch(int n, int *a, int m, bool trace = false)
{
    int left = 0;                       // estremo sinistro (1 nello pseudocodice)
    int right = n - 1;                  // estremo destro (n nello pseudocodice)
    int iteration = 0;                  // solo per la traccia

    while (left <= right)
    {
        iteration++;
        int mid = left + (right - left) / 2;    // indice centrale

        if (a[mid] == m)
        {
            if (trace)
                printf("%10d %5d %5d %4d  a[%d] = %d = m => Trovato! Restituisce %d\n", iteration, left, right, mid, mid, a[mid], mid);
            return mid;                 // elemento trovato
        }

        if (a[mid] < m)
        {
            if (trace)
                printf("%10d %5d %5d %4d  a[%d] = %d < %d => left = %d + 1 = %d\n", iteration, left, right, mid, mid, a[mid], m, mid, mid + 1);
            left = mid + 1;             // cerca nella metà destra
        }
        else
        {
            if (trace)
                printf("%10d %5d %5d %4d  a[%d] = %d > %d => right = %d - 1 = %d\n", iteration, left, right, mid, mid, a[mid], m, mid, mid - 1);
            right = mid - 1;            // cerca nella metà sinistra
        }
    }
    if (trace)
        printf("%10s %5d %5d %4s  left > right => Non trovato! Restituisce -1\n", "-", left, right, "-");
    return -1;                          // elemento non trovato
}

void Run(vector<int> v, int m)
{
    int n = v.size();
    printf("Ricerca: m = %d, a = %s, n = %d\n", m, Str(n, v.data()).c_str(), n);
    printf("%10s %5s %5s %4s  %s\n", "iterazione", "left", "right", "mid", "azione");
    int result = BinarySearch(n, v.data(), m, true);
    printf("Restituisce: %d\n\n", result);
}

int main(int argc, char *argv[])
{
    if (argc > 2)   // «Prova tu»: primo numero = m, gli altri = array ordinato
    {
        vector<int> v;
        for (int k = 2; k < argc; k++) v.push_back(atoi(argv[k]));
        Run(v, atoi(argv[1]));
        return 0;
    }
    vector<int> a = {3, 9, 11, 12, 12, 16, 19};
    Run(a, 12);
    Run(a, 10);
    Run(a, 3);
    Run(a, 19);
    Run({1, 2, 2, 2, 2, 2, 3}, 2);      // m compare più volte: restituisce una sua occorrenza
    return 0;
}
```
