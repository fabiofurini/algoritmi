## LinearSearch

**Input:** un array $a[1..n]$ di $n$ numeri naturali e un valore cercato $m$. **Output:** l'indice di $m$ in $a$, oppure $-1$ se $m$ non c'è.

L'algoritmo confronta $m$ con gli elementi dell'array uno dopo l'altro e si ferma al primo uguale a $m$: l'indice restituito è il più piccolo $k$ con $a[k]=m$. Nel codice gli indici partono da 0, quindi anche l'indice restituito è quello dello pseudocodice meno 1.

```python
%%writefile LinearSearch.java
import java.util.Arrays;

public class LinearSearch {

    // Restituisce l'indice di m in a[0..n-1], oppure -1 se m non c'è
    static int linearSearch(int[] a, int n, int m, boolean trace) {
        for (int i = 0; i < n; i++) {   // i = 1..n nello pseudocodice
            if (a[i] == m) {
                if (trace)
                    System.out.printf("%10d %3d  %-16s  %s\n", i + 1, i, "a[" + i + "] = " + a[i] + " = m", "Trovato! Restituisce " + i);
                return i;               // elemento trovato in posizione i
            }
            if (trace)
                System.out.printf("%10d %3d  %-16s  %s\n", i + 1, i, "a[" + i + "] = " + a[i] + " != m", "continua");
        }
        if (trace)
            System.out.printf("%10s %3d  %-16s  %s\n", "-", n, "-", "i = n: il ciclo termina, non trovato! Restituisce -1");
        return -1;                      // elemento non trovato
    }

    static void run(int[] a, int m) {
        int n = a.length;
        System.out.printf("Ricerca: m = %d, a = %s, n = %d\n", m, Arrays.toString(a), n);
        System.out.printf("%10s %3s  %-16s  %s\n", "iterazione", "i", "confronto", "azione");
        int result = linearSearch(a, n, m, true);
        System.out.printf("Restituisce: %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 1) {          // «Prova tu»: primo numero = m, gli altri = array
            int[] v = new int[args.length - 1];
            for (int k = 1; k < args.length; k++) v[k - 1] = Integer.parseInt(args[k]);
            run(v, Integer.parseInt(args[0]));
            return;
        }
        int[] a = {4, 2, 7, 1, 5, 9, 3};
        run(a, 5);
        run(a, 6);
        run(new int[]{1, 2, 2, 2, 2, 2, 3}, 2);     // m compare più volte: restituisce il primo indice
    }
}
```
