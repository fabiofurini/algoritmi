**Indici.** Nello pseudocodice gli array sono indicizzati da 1; nel codice Python da 0: l'elemento `a[i]` dello pseudocodice è `a[i-1]` nel codice (e un ciclo `while i <= n` da 1 diventa `while i < n` da 0). Le tabelle stampate usano gli indici dello pseudocodice.

## Strumenti: il contatore dei passi

La cella definisce l'array `cnt` (una casella per riga dello pseudocodice), `reset()` per azzerarlo e `print_steps()` per stampare la tabella delle esecuzioni.

```python
cnt = [0] * 20      # cnt[r] = numero di esecuzioni della riga r (righe numerate come nelle dispense)


def reset():
    for r in range(len(cnt)):
        cnt[r] = 0


def print_steps(*lines):
    print("Riga | Esecuzioni")
    for r in lines:
        print("%4d | %10d" % (r, cnt[r]))
    print("Passi totali S = %d" % sum(cnt))
```

## 1. Ricerca del massimo: FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il massimo valore di $a$. Nel caso peggiore (array crescente) $S(n) = 4n = \Theta(n)$.

Pseudocodice: sezione 12.2 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi-ricorsione/12-tempo-esecuzione/).

```python
def find_maximum(a):
    n = len(a)
    max_val = a[0]; cnt[2] += 1        # max := a[1]
    i = 1; cnt[3] += 1                 # i := 2 nello pseudocodice
    cnt[4] += 1                        # la condizione del while si conta a ogni valutazione
    while i < n:                       # i <= n nello pseudocodice
        cnt[5] += 1
        if a[i] > max_val:
            max_val = a[i]; cnt[6] += 1
        i = i + 1; cnt[8] += 1
        cnt[4] += 1
    cnt[10] += 1
    return max_val
```

```python
def steps_of(a):
    reset()
    find_maximum(a)
    return sum(cnt)


# Esempio delle dispense: a[1..4] = (3, 1, 4, 5)
a = [3, 1, 4, 5]
n = len(a)
reset()
m = find_maximum(a)
print("a = %s -> max = %d" % (a, m))
print_steps(2, 3, 4, 5, 6, 8, 10)
t = cnt[6]
print("sum t(i) = %d, S(n) = 3n + 1 + sum t(i) = %d" % (t, 3 * n + 1 + t))

# Caso peggiore (array crescente) e caso migliore (array decrescente)
print()
a = [1, 2, 3, 4]
print("a = %s (crescente): S = %d = 4n = %d" % (a, steps_of(a), 4 * n))
a = [4, 3, 2, 1]
print("a = %s (decrescente): S = %d = 3n + 1 = %d" % (a, steps_of(a), 3 * n + 1))

print()
print("Array crescente (1, 2, ..., n):")
print("  n | S(n) | 4n")
for n in range(1, 9):
    print("%3d | %4d | %d" % (n, steps_of(list(range(1, n + 1))), 4 * n))
```

**Prova tu.** Cambia l'array: i passi stanno sempre fra $3n + 1$ e $4n$.

```python
a = [7, 2, 9, 4, 9, 1]      # <-- cambia l'array
n = len(a)
reset()
m = find_maximum(a)
print("a = %s -> max = %d" % (a, m))
print("S = %d, tra 3n + 1 = %d e 4n = %d" % (sum(cnt), 3 * n + 1, 4 * n))
```

## 2. Ordinamento per inserimento: InsertionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in modo non decrescente. Se $t(i)$ è il numero di spostamenti all'iterazione $i$, allora $S(n) = 6n - 4 + 3\sum_{i=2}^{n} t(i)$; nel caso peggiore (array decrescente) $S(n) = \frac{3n^2+9n-8}{2} = \Theta(n^2)$.

Pseudocodice: sezione 12.3 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi-ricorsione/12-tempo-esecuzione/).

```python
def insertion_sort(a, trace=False):
    n = len(a)
    prev = 0                               # solo per la traccia
    i = 1; cnt[2] += 1                     # i := 2 nello pseudocodice
    cnt[3] += 1
    while i < n:                           # i <= n nello pseudocodice
        key = a[i]; cnt[4] += 1
        j = i - 1; cnt[5] += 1
        cnt[6] += 1                        # la condizione del while si conta a ogni valutazione
        while j >= 0 and a[j] > key:       # j >= 1 and a[j] > key nello pseudocodice
            a[j + 1] = a[j]; cnt[7] += 1
            j = j - 1; cnt[8] += 1
            cnt[6] += 1
        a[j + 1] = key; cnt[10] += 1
        i = i + 1; cnt[11] += 1
        cnt[3] += 1
        if trace:                          # traccia: dopo l'incremento, i e' l'indice dello pseudocodice dell'iterazione appena finita
            print("%2d | %3d | %4d | %s" % (i, key, cnt[7] - prev, a))
            prev = cnt[7]
```

```python
def steps_of(a):
    reset()
    insertion_sort(a)
    return sum(cnt)


# Esempio delle dispense: a[1..6] = (5, 2, 4, 6, 1, 3)
a = [5, 2, 4, 6, 1, 3]
n = len(a)
print("a = %s" % a)
reset()
print(" i | key | t(i) | array dopo l'iterazione")
insertion_sort(a, True)
print("ordinato: %s" % a)
print_steps(2, 3, 4, 5, 6, 7, 8, 10, 11)
t = cnt[7]
print("sum t(i) = %d, S(n) = 6n - 4 + 3 sum t(i) = %d" % (t, 6 * n - 4 + 3 * t))

# Caso peggiore: array decrescente, t(i) = i - 1
print()
a = [6, 5, 4, 3, 2, 1]
print("a = %s (decrescente): S = %d = (3n^2 + 9n - 8)/2 = %d" % (a, steps_of(a[:]), (3 * n * n + 9 * n - 8) // 2))
print("sum t(i) = %d = n(n-1)/2 = %d" % (cnt[7], n * (n - 1) // 2))

print()
print("Array decrescente (n, ..., 2, 1):")
print("  n | S(n) | (3n^2 + 9n - 8)/2")
for n in range(1, 9):
    print("%3d | %4d | %d" % (n, steps_of(list(range(n, 0, -1))), (3 * n * n + 9 * n - 8) // 2))
```

**Prova tu.** Cambia l'array e controlla la formula $6n - 4 + 3\sum t(i)$.

```python
a = [8, 3, 5, 1, 9, 2, 7]      # <-- cambia l'array
n = len(a)
reset()
print(" i | key | t(i) | array dopo l'iterazione")
insertion_sort(a, True)
print_steps(2, 3, 4, 5, 6, 7, 8, 10, 11)
print("6n - 4 + 3 sum t(i) = %d" % (6 * n - 4 + 3 * cnt[7]))
```

## 3. Cambio di base: ChangeToBase2

**Input:** un numero naturale $n \ge 1$ in base 10. **Output:** l'array $d[0..nd-1]$ delle cifre binarie e il numero di cifre $nd$. Per ogni $n$: $S(n) = 6\lfloor\log_2 n\rfloor + 11 = \Theta(\log_2 n)$.

Pseudocodice: sezione 12.4 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi-ricorsione/12-tempo-esecuzione/).

```python
def floor_log2(n):                         # floor(log2 n) calcolato con un ciclo (non conta come passo)
    k = 0
    while n > 1:
        n = n // 2
        k = k + 1
    return k


def change_to_base2(n, trace=False):
    nd = floor_log2(n) + 1; cnt[2] += 1
    d = [0] * nd; cnt[3] += 1              # array d[0..nd-1], indicizzato da 0 come nello pseudocodice
    k = 0; cnt[4] += 1
    cnt[5] += 1                            # la condizione del while si conta a ogni valutazione
    while n > 0:
        q = n // 2; cnt[6] += 1
        r = n % 2; cnt[7] += 1
        d[k] = r; cnt[8] += 1
        if trace:                          # traccia
            print("%10d | %3d | %3d | %3d | d[%d] = %d" % (k + 1, n, q, r, k, r))
        n = q; cnt[9] += 1
        k = k + 1; cnt[10] += 1
        cnt[5] += 1
    cnt[12] += 1
    return d, nd
```

```python
# Esempio delle dispense: n = 13
n = 13
reset()
print("iterazione |   n |   q |   r | assegnamento")
d, nd = change_to_base2(n, True)
print("n = %d = (%s)_2: nd = %d, cifre d[nd-1..0] = %s" % (n, "".join(str(x) for x in reversed(d)), nd, "".join(str(x) for x in reversed(d))))
print_steps(2, 3, 4, 5, 6, 7, 8, 9, 10, 12)
print("S(n) = 6 floor(log2 n) + 11 = %d" % (6 * floor_log2(n) + 11))

# Il numero di passi cresce come log2 n
print()
print("       n | nd | S(n) | 6 floor(log2 n) + 11")
for n in [1, 2, 13, 100, 1000, 1000000]:
    reset()
    d, nd = change_to_base2(n)
    print("%8d | %2d | %4d | %d" % (n, nd, sum(cnt), 6 * floor_log2(n) + 11))
```

**Prova tu.** Cambia $n$ e verifica che $S(n) = 6\lfloor\log_2 n\rfloor + 11$.

```python
n = 2026        # <-- cambia il numero
reset()
d, nd = change_to_base2(n)
print("n = %d = (%s)_2, nd = %d, S = %d" % (n, "".join(str(x) for x in reversed(d)), nd, sum(cnt)))
```

## 4. Test di primalità: PrimalityTest

**Input:** un numero naturale $n$. **Output:** `true` se $n$ è primo, `false` altrimenti. Nel caso peggiore ($n$ primo) $S(n) = 3\lfloor\sqrt{n}\rfloor + 1 = \Theta(\sqrt{n})$, cioè $2^{\Theta(|I|)}$ in funzione della dimensione dell'input.

Pseudocodice: sezione 12.5 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi-ricorsione/12-tempo-esecuzione/).

```python
from math import isqrt             # floor(sqrt(n)) in aritmetica intera


def primality_test(n):
    cnt[2] += 1
    if n < 2:
        cnt[3] += 1
        return False
    d = 2; cnt[5] += 1
    cnt[6] += 1                        # la condizione del while si conta a ogni valutazione
    while d <= isqrt(n):
        cnt[7] += 1
        if n % d == 0:
            cnt[8] += 1
            return False
        d = d + 1; cnt[10] += 1
        cnt[6] += 1
    cnt[12] += 1
    return True
```

```python
def risposta(r):
    return "true" if r else "false"


# Esempio delle dispense: un numero primo e uno composto
n = 97
reset()
r = primality_test(n)
print("n = %d: %s" % (n, risposta(r)))
print_steps(2, 3, 5, 6, 7, 8, 10, 12)
print("S(n) = 3 floor(sqrt(n)) + 1 = %d" % (3 * isqrt(n) + 1))

print()
n = 91
reset()
r = primality_test(n)
p = 2
while n % p != 0:
    p = p + 1
print("n = %d: %s, p = %d" % (n, risposta(r), p))
print_steps(2, 3, 5, 6, 7, 8, 10, 12)
print("S(n) = 3p - 1 = %d, mentre floor(sqrt(n)) = %d" % (3 * p - 1, isqrt(n)))

print()
n = 1
reset()
r = primality_test(n)
print("n = %d: %s, S = %d" % (n, risposta(r), sum(cnt)))

# Il caso peggiore sono i numeri primi: S cresce come sqrt(n), cioe' raddoppia ogni 2 bit in piu'
print()
print("|I| |        n |  S(n) | 3 floor(sqrt(n)) + 1")
for n in [1021, 4093, 16381, 65521, 262139, 1048573, 4194301, 16777213]:
    reset()
    r = primality_test(n)
    print("%3d | %8d | %5d | %d" % (n.bit_length(), n, sum(cnt), 3 * isqrt(n) + 1))
```

**Prova tu.** Cambia $n$: per un primo $S = 3\lfloor\sqrt{n}\rfloor + 1$, per un composto $S = 3p - 1$.

```python
n = 7919        # <-- cambia il numero
reset()
r = primality_test(n)
print("n = %d: %s, S = %d, 3 floor(sqrt(n)) + 1 = %d" % (n, "true" if r else "false", sum(cnt), 3 * isqrt(n) + 1))
```
