## ChangeOfBase

**Input:** un numero naturale $n \ge 1$ (in base 10) e una base $b \ge 2$. **Output:** l'array $d[1..nd]$ delle cifre di $n$ in base $b$ e il numero di cifre $nd$.

Una divisione per $b$ estrae la cifra meno significativa ($r = n \bmod b$) e lascia come quoziente il numero scritto con le altre cifre: si ripete finché il quoziente è 0. Le cifre escono dalla meno significativa alla più significativa, quindi nell'array `d` la cifra di peso $b^0$ sta in `d[0]` (in $d[1]$ nello pseudocodice).

Il numero di cifre $nd = \lfloor \log_b n \rfloor + 1$ è calcolato qui con le sole moltiplicazioni intere, senza logaritmi in virgola mobile: in doppia precisione `log(1000)/log(10)` vale $2.9999999999999996$ e la formula darebbe 3 cifre invece di 4.

```python
def change_of_base(n, b, trace=False):
    nd = 1                              # nd = ⌊log_b n⌋ + 1, con sole operazioni intere
    p = b
    while p <= n:
        p = p * b
        nd = nd + 1
    d = [0] * nd                        # array delle cifre
    k = 0                               # k = 1 nello pseudocodice
    while n > 0:
        q = n // b                      # quoziente
        r = n % b                       # resto (cifra in base b)
        d[k] = r
        n = q
        k = k + 1
        if trace:
            print("%10d %5d %3d %5d %3d  %s" % (k, q, r, n, k, d[:k]))
    return d, nd
```

### Esecuzione sugli esempi del capitolo

```python
def digit_string(d, nd):                # le cifre dalla più alla meno significativa
    return "".join("0123456789ABCDEF"[d[i]] for i in range(nd - 1, -1, -1))


def run(n, b):
    print("Conversione: n = %d, b = %d" % (n, b))
    print("%10s %5s %3s %5s %3s  %s" % ("iterazione", "q", "r", "n", "nd", "cifre"))
    d, nd = change_of_base(n, b, trace=True)
    print("Restituisce: nd = %d, d = %s" % (nd, d))
    print("%d = (%s)_%d" % (n, digit_string(d, nd), b))
    print()


run(7, 2)
run(8, 2)
run(25, 3)
run(100, 8)
run(1000, 10)    # 4 cifre: il caso in cui log(1000)/log(10) in virgola mobile sbaglia
```

## BasicPrimalityTest

**Input:** un numero naturale $n$. **Output:** `true` se $n$ è primo, `false` altrimenti.

Un divisore $d < n$ di $n$ è al più $n/2$: basta provare i candidati $d = 2, 3, \dots, \lfloor n/2 \rfloor$. Per un numero primo il ciclo viene eseguito per intero: per $n = 29$ i candidati sono 13. La funzione restituisce anche il numero di candidati provati.

```python
def basic_primality_test(n, trace=False):
    tested = 0                          # candidati provati
    if n < 2:
        if trace:
            print("%3s %8s  %s" % ("-", "-", "n < 2: restituisce false"))
        return False, tested            # 0 e 1 non sono primi
    for d in range(2, n // 2 + 1):      # d = 2..⌊n/2⌋
        tested += 1
        if n % d == 0:
            if trace:
                print("%3d %8d  %s" % (d, n % d, "restituisce false: d divide n"))
            return False, tested        # d divide n: non è primo
        if trace:
            print("%3d %8d  %s" % (d, n % d, "continua"))
    if trace:
        print("%3s %8s  %s" % ("-", "-", "il ciclo termina: restituisce true"))
    return True, tested                 # nessun divisore: n è primo
```

### Esecuzione sugli esempi del capitolo

```python
def run(n):
    print("Controllo: n = %d, floor(n/2) = %d" % (n, n // 2))
    print("%3s %8s  %s" % ("d", "n mod d", "azione"))
    result, tested = basic_primality_test(n, trace=True)
    print("Restituisce: %s, candidati provati: %d" % (str(result).lower(), tested))
    print()


run(36)
run(29)
run(25)
```

## PrimalityTest

**Input:** un numero naturale $n$. **Output:** `true` se $n$ è primo, `false` altrimenti.

I divisori vengono in coppie $d$ e $n/d$ e in ogni coppia uno dei due è al più $\sqrt{n}$: un numero $n \ge 2$ è primo se e solo se nessun numero di $\{2, 3, \dots, \lfloor \sqrt{n} \rfloor\}$ lo divide. Per $n = 2$ e $n = 3$ il ciclo non viene eseguito e la funzione restituisce `true`. Per $n = 29$ i candidati sono 4, invece dei 13 dell'algoritmo di base.

```python
from math import isqrt                 # parte intera della radice quadrata, esatta sugli interi


def primality_test(n, trace=False):
    tested = 0                          # candidati provati
    if n < 2:
        if trace:
            print("%3s %8s  %s" % ("-", "-", "n < 2: restituisce false"))
        return False, tested            # 0 e 1 non sono primi
    d = 2
    while d <= isqrt(n):                # d <= ⌊√n⌋
        tested += 1
        if n % d == 0:
            if trace:
                print("%3d %8d  %s" % (d, n % d, "restituisce false: d divide n"))
            return False, tested        # d divide n: non è primo
        if trace:
            print("%3d %8d  %s" % (d, n % d, "continua"))
        d = d + 1
    if trace:
        print("%3s %8s  %s" % ("-", "-", "il ciclo termina: restituisce true"))
    return True, tested                 # nessun divisore: n è primo
```

### Esecuzione sugli esempi del capitolo

```python
def run(n):
    print("Controllo: n = %d, floor(sqrt(n)) = %d" % (n, isqrt(n)))
    print("%3s %8s  %s" % ("d", "n mod d", "azione"))
    result, tested = primality_test(n, trace=True)
    print("Restituisce: %s, candidati provati: %d" % (str(result).lower(), tested))
    print()


run(36)
run(29)
run(25)
run(2)       # il ciclo non viene eseguito
```

## SieveOfEratosthenes

**Input:** un numero naturale $n \ge 2$. **Output:** un array booleano $prime[1..n]$ in cui $prime[i] = 1$ se e solo se $i$ è primo.

Si parte con tutti i numeri segnati come primi (tranne 1) e, per ogni primo $p \le \lfloor \sqrt{n} \rfloor$, si segnano come composti i suoi multipli a partire da $p^2$. In Python, C++ e Java l'array parte da 0: `prime[i]` è associato al numero $i$, quindi ha $n+1$ elementi `prime[0..n]` e `prime[0]` (il numero 0, non primo) vale 0.

```python
from math import isqrt


def sieve_of_eratosthenes(n, trace=False):
    prime = [1] * (n + 1)               # prime[0..n], tutti a 1
    prime[0] = 0                        # 0 non è primo
    prime[1] = 0                        # 1 non è primo
    if trace:
        print("%5s  %s" % ("start", " ".join(map(str, prime))))
    p = 2
    while p <= isqrt(n):                # p <= ⌊√n⌋
        js = []                         # i valori di j (solo per la traccia)
        if prime[p] == 1:
            j = p * p                   # si parte da p^2
            while j <= n:
                prime[j] = 0            # j non è primo
                js.append(j)
                j = j + p
        if trace:
            print("%5d  %s  %s" % (p, " ".join(map(str, prime)),
                  "j = " + ", ".join(map(str, js)) if js else "nessun ciclo interno: prime[%d] è 0" % p))
        p = p + 1
    return prime
```

### Esecuzione sugli esempi del capitolo

```python
def run(n):
    print("Crivello: n = %d, floor(sqrt(n)) = %d" % (n, isqrt(n)))
    print("%5s  %s" % ("p", "prime[0..%d]" % n))
    prime = sieve_of_eratosthenes(n, trace=True)
    print("Numeri primi <= %d: %s" % (n, [i for i in range(n + 1) if prime[i] == 1]))
    print()


run(10)
run(20)
```

## EuclideanAlgorithm

**Input:** due numeri naturali $n, m \ge 1$. **Output:** $\gcd(n, m)$, il massimo comun divisore.

La coppia $(n, m)$ viene sostituita con $(m,\ n \bmod m)$, che ha gli stessi divisori comuni, finché il secondo numero è 0: allora $\gcd(n, 0) = n$. I parametri sono passati per valore: la funzione modifica le proprie copie. Due numeri sono coprimi se e solo se il risultato è 1.

```python
def euclidean_algorithm(n, m, trace=False):
    step = 0                            # solo per la traccia
    while m != 0:
        step += 1
        r = n % m
        if trace:
            print("%5d %5d %5d  %d mod %d = %d" % (step, n, m, n, m, r))
        n = m
        m = r
    if trace:
        print("%5d %5d %5d  %s" % (step + 1, n, m, "--- (il ciclo termina)"))
    return n                            # m = 0: n è il MCD
```

### Esecuzione sugli esempi del capitolo

```python
def run(n, m):
    print("Calcolo del MCD: n = %d, m = %d" % (n, m))
    print("%5s %5s %5s  %s" % ("passo", "n", "m", "r = n mod m"))
    g = euclidean_algorithm(n, m, trace=True)
    print("Restituisce: gcd(%d, %d) = %d" % (n, m, g))
    print("coprimi: %s" % str(g == 1).lower())
    print()


run(48, 18)
run(252, 105)
run(35, 12)
```
