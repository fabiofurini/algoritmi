## FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il valore massimo dell'array.

La variabile `max` parte dal primo elemento e viene aggiornata ogni volta che si incontra un elemento più grande.

```python
def find_maximum(a, n, trace=False):
    max_val = a[0]                  # parte dal primo elemento
    if trace:
        print("%3s  %4s  %4d  %s" % ("--", "--", max_val, "--"))
    for i in range(1, n):           # i = 2..n nello pseudocodice
        bigger = a[i] > max_val     # confronto, salvato per la traccia
        if bigger:
            max_val = a[i]          # aggiorna il massimo
        if trace:
            print("%3d  %4d  %4d  %s" % (i, a[i], max_val, "Sì" if bigger else "No"))
    return max_val
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s  %4s  %4s  %s" % ("i", "a[i]", "max", "a[i]>max?"))
    result = find_maximum(a, n, trace=True)
    print("Restituisce: max = %d" % result)
    print()


run("Istanza 1", [9, 12, 3, 16, 19, 11, 12])
run("Istanza 2", [3, 7, 2, 9, 1])
```

## CountPositive

**Input:** un array $a[1..n]$ di $n$ numeri interi. **Output:** il numero di elementi positivi dell'array.

```python
def count_positive(a, n, trace=False):
    count = 0
    if trace:
        print("%3s  %4s  %5s  %s" % ("--", "--", count, "--"))
    for i in range(n):              # i = 1..n nello pseudocodice
        positive = a[i] > 0         # confronto, salvato per la traccia
        if positive:
            count = count + 1
        if trace:
            print("%3d  %4d  %5d  %s" % (i, a[i], count, "Sì" if positive else "No"))
    return count
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s  %4s  %5s  %s" % ("i", "a[i]", "count", "a[i]>0?"))
    result = count_positive(a, n, trace=True)
    print("Restituisce: count = %d" % result)
    print()


run("Istanza 1", [-2, 5, -1, 3, 0])
run("Istanza 2", [4, -3, 7, 2])
```

## IsSorted

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** `true` se i valori dell'array sono in ordine non decrescente, `false` altrimenti.

L'istruzione `return` dentro il ciclo termina la funzione appena trova due elementi consecutivi nell'ordine sbagliato.

```python
def is_sorted(a, n, trace=False):
    for i in range(n - 1):          # i = 1..n-1 nello pseudocodice
        wrong = a[i] > a[i + 1]     # confronto, salvato per la traccia
        if trace:
            print("%3d  %4d  %6d  %s" % (i, a[i], a[i + 1], "Sì" if wrong else "No"))
        if wrong:
            return False
    return True
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s  %4s  %6s  %s" % ("i", "a[i]", "a[i+1]", "a[i]>a[i+1]?"))
    result = is_sorted(a, n, trace=True)
    print("Restituisce: %s" % str(result).lower())
    print()


run("Istanza 1", [2, 5, 7, 9])
run("Istanza 2", [3, 1, 4, 2])
```

## ScalarProduct

**Input:** due array $a[1..n]$ e $b[1..n]$ di dimensione $n \ge 1$. **Output:** il prodotto scalare $\sum_{i=1}^{n} a[i] \cdot b[i]$ dei due array.

```python
def scalar_product(a, b, n, trace=False):
    prod = a[0] * b[0]
    if trace:
        print("%3s  %4s  %4s  %9s  %5d" % ("--", "--", "--", "--", prod))
    for i in range(1, n):           # i = 2..n nello pseudocodice
        prod = prod + a[i] * b[i]
        if trace:
            print("%3d  %4d  %4d  %9d  %5d" % (i, a[i], b[i], a[i] * b[i], prod))
    return prod
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a, b):
    n = len(a)
    print("%s: a = %s, b = %s, n = %d" % (title, a, b, n))
    print("%3s  %4s  %4s  %9s  %5s" % ("i", "a[i]", "b[i]", "a[i]*b[i]", "prod"))
    result = scalar_product(a, b, n, trace=True)
    print("Restituisce: prod = %d" % result)
    print()


run("Istanza 1", [1, 2, 3], [4, 5, 6])
run("Istanza 2", [2, 0, -1], [3, 4, 5])
```

## ReverseArray

**Input:** un array $a[1..n]$ di $n$ numeri. **Output:** lo stesso array con gli elementi in ordine inverso.

L'elemento $a[i]$ viene scambiato con l'elemento simmetrico $a[n-i+1]$; bastano $\lfloor n/2 \rfloor$ scambi (se $n$ è dispari l'elemento centrale resta al suo posto).

```python
def reverse_array(a, n, trace=False):
    if trace:
        print("%3s  %-14s  %s" % ("--", "--", a))
    for i in range(n // 2):         # i = 1..⌊n/2⌋ nello pseudocodice
        tmp = a[i]
        a[i] = a[n - i - 1]         # a[n-i+1] nello pseudocodice
        a[n - i - 1] = tmp
        if trace:
            print("%3d  %-14s  %s" % (i, "a[%d] <-> a[%d]" % (i, n - i - 1), a))
    return a
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d, %d iterazioni" % (title, a, n, n // 2))
    print("%3s  %-14s  %s" % ("i", "scambio", "array a (dopo lo scambio)"))
    result = reverse_array(a, n, trace=True)
    print("Restituisce: a = %s" % result)
    print()


run("Istanza 1", [1, 2, 3, 4, 5])
run("Istanza 2", [10, 20, 30, 40])
```
