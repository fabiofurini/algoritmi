## 1. RecursiveSum e il conteggio dei passi

**RecursiveSum**$(a,n)$ — *Input*: un array $a[1..n]$ di $n\ge1$ interi. *Output*: la somma $a[1]+\dots+a[n]$.

Il codice è fedele allo pseudocodice (righe 2, 3, 5, 6); un contatore `steps` somma i passi eseguiti da ogni chiamata, come nel capitolo: 2 passi se $n=1$, 3 passi più la chiamata ricorsiva se $n>1$. La seconda funzione è `sumRecDecrease` del codice del corso, che si ferma all'array vuoto ($n=0$) e quindi esegue una chiamata in più.

```python
steps = 0   # passi eseguiti (righe di pseudocodice)
calls = 0   # numero di chiamate

def recursive_sum(a, n):
    """Somma di a[0..n-1] (pseudocodice: a[1..n]), n >= 1."""
    global steps, calls
    calls += 1
    steps += 1                    # riga 2: if n = 1
    if n == 1:
        steps += 1                # riga 3: return a[1]
        return a[0]
    steps += 2                    # righe 5 e 6
    s = recursive_sum(a, n - 1)
    return s + a[n - 1]

def sum_rec_decrease(a, n):
    """Variante del codice del corso: caso base n = 0."""
    global calls
    calls += 1
    if n == 0:
        return 0
    return sum_rec_decrease(a, n - 1) + a[n - 1]

a = [3, 1, 4, 1, 5, 9, 2, 6]
print("a =", a)
calls = 0
print("RecursiveSum(a, 8) =", recursive_sum(a, len(a)), "con", calls, "chiamate")
calls = 0
print("sumRecDecrease(a, 8) =", sum_rec_decrease(a, len(a)), "con", calls, "chiamate")
print("n  S(n) misurato  3n-1")
for n in range(1, 9):
    steps = 0
    recursive_sum(a, n)
    print(f"{n:<2d} {steps:>12d} {3 * n - 1:>5d}")
```

**Prova tu.** Cambia l'array `a` (anche la lunghezza) e verifica che i passi misurati siano sempre $3n-1$.

```python
# Cambia l'array e controlla che i passi misurati siano sempre 3n-1.
a = [7, -2, 5, 0, 11]
steps = 0
print(recursive_sum(a, len(a)), steps, 3 * len(a) - 1)
```

## 2. Torre di Hanoi: $T(n)=2T(n-1)+1$

**Hanoi**$(n, from, to, aux)$ — *Input*: il numero di dischi $n\ge1$ e i tre pioli. *Output*: la sequenza di mosse che trasferisce i dischi $1,\dots,n$ da *from* a *to*.

Il codice conta le mosse (e, per $n=3$, le stampa): il capitolo mostra che sono esattamente $2^n-1$. La chiamata iniziale è `Hanoi(n, 'A', 'C', 'B')`.

```python
moves = 0       # numero di mosse
log = []        # elenco delle mosse (solo per stamparle)

def hanoi(n, src, dst, aux):
    """Hanoi(n, from, to, aux)"""
    global moves
    if n == 1:
        moves += 1                              # sposta il disco 1 da src a dst
        log.append(f"disco 1: {src} -> {dst}")
        return
    hanoi(n - 1, src, aux, dst)
    moves += 1                                  # sposta il disco n
    log.append(f"disco {n}: {src} -> {dst}")
    hanoi(n - 1, aux, dst, src)

hanoi(3, 'A', 'C', 'B')
print("Mosse per n = 3:")
for m in log:
    print("  " + m)
print("n   mosse  2^n-1")
for n in [1, 2, 3, 4, 5, 6, 7, 8, 10]:
    moves = 0
    log = []
    hanoi(n, 'A', 'C', 'B')
    print(f"{n:<3d} {moves:>5d} {2 ** n - 1:>5d}")
```

**Prova tu.** Cambia il numero di dischi $n$: le mosse crescono come $2^n-1$ (con $n=15$ sono già $32\,767$).

```python
# Prova con più dischi: quante mosse servono per n = 15?
n = 15
moves = 0
log = []
hanoi(n, 'A', 'C', 'B')
print(n, moves, 2 ** n - 1)
```

## 3. Ricorrenze calcolate direttamente e forme chiuse

Ogni ricorrenza è calcolata con una funzione ricorsiva che segue la definizione e confrontata con la forma chiusa ottenuta nel capitolo (metodo dell'espansione). Per le ricorrenze con $n/2$ i valori di $n$ sono potenze di 2.

| ricorrenza | forma chiusa |
|---|---|
| $T(n)=T(n-1)+n$, $T(1)=1$ | $n(n+1)/2$ |
| $S(n)=S(n-1)+3$, $S(1)=2$ (RecursiveSum) | $3n-1$ |
| $T(n)=2T(n-1)+1$, $T(1)=1$ (Hanoi) | $2^n-1$ |
| $T(n)=T(n/2)+1$, $T(1)=1$ | $\log_2 n+1$ |
| $T(n)=2T(n/2)+n$, $T(1)=1$ | $n\log_2 n+n$ |

```python
def t_quad(n):  return 1 if n == 1 else t_quad(n - 1) + n
def t_sum(n):   return 2 if n == 1 else t_sum(n - 1) + 3
def t_hanoi(n): return 1 if n == 1 else 2 * t_hanoi(n - 1) + 1
def t_log(n):   return 1 if n == 1 else t_log(n // 2) + 1
def t_nlogn(n): return 1 if n == 1 else 2 * t_nlogn(n // 2) + n

def confronta(nome, valori, ric, chiusa):
    print(nome)
    for n in valori:
        r, c = ric(n), chiusa(n)
        print(f"  n={n:<7d} {r:>12d} {c:>12d}  {'OK' if r == c else 'DIVERSI'}")

def log2(n):  # n è una potenza di 2
    return n.bit_length() - 1

confronta("T(n) = T(n-1) + n  ->  n(n+1)/2", [1, 2, 3, 4, 10, 100], t_quad, lambda n: n * (n + 1) // 2)
confronta("S(n) = S(n-1) + 3  ->  3n-1", [1, 2, 3, 4, 100], t_sum, lambda n: 3 * n - 1)
confronta("T(n) = 2T(n-1) + 1  ->  2^n-1", [1, 2, 3, 4, 10, 20], t_hanoi, lambda n: 2 ** n - 1)
confronta("T(n) = T(n/2) + 1  ->  log2(n)+1", [1, 2, 4, 8, 1024, 2 ** 20], t_log, lambda n: log2(n) + 1)
confronta("T(n) = 2T(n/2) + n  ->  n log2(n) + n", [1, 2, 4, 8, 1024, 2 ** 20], t_nlogn, lambda n: n * log2(n) + n)
```

**Prova tu.** Aggiungi una ricorrenza tua (qui $T(n)=T(n-1)+2n-1$, $T(1)=1$) e confrontala con la sua forma chiusa.

```python
# Aggiungi una ricorrenza: T(n) = T(n-1) + 2n - 1 con T(1) = 1 ha forma chiusa n^2.
def t_mia(n): return 1 if n == 1 else t_mia(n - 1) + 2 * n - 1
confronta("T(n) = T(n-1) + 2n - 1  ->  n^2", [1, 2, 3, 10, 100], t_mia, lambda n: n * n)
```
