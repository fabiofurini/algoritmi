## Strumenti: il contatore dei passi

La cella definisce l'array `cnt` (una casella per riga dello pseudocodice), `reset()` per azzerarlo e `print_steps()` per stampare la tabella delle esecuzioni.

```python
cnt = [0] * 20      # cnt[r] = numero di esecuzioni della riga r (righe numerate come nelle dispense)


def reset():
    for r in range(len(cnt)):
        cnt[r] = 0


def print_steps(*lines):
    print("Riga | Esecuzioni")
    for r in lines:
        print("%4d | %10d" % (r, cnt[r]))
    print("Passi totali S = %d" % sum(cnt))
```

## 1. La dimensione dell'input degli esempi

**Input:** un numero naturale, due numeri naturali o un array di numeri naturali. **Output:** la dimensione dell'input $|I|$, cioè il numero di bit necessari per scriverlo in binario ($\lfloor \log_2 n \rfloor + 1$ bit per un numero $n \ge 1$).

Si veda: sezioni 11.1, 11.3, 11.4, 11.5 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi-ricorsione/11-dimensione-input/).

```python
def bits(n):
    """Numero di bit di un naturale n >= 1: floor(log2 n) + 1."""
    k = 0
    while n > 0:
        n = n // 2        # si dimezza fino ad arrivare a 0
        k = k + 1
    return k


def binary(n):
    """Cifre di n in base 2 (come stringa)."""
    s = ""
    while n > 0:
        s = str(n % 2) + s
        n = n // 2
    return s
```

```python
from math import log2

# Esempio delle dispense: il numero di bit di alcune istanze
n = 37
print("Un numero: n = %d = (%s)_2, |I| = %d" % (n, binary(n), bits(n)))
m = 200
print("Due numeri: n = %d, m = %d = (%s)_2, |I| = %d + %d = %d" % (n, m, binary(m), bits(n), bits(m), bits(n) + bits(m)))
a = [6, 1, 3, 8]
b = [bits(x) for x in a]
print("Un array: a = %s, bit per voce = %s, |I| = %d" % (a, b, sum(b)))

# I limiti delle dispense (valori arrotondati a due decimali)
print()
print("Un numero: log2(n) = %.2f <= |I| = %d <= 2 log2(n) = %.2f" % (log2(n), bits(n), 2 * log2(n)))
s = log2(n) + log2(m)
print("Due numeri: log2(n) + log2(m) = %.2f <= |I| = %d <= 2 (log2(n) + log2(m)) = %.2f" % (s, bits(n) + bits(m), 2 * s))
print("Due numeri e massimo: log2(max) = %.2f <= |I| = %d <= 4 log2(max) = %.2f" % (log2(max(n, m)), bits(n) + bits(m), 4 * log2(max(n, m))))
print("Un array: n = %d <= |I| = %d <= n (1 + log2 M) = %.2f" % (len(a), sum(b), len(a) * (1 + log2(max(a)))))

# Il limite superiore non risulta stretto: a[1] = M = 2^n, a[2..n] = 1 ha |I| = 2n, ma n(1 + log2 M) = n(n + 1)
print()
print("   n |      M |  |I| = 2n | n (1 + log2 M)")
for n in range(1, 11):
    M = 2 ** n
    a = [M] + [1] * (n - 1)
    print("%4d | %6d | %9d | %13d" % (n, M, sum(bits(x) for x in a), n * (n + 1)))
```

## 2. Somma delle componenti di un array: SumArray

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri naturali. **Output:** $a[1]+a[2]+\cdots+a[n]$. Numero di passi: $S(n) = 3n + 4$, qualunque siano i valori.

Pseudocodice: sezione 11.6 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi-ricorsione/11-dimensione-input/).

```python
def sum_array(a):
    n = len(a)
    s = 0; cnt[2] += 1
    i = 0; cnt[3] += 1                 # i := 1 nello pseudocodice
    cnt[4] += 1                        # la condizione del while si conta a ogni valutazione
    while i < n:                       # i <= n nello pseudocodice
        s = s + a[i]; cnt[5] += 1
        i = i + 1; cnt[6] += 1
        cnt[4] += 1
    cnt[8] += 1
    return s
```

```python
a = [6, 1, 3, 8]
n = len(a)
reset()
s = sum_array(a)
print("a = %s, n = %d, somma = %d" % (a, n, s))
print_steps(2, 3, 4, 5, 6, 8)
print("Formula: S(n) = 3n + 4 = %d" % (3 * n + 4))

# Caso peggiore: l'array di |I| uni ha dimensione dell'input |I| e richiede esattamente 3|I| + 4 passi
print()
print("Array di |I| uni:")
for size in [4, 8, 16, 32]:
    reset()
    sum_array([1] * size)
    print("|I| = %2d: S = %3d = 3|I| + 4 = %d" % (size, sum(cnt), 3 * size + 4))
```

**Prova tu.** Cambia l'array e verifica che i passi siano sempre $3n + 4$.

```python
a = [5, 1, 9, 2, 7]      # <-- cambia l'array
reset()
s = sum_array(a)
print("somma = %d, S = %d, 3n + 4 = %d" % (s, sum(cnt), 3 * len(a) + 4))
```
