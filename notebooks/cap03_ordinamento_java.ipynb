## InsertionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in ordine non decrescente.

Al passo $i$ il sotto-array $a[1..i-1]$ è ordinato: la chiave $a[i]$ viene inserita al suo posto, dopo aver spostato di una posizione a destra le chiavi maggiori. La funzione restituisce anche i contatori usati nella dispensa: il numero di confronti tra chiavi e il numero di spostamenti (nessuno scambio: uno spostamento è una sola assegnazione).

```python
%%writefile InsertionSort.java
import java.util.Arrays;

public class InsertionSort {

    // Una riga della traccia
    static void row(String i, String j, String key, int[] a, String action) {
        System.out.printf("%3s %3s %4s  %-20s  %s\n", i, j, key, Arrays.toString(a), action);
    }

    // Ordina a[0..n-1] sul posto; restituisce {confronti tra chiavi, spostamenti}
    static int[] insertionSort(int[] a, int n, boolean trace) {
        int comparisons = 0;                // confronti tra chiavi
        int shifts = 0;                     // spostamenti
        if (trace)
            row("-", "-", "-", a, "stato iniziale");

        for (int i = 1; i < n; i++) {       // i = 2..n nello pseudocodice
            int key = a[i];                 // elemento da inserire
            int j = i - 1;
            if (trace)
                row("" + i, "" + j, "" + key, a, String.format("key <- a[%d]=%d; j <- %d", i, key, j));

            // sposta a destra le chiavi maggiori di key
            while (j >= 0 && a[j] > key) {
                comparisons++;
                shifts++;
                a[j + 1] = a[j];
                j = j - 1;
                if (trace)
                    row("", "" + j, "" + key, a, String.format("a[%d]=%d>%d: sposta a[%d] <- %d; j <- %d", j + 1, a[j + 1], key, j + 2, a[j + 1], j));
            }
            if (j >= 0)
                comparisons++;              // il confronto che ha fermato il ciclo (a[j] <= key)

            // inserisce key nella posizione giusta
            a[j + 1] = key;
            if (trace) {
                if (j < 0)
                    row("", "-", "" + key, a, String.format("j=%d: esce; inserisce a[%d] <- %d", j, j + 1, key));
                else
                    row("", "-", "" + key, a, String.format("a[%d]=%d<=%d: esce; inserisce a[%d] <- %d", j, a[j], key, j + 1, key));
            }
        }
        return new int[]{comparisons, shifts};
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s %3s %4s  %-20s  %s\n", "i", "j", "key", "a", "azione");
        int[] c = insertionSort(a, n, true);
        System.out.printf("Array ordinato: %s\n", Arrays.toString(a));
        System.out.printf("Confronti tra chiavi: %d, spostamenti: %d\n\n", c[0], c[1]);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // «Prova tu»: l'array arriva dalla riga di comando
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Prova tu", v);
            return;
        }
        run("Istanza 1", new int[]{5, 2, 4, 1, 3});
        run("Istanza 2", new int[]{12, 11, 13, 5, 6});
    }
}
```

## ExchangeSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in ordine non decrescente.

Al passo $i$ l'elemento $a[i]$ viene confrontato con ogni $a[j]$ con $j > i$ e i due elementi vengono scambiati ogni volta che $a[i] > a[j]$: alla fine del passo $a[i]$ è il minimo di $a[i..n]$. La funzione restituisce il numero di confronti tra chiavi e il numero di scambi.

```python
%%writefile ExchangeSort.java
import java.util.Arrays;

public class ExchangeSort {

    // Una riga della traccia
    static void row(String i, String j, int[] a, String action) {
        System.out.printf("%3s %3s  %-20s  %s\n", i, j, Arrays.toString(a), action);
    }

    // Ordina a[0..n-1] sul posto; restituisce {confronti tra chiavi, scambi}
    static int[] exchangeSort(int[] a, int n, boolean trace) {
        int comparisons = 0;                // confronti tra chiavi
        int swaps = 0;                      // scambi
        if (trace)
            row("-", "-", a, "stato iniziale");

        for (int i = 0; i < n - 1; i++) {       // i = 1..n-1 nello pseudocodice
            for (int j = i + 1; j < n; j++) {   // j = i+1..n nello pseudocodice
                comparisons++;
                int ai = a[i], aj = a[j];       // valori prima dello scambio, per la traccia
                if (a[i] > a[j]) {
                    int tmp = a[i];             // scambia a[i] e a[j]
                    a[i] = a[j];
                    a[j] = tmp;
                    swaps++;
                    if (trace)
                        row(j == i + 1 ? "" + i : "", "" + j, a, String.format("a[%d]=%d>a[%d]=%d: scambio a[%d] <-> a[%d]", i, ai, j, aj, i, j));
                } else if (trace)
                    row(j == i + 1 ? "" + i : "", "" + j, a, String.format("a[%d]=%d<=a[%d]=%d: nessuno scambio", i, ai, j, aj));
            }
        }
        return new int[]{comparisons, swaps};
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s %3s  %-20s  %s\n", "i", "j", "a", "azione");
        int[] c = exchangeSort(a, n, true);
        System.out.printf("Array ordinato: %s\n", Arrays.toString(a));
        System.out.printf("Confronti tra chiavi: %d, scambi: %d\n\n", c[0], c[1]);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // «Prova tu»: l'array arriva dalla riga di comando
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Prova tu", v);
            return;
        }
        run("Istanza 1", new int[]{5, 2, 4, 1, 3});
        run("Istanza 2", new int[]{12, 11, 13, 5, 6});
    }
}
```

## SelectionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in ordine non decrescente.

Al passo $i$ il ciclo interno cerca l'indice $\mathit{minIdx}$ del minimo di $a[i..n]$ (il confronto è stretto: in caso di parità vince il primo minimo) e poi, se $\mathit{minIdx} \ne i$, scambia $a[i]$ con $a[\mathit{minIdx}]$: al più uno scambio per passo. La funzione restituisce il numero di confronti tra chiavi e il numero di scambi.

```python
%%writefile SelectionSort.java
import java.util.Arrays;

public class SelectionSort {

    // Una riga della traccia
    static void row(String i, String j, String minIdx, int[] a, String action) {
        System.out.printf("%3s %3s %6s  %-20s  %s\n", i, j, minIdx, Arrays.toString(a), action);
    }

    // Ordina a[0..n-1] sul posto; restituisce {confronti tra chiavi, scambi}
    static int[] selectionSort(int[] a, int n, boolean trace) {
        int comparisons = 0;                // confronti tra chiavi
        int swaps = 0;                      // scambi
        if (trace)
            row("-", "-", "-", a, "stato iniziale");

        for (int i = 0; i < n - 1; i++) {   // i = 1..n-1 nello pseudocodice
            int minIdx = i;                 // indice del minimo in a[i..n]
            if (trace)
                row("" + i, "-", "" + minIdx, a, String.format("minIdx := %d (a[%d]=%d)", i, i, a[i]));
            for (int j = i + 1; j < n; j++) {   // j = i+1..n nello pseudocodice
                comparisons++;
                if (a[j] < a[minIdx]) {
                    int old = minIdx;       // solo per la traccia
                    minIdx = j;             // trovato un nuovo minimo
                    if (trace)
                        row("", "" + j, "" + minIdx, a, String.format("a[%d]=%d<a[%d]=%d: minIdx <- %d", j, a[j], old, a[old], j));
                } else if (trace)
                    row("", "" + j, "" + minIdx, a, String.format("a[%d]=%d>=a[%d]=%d: nessun aggiornamento", j, a[j], minIdx, a[minIdx]));
            }
            if (minIdx != i) {
                int tmp = a[i];             // scambia a[i] e a[minIdx]
                a[i] = a[minIdx];
                a[minIdx] = tmp;
                swaps++;
                if (trace)
                    row("", "-", "-", a, String.format("minIdx=%d != %d: scambio a[%d] <-> a[%d]", minIdx, i, i, minIdx));
            } else if (trace)
                row("", "-", "-", a, String.format("minIdx=%d: nessuno scambio", minIdx));
        }
        return new int[]{comparisons, swaps};
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s %3s %6s  %-20s  %s\n", "i", "j", "minIdx", "a", "azione");
        int[] c = selectionSort(a, n, true);
        System.out.printf("Array ordinato: %s\n", Arrays.toString(a));
        System.out.printf("Confronti tra chiavi: %d, scambi: %d\n\n", c[0], c[1]);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // «Prova tu»: l'array arriva dalla riga di comando
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Prova tu", v);
            return;
        }
        run("Istanza 1", new int[]{5, 2, 4, 1, 3});
        run("Istanza 2", new int[]{12, 11, 13, 5, 6});
    }
}
```

### Confronto dei tre algoritmi

Sugli stessi due array, i tre algoritmi eseguono (come nell'esempio di confronto della dispensa):

| algoritmo | $[5, 2, 4, 1, 3]$: confronti | movimenti | $[12, 11, 13, 5, 6]$: confronti | movimenti |
|---|---|---|---|---|
| Insertion Sort | 9 | 7 spostamenti | 9 | 7 spostamenti |
| Exchange Sort | 10 | 7 scambi | 10 | 7 scambi |
| Selection Sort | 10 | 3 scambi | 10 | 3 scambi |

Selection Sort esegue sempre $n(n-1)/2$ confronti ma al più $n-1$ scambi; Insertion Sort si adatta all'input (su un array già ordinato esegue solo $n-1$ confronti e nessuno spostamento) ed è l'unico dei tre ad essere stabile.
