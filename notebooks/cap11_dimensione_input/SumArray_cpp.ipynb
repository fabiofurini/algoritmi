## Somma delle componenti di un array: SumArray

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri naturali. **Output:** $a[1]+a[2]+\cdots+a[n]$. Numero di passi: $S(n) = 3n + 4$, qualunque siano i valori.

Pseudocodice: sezione 11.6 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/11-dimensione-input/).

```python
%%writefile sum_array.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Riga | Esecuzioni\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Passi totali S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// SumArray(a, n)
int SumArray(int a[], int n)
{
    int s = 0; cnt[2]++;
    int i = 0; cnt[3]++;               // i := 1 nello pseudocodice
    cnt[4]++;                          // la condizione del while si conta a ogni valutazione
    while (i < n)                      // i <= n nello pseudocodice
    {
        s = s + a[i]; cnt[5]++;
        i = i + 1; cnt[6]++;
        cnt[4]++;
    }
    cnt[8]++;
    return s;
}

#ifndef SOLO_LIBRERIA
int main()
{
    int a[] = {6, 1, 3, 8};
    int n = 4;
    Reset();
    int s = SumArray(a, n);
    printf("a = %s, n = %d, somma = %d\n", Fmt(a, 0, n).c_str(), n, s);
    PrintSteps({2, 3, 4, 5, 6, 8});
    printf("Formula: S(n) = 3n + 4 = %d\n", 3 * n + 4);

    // Caso peggiore: l'array di |I| uni ha dimensione dell'input |I| e richiede esattamente 3|I| + 4 passi
    printf("\n");
    printf("Array di |I| uni:\n");
    int sizes[] = {4, 8, 16, 32};
    for (int k = 0; k < 4; k++)
    {
        int size = sizes[k];
        int ones[32];
        for (int i = 0; i < size; i++) ones[i] = 1;
        Reset();
        SumArray(ones, size);
        printf("|I| = %2d: S = %3lld = 3|I| + 4 = %d\n", size, Total(), 3 * size + 4);
    }
    return 0;
}
#endif
```

**Prova tu.** Cambia l'array e verifica che i passi siano sempre $3n + 4$.

```python
%%writefile prova_sum_array.cpp
#define SOLO_LIBRERIA   // non eseguire il main di sum_array.cpp
#include "sum_array.cpp"

int main()
{
    int a[] = {5, 1, 9, 2, 7};      // <-- cambia l'array
    int n = 5;
    Reset();
    int s = SumArray(a, n);
    printf("somma = %d, S = %lld, 3n + 4 = %d\n", s, Total(), 3 * n + 4);
    return 0;
}
```
