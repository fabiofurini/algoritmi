**Indici.** Nello pseudocodice gli array sono indicizzati da 1; nel codice Python da 0: l'elemento `a[i]` dello pseudocodice è `a[i-1]` nel codice (e un ciclo `while i <= n` da 1 diventa `while i < n` da 0). Le tabelle stampate usano gli indici dello pseudocodice.

## Strumenti: il contatore dei passi

La cella definisce l'array `cnt` (una casella per riga dello pseudocodice), `reset()` per azzerarlo e `print_steps()` per stampare la tabella delle esecuzioni.

```python
cnt = [0] * 20      # cnt[r] = numero di esecuzioni della riga r (righe numerate come nelle dispense)


def reset():
    for r in range(len(cnt)):
        cnt[r] = 0


def print_steps(*lines):
    print("Riga | Esecuzioni")
    for r in lines:
        print("%4d | %10d" % (r, cnt[r]))
    print("Passi totali S = %d" % sum(cnt))
```

## Somma delle componenti di un array: SumArray

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri naturali. **Output:** $a[1]+a[2]+\cdots+a[n]$. Numero di passi: $S(n) = 3n + 4$, qualunque siano i valori.

Pseudocodice: sezione 11.6 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/11-dimensione-input/).

```python
def sum_array(a):
    n = len(a)
    s = 0; cnt[2] += 1
    i = 0; cnt[3] += 1                 # i := 1 nello pseudocodice
    cnt[4] += 1                        # la condizione del while si conta a ogni valutazione
    while i < n:                       # i <= n nello pseudocodice
        s = s + a[i]; cnt[5] += 1
        i = i + 1; cnt[6] += 1
        cnt[4] += 1
    cnt[8] += 1
    return s
```

```python
a = [6, 1, 3, 8]
n = len(a)
reset()
s = sum_array(a)
print("a = %s, n = %d, somma = %d" % (a, n, s))
print_steps(2, 3, 4, 5, 6, 8)
print("Formula: S(n) = 3n + 4 = %d" % (3 * n + 4))

# Caso peggiore: l'array di |I| uni ha dimensione dell'input |I| e richiede esattamente 3|I| + 4 passi
print()
print("Array di |I| uni:")
for size in [4, 8, 16, 32]:
    reset()
    sum_array([1] * size)
    print("|I| = %2d: S = %3d = 3|I| + 4 = %d" % (size, sum(cnt), 3 * size + 4))
```

**Prova tu.** Cambia l'array e verifica che i passi siano sempre $3n + 4$.

```python
a = [5, 1, 9, 2, 7]      # <-- cambia l'array
reset()
s = sum_array(a)
print("somma = %d, S = %d, 3n + 4 = %d" % (s, sum(cnt), 3 * len(a) + 4))
```
