## LinearSearch

**Input:** un array $a[1..n]$ di $n$ numeri naturali e un valore cercato $m$. **Output:** l'indice di $m$ in $a$, oppure $-1$ se $m$ non c'è.

L'algoritmo confronta $m$ con gli elementi dell'array uno dopo l'altro e si ferma al primo uguale a $m$: l'indice restituito è il più piccolo $k$ con $a[k]=m$. Nel codice gli indici partono da 0, quindi anche l'indice restituito è quello dello pseudocodice meno 1.

```python
def linear_search(a, n, m, trace=False):
    i = 0                               # i = 1 nello pseudocodice
    while i < n:                        # i <= n nello pseudocodice
        if a[i] == m:
            if trace:
                print("%10d %3d  %-16s  %s" % (i + 1, i, "a[%d] = %d = m" % (i, a[i]), "Trovato! Restituisce %d" % i))
            return i                    # elemento trovato in posizione i
        if trace:
            print("%10d %3d  %-16s  %s" % (i + 1, i, "a[%d] = %d != m" % (i, a[i]), "continua"))
        i = i + 1
    if trace:
        print("%10s %3d  %-16s  %s" % ("-", i, "-", "i = n: il ciclo termina, non trovato! Restituisce -1"))
    return -1                           # elemento non trovato
```

### Esecuzione sugli esempi del capitolo

```python
def run(a, m):
    n = len(a)
    print("Ricerca: m = %d, a = %s, n = %d" % (m, a, n))
    print("%10s %3s  %-16s  %s" % ("iterazione", "i", "confronto", "azione"))
    result = linear_search(a, n, m, trace=True)
    print("Restituisce: %d" % result)
    print()


a = [4, 2, 7, 1, 5, 9, 3]
run(a, 5)
run(a, 6)
run([1, 2, 2, 2, 2, 2, 3], 2)    # m compare più volte: restituisce il primo indice
```

## BinarySearch

**Input:** un array $a[1..n]$ di $n$ numeri naturali **ordinato in ordine non decrescente** e un valore cercato $m$. **Output:** l'indice di $m$ in $a$, oppure $-1$ se $m$ non c'è.

L'algoritmo mantiene l'intervallo di ricerca $a[left..right]$ e a ogni iterazione confronta $m$ con l'elemento centrale $a[mid]$, dimezzando l'intervallo: esegue al più $\lfloor \log_2 n \rfloor + 1$ iterazioni. Se $m$ compare più volte restituisce l'indice di una delle sue occorrenze, non necessariamente la prima.

```python
def binary_search(a, n, m, trace=False):
    left = 0                            # estremo sinistro (1 nello pseudocodice)
    right = n - 1                       # estremo destro (n nello pseudocodice)
    iteration = 0                       # solo per la traccia
    while left <= right:
        iteration += 1
        mid = left + (right - left) // 2    # indice centrale
        if a[mid] == m:
            if trace:
                print("%10d %5d %5d %4d  %s" % (iteration, left, right, mid, "a[%d] = %d = m => Trovato! Restituisce %d" % (mid, a[mid], mid)))
            return mid                  # elemento trovato
        if a[mid] < m:
            if trace:
                print("%10d %5d %5d %4d  %s" % (iteration, left, right, mid, "a[%d] = %d < %d => left = %d + 1 = %d" % (mid, a[mid], m, mid, mid + 1)))
            left = mid + 1              # cerca nella metà destra
        else:
            if trace:
                print("%10d %5d %5d %4d  %s" % (iteration, left, right, mid, "a[%d] = %d > %d => right = %d - 1 = %d" % (mid, a[mid], m, mid, mid - 1)))
            right = mid - 1             # cerca nella metà sinistra
    if trace:
        print("%10s %5d %5d %4s  %s" % ("-", left, right, "-", "left > right => Non trovato! Restituisce -1"))
    return -1                           # elemento non trovato
```

### Esecuzione sugli esempi del capitolo

```python
def run(a, m):
    n = len(a)
    print("Ricerca: m = %d, a = %s, n = %d" % (m, a, n))
    print("%10s %5s %5s %4s  %s" % ("iterazione", "left", "right", "mid", "azione"))
    result = binary_search(a, n, m, trace=True)
    print("Restituisce: %d" % result)
    print()


a = [3, 9, 11, 12, 12, 16, 19]
run(a, 12)
run(a, 10)
run(a, 3)
run(a, 19)
run([1, 2, 2, 2, 2, 2, 3], 2)    # m compare più volte: restituisce una sua occorrenza
```

### Prova tu

Cambia l'input nella cella qui sotto e riesegui la cella. Il primo numero è il valore cercato $m$, gli altri formano l'array (che per la ricerca binaria deve essere ordinato in ordine non decrescente).

```python
run([2, 5, 8, 11, 14, 17, 20], 11)
```

### Confronto con la ricerca lineare

Per $a = [1, 2, 2, 2, 2, 2, 3]$ e $m = 2$ la ricerca lineare restituisce $1$ (la prima occorrenza), la ricerca binaria restituisce $3$: è l'indice dello pseudocodice, $4$, meno 1. Su un array ordinato di $n$ elementi la ricerca lineare esegue fino a $n$ confronti, la ricerca binaria al più $\lfloor \log_2 n \rfloor + 1$ iterazioni.
