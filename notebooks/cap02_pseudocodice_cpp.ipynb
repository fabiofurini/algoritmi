## FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il valore massimo dell'array.

La variabile `max` parte dal primo elemento e viene aggiornata ogni volta che si incontra un elemento più grande.

```python
%%writefile find_maximum.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Restituisce il massimo di a[0..n-1], con n >= 1
int FindMaximum(int n, int *a, bool trace = false)
{
    int max = a[0];                 // parte dal primo elemento
    if (trace)
        printf("%3s  %4s  %4d  %s\n", "--", "--", max, "--");
    for (int i = 1; i < n; i++)     // i = 2..n nello pseudocodice
    {
        bool bigger = a[i] > max;   // confronto, salvato per la traccia
        if (bigger)
        {
            max = a[i];             // aggiorna il massimo
        }
        if (trace)
            printf("%3d  %4d  %4d  %s\n", i, a[i], max, bigger ? "Sì" : "No");
    }
    return max;
}

void Run(string title, vector<int> v)
{
    int n = v.size();
    printf("%s: a = %s, n = %d\n", title.c_str(), Str(n, v.data()).c_str(), n);
    printf("%3s  %4s  %4s  %s\n", "i", "a[i]", "max", "a[i]>max?");
    int result = FindMaximum(n, v.data(), true);
    printf("Restituisce: max = %d\n\n", result);
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // «Prova tu»: l'array arriva dalla riga di comando
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        Run("Prova tu", v);
        return 0;
    }
    Run("Istanza 1", {9, 12, 3, 16, 19, 11, 12});
    Run("Istanza 2", {3, 7, 2, 9, 1});
    return 0;
}
```

## CountPositive

**Input:** un array $a[1..n]$ di $n$ numeri interi. **Output:** il numero di elementi positivi dell'array.

```python
%%writefile count_positive.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Restituisce il numero di elementi positivi di a[0..n-1]
int CountPositive(int n, int *a, bool trace = false)
{
    int count = 0;
    if (trace)
        printf("%3s  %4s  %5d  %s\n", "--", "--", count, "--");
    for (int i = 0; i < n; i++)     // i = 1..n nello pseudocodice
    {
        bool positive = a[i] > 0;   // confronto, salvato per la traccia
        if (positive)
        {
            count = count + 1;
        }
        if (trace)
            printf("%3d  %4d  %5d  %s\n", i, a[i], count, positive ? "Sì" : "No");
    }
    return count;
}

void Run(string title, vector<int> v)
{
    int n = v.size();
    printf("%s: a = %s, n = %d\n", title.c_str(), Str(n, v.data()).c_str(), n);
    printf("%3s  %4s  %5s  %s\n", "i", "a[i]", "count", "a[i]>0?");
    int result = CountPositive(n, v.data(), true);
    printf("Restituisce: count = %d\n\n", result);
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // «Prova tu»: l'array arriva dalla riga di comando
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        Run("Prova tu", v);
        return 0;
    }
    Run("Istanza 1", {-2, 5, -1, 3, 0});
    Run("Istanza 2", {4, -3, 7, 2});
    return 0;
}
```

## IsSorted

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** `true` se i valori dell'array sono in ordine non decrescente, `false` altrimenti.

L'istruzione `return` dentro il ciclo termina la funzione appena trova due elementi consecutivi nell'ordine sbagliato.

```python
%%writefile is_sorted.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Restituisce true se a[0..n-1] è in ordine non decrescente
bool IsSorted(int n, int *a, bool trace = false)
{
    for (int i = 0; i < n - 1; i++)     // i = 1..n-1 nello pseudocodice
    {
        bool wrong = a[i] > a[i + 1];   // confronto, salvato per la traccia
        if (trace)
            printf("%3d  %4d  %6d  %s\n", i, a[i], a[i + 1], wrong ? "Sì" : "No");
        if (wrong)
        {
            return false;
        }
    }
    return true;
}

void Run(string title, vector<int> v)
{
    int n = v.size();
    printf("%s: a = %s, n = %d\n", title.c_str(), Str(n, v.data()).c_str(), n);
    printf("%3s  %4s  %6s  %s\n", "i", "a[i]", "a[i+1]", "a[i]>a[i+1]?");
    bool result = IsSorted(n, v.data(), true);
    printf("Restituisce: %s\n\n", result ? "true" : "false");
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // «Prova tu»: l'array arriva dalla riga di comando
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        Run("Prova tu", v);
        return 0;
    }
    Run("Istanza 1", {2, 5, 7, 9});
    Run("Istanza 2", {3, 1, 4, 2});
    return 0;
}
```

## ScalarProduct

**Input:** due array $a[1..n]$ e $b[1..n]$ di dimensione $n \ge 1$. **Output:** il prodotto scalare $\sum_{i=1}^{n} a[i] \cdot b[i]$ dei due array.

```python
%%writefile scalar_product.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Restituisce il prodotto scalare di a[0..n-1] e b[0..n-1], con n >= 1
int ScalarProduct(int n, int *a, int *b, bool trace = false)
{
    int prod = a[0] * b[0];
    if (trace)
        printf("%3s  %4s  %4s  %9s  %5d\n", "--", "--", "--", "--", prod);
    for (int i = 1; i < n; i++)     // i = 2..n nello pseudocodice
    {
        prod = prod + a[i] * b[i];
        if (trace)
            printf("%3d  %4d  %4d  %9d  %5d\n", i, a[i], b[i], a[i] * b[i], prod);
    }
    return prod;
}

void Run(string title, vector<int> a, vector<int> b)
{
    int n = a.size();
    printf("%s: a = %s, b = %s, n = %d\n", title.c_str(), Str(n, a.data()).c_str(), Str(n, b.data()).c_str(), n);
    printf("%3s  %4s  %4s  %9s  %5s\n", "i", "a[i]", "b[i]", "a[i]*b[i]", "prod");
    int result = ScalarProduct(n, a.data(), b.data(), true);
    printf("Restituisce: prod = %d\n\n", result);
}

int main(int argc, char *argv[])
{
    if (argc > 2)   // «Prova tu»: i numeri arrivano dalla riga di comando (metà per a, metà per b)
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        int n = v.size() / 2;
        Run("Prova tu", vector<int>(v.begin(), v.begin() + n), vector<int>(v.begin() + n, v.begin() + 2 * n));
        return 0;
    }
    Run("Istanza 1", {1, 2, 3}, {4, 5, 6});
    Run("Istanza 2", {2, 0, -1}, {3, 4, 5});
    return 0;
}
```

## ReverseArray

**Input:** un array $a[1..n]$ di $n$ numeri. **Output:** lo stesso array con gli elementi in ordine inverso.

L'elemento $a[i]$ viene scambiato con l'elemento simmetrico $a[n-i+1]$; bastano $\lfloor n/2 \rfloor$ scambi (se $n$ è dispari l'elemento centrale resta al suo posto).

```python
%%writefile reverse_array.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Inverte l'ordine degli elementi di a[0..n-1] (sul posto)
void ReverseArray(int n, int *a, bool trace = false)
{
    if (trace)
        printf("%3s  %-14s  %s\n", "--", "--", Str(n, a).c_str());
    for (int i = 0; i < n / 2; i++)     // i = 1..⌊n/2⌋ nello pseudocodice
    {
        int tmp = a[i];
        a[i] = a[n - i - 1];            // a[n-i+1] nello pseudocodice
        a[n - i - 1] = tmp;
        if (trace)
        {
            string ex = "a[" + to_string(i) + "] <-> a[" + to_string(n - i - 1) + "]";
            printf("%3d  %-14s  %s\n", i, ex.c_str(), Str(n, a).c_str());
        }
    }
}

void Run(string title, vector<int> v)
{
    int n = v.size();
    printf("%s: a = %s, n = %d, %d iterazioni\n", title.c_str(), Str(n, v.data()).c_str(), n, n / 2);
    printf("%3s  %-14s  %s\n", "i", "scambio", "array a (dopo lo scambio)");
    ReverseArray(n, v.data(), true);
    printf("Restituisce: a = %s\n\n", Str(n, v.data()).c_str());
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // «Prova tu»: l'array arriva dalla riga di comando
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        Run("Prova tu", v);
        return 0;
    }
    Run("Istanza 1", {1, 2, 3, 4, 5});
    Run("Istanza 2", {10, 20, 30, 40});
    return 0;
}
```
