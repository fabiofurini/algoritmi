## 1. La dimensione dell'input degli esempi

**Input:** un numero naturale, due numeri naturali o un array di numeri naturali. **Output:** la dimensione dell'input $|I|$, cioè il numero di bit necessari per scriverlo in binario ($\lfloor \log_2 n \rfloor + 1$ bit per un numero $n \ge 1$).

Si veda: sezioni 11.1, 11.3, 11.4, 11.5 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/11-dimensione-input/).

```python
%%writefile dimensione_input_esempi.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Riga | Esecuzioni\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Passi totali S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// Numero di bit di un naturale n >= 1: floor(log2 n) + 1
int Bits(int n)
{
    int k = 0;
    while (n > 0)
    {
        n = n / 2;        // si dimezza fino ad arrivare a 0
        k = k + 1;
    }
    return k;
}

// Cifre di n in base 2 (come stringa)
string Binary(int n)
{
    string s = "";
    while (n > 0)
    {
        s = to_string(n % 2) + s;
        n = n / 2;
    }
    return s;
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Esempio delle dispense: il numero di bit di alcune istanze
    int n = 37, m = 200;
    printf("Un numero: n = %d = (%s)_2, |I| = %d\n", n, Binary(n).c_str(), Bits(n));
    printf("Due numeri: n = %d, m = %d = (%s)_2, |I| = %d + %d = %d\n", n, m, Binary(m).c_str(), Bits(n), Bits(m), Bits(n) + Bits(m));
    int a[] = {6, 1, 3, 8};
    int b[4], sum = 0, M = 0;
    for (int i = 0; i < 4; i++)
    {
        b[i] = Bits(a[i]);
        sum += b[i];
        if (a[i] > M) M = a[i];
    }
    printf("Un array: a = %s, bit per voce = %s, |I| = %d\n", Fmt(a, 0, 4).c_str(), Fmt(b, 0, 4).c_str(), sum);

    // I limiti delle dispense (valori arrotondati a due decimali)
    printf("\n");
    printf("Un numero: log2(n) = %.2f <= |I| = %d <= 2 log2(n) = %.2f\n", log2(n), Bits(n), 2 * log2(n));
    double s = log2(n) + log2(m);
    printf("Due numeri: log2(n) + log2(m) = %.2f <= |I| = %d <= 2 (log2(n) + log2(m)) = %.2f\n", s, Bits(n) + Bits(m), 2 * s);
    int mx = n > m ? n : m;
    printf("Due numeri e massimo: log2(max) = %.2f <= |I| = %d <= 4 log2(max) = %.2f\n", log2(mx), Bits(n) + Bits(m), 4 * log2(mx));
    printf("Un array: n = %d <= |I| = %d <= n (1 + log2 M) = %.2f\n", 4, sum, 4 * (1 + log2(M)));

    // Il limite superiore non risulta stretto: a[1] = M = 2^n, a[2..n] = 1 ha |I| = 2n, ma n(1 + log2 M) = n(n + 1)
    printf("\n");
    printf("   n |      M |  |I| = 2n | n (1 + log2 M)\n");
    for (int k = 1; k <= 10; k++)
    {
        int big = 1 << k;
        int size = Bits(big) + (k - 1);     // a[1] = M, a[2..n] = 1 (1 bit each)
        printf("%4d | %6d | %9d | %13d\n", k, big, size, k * (k + 1));
    }
    return 0;
}
#endif
```

## 2. Somma delle componenti di un array: SumArray

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri naturali. **Output:** $a[1]+a[2]+\cdots+a[n]$. Numero di passi: $S(n) = 3n + 4$, qualunque siano i valori.

Pseudocodice: sezione 11.6 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/11-dimensione-input/).

```python
%%writefile sum_array.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Riga | Esecuzioni\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Passi totali S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// SumArray(a, n)
int SumArray(int a[], int n)
{
    int s = 0; cnt[2]++;
    int i = 0; cnt[3]++;               // i := 1 nello pseudocodice
    cnt[4]++;                          // la condizione del while si conta a ogni valutazione
    while (i < n)                      // i <= n nello pseudocodice
    {
        s = s + a[i]; cnt[5]++;
        i = i + 1; cnt[6]++;
        cnt[4]++;
    }
    cnt[8]++;
    return s;
}

#ifndef SOLO_LIBRERIA
int main()
{
    int a[] = {6, 1, 3, 8};
    int n = 4;
    Reset();
    int s = SumArray(a, n);
    printf("a = %s, n = %d, somma = %d\n", Fmt(a, 0, n).c_str(), n, s);
    PrintSteps({2, 3, 4, 5, 6, 8});
    printf("Formula: S(n) = 3n + 4 = %d\n", 3 * n + 4);

    // Caso peggiore: l'array di |I| uni ha dimensione dell'input |I| e richiede esattamente 3|I| + 4 passi
    printf("\n");
    printf("Array di |I| uni:\n");
    int sizes[] = {4, 8, 16, 32};
    for (int k = 0; k < 4; k++)
    {
        int size = sizes[k];
        int ones[32];
        for (int i = 0; i < size; i++) ones[i] = 1;
        Reset();
        SumArray(ones, size);
        printf("|I| = %2d: S = %3lld = 3|I| + 4 = %d\n", size, Total(), 3 * size + 4);
    }
    return 0;
}
#endif
```

```python
%%writefile prova_sum_array.cpp
#define SOLO_LIBRERIA   // non eseguire il main di sum_array.cpp
#include "sum_array.cpp"

int main()
{
    int a[] = {5, 1, 9, 2, 7};      // <-- cambia l'array
    int n = 5;
    Reset();
    int s = SumArray(a, n);
    printf("somma = %d, S = %lld, 3n + 4 = %d\n", s, Total(), 3 * n + 4);
    return 0;
}
```
