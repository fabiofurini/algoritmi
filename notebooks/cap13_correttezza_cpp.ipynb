## 1. Ricerca del massimo: FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il massimo di $a$. **Invariante:** ogni volta che si valuta la condizione, $max = \max\{a[1],\dots,a[i-1]\}$.

Pseudocodice: sezione 13.2 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/13-correttezza/).

```python
%%writefile find_maximum.cpp
#include <algorithm>
#include <cstdio>
#include <string>
using namespace std;

const int MAXN = 64;

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

bool Equal(const int* x, const int* y, int len)     // x[0..len-1] == y[0..len-1]
{
    for (int k = 0; k < len; k++)
        if (x[k] != y[k]) return false;
    return true;
}

int MaxOf(const int* a, int from, int to)           // massimo di a[from..to-1]
{
    int m = a[from];
    for (int k = from + 1; k < to; k++)
        if (a[k] > m) m = a[k];
    return m;
}

// Invariante: max = massimo di a[1..i-1]
void PrintInvariant(const int* a, int i, int max)
{
    int m = MaxOf(a, 0, i);                         // a[1..i-1] dello pseudocodice
    printf("%2d | %3d | %-26s | %14d | %s\n", i + 1, max, Fmt(a, 0, i).c_str(), m, max == m ? "ok" : "FALSA");
}

// FindMaximum(a, n)
int FindMaximum(int a[], int n)
{
    int max = a[0];
    int i = 1;                                      // i := 2
    PrintInvariant(a, i, max);                      // prima della prima valutazione della condizione
    while (i < n)                                   // i <= n
    {
        if (a[i] > max)
        {
            max = a[i];
        }
        i = i + 1;
        PrintInvariant(a, i, max);                  // dopo ogni iterazione, cioe' prima della valutazione successiva
    }
    return max;
}

#ifndef SOLO_LIBRERIA
int main()
{
    int a[] = {9, 12, 3, 16, 19, 11, 12};           // esempio delle dispense
    int n = 7;
    printf("a = %s, n = %d\n", Fmt(a, 0, n).c_str(), n);
    printf(" i | max | a[1..i-1]                  | max{a[1..i-1]} | invariante\n");
    int m = FindMaximum(a, n);
    printf("Il ciclo termina con i = n + 1 = %d: FindMaximum restituisce %d\n", n + 1, m);
    return 0;
}
#endif
```

## 2. Ordinamento per inserimento: InsertionSort

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri naturali. **Output:** l'array ordinato in modo non decrescente. **Invariante esterno:** $a[1..i-1]$ contiene gli elementi originali di $a[1..i-1]$ ordinati e $a[i..n]$ è uguale all'originale. **Invariante interno** (I1)–(I3): $a[1..j] = b[1..j]$, $a[j+2..i] = b[j+1..i-1]$ e questi elementi sono $> key$.

Pseudocodice: sezione 13.3 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/13-correttezza/).

```python
%%writefile insertion_sort.cpp
#include <algorithm>
#include <cstdio>
#include <string>
using namespace std;

const int MAXN = 64;

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

bool Equal(const int* x, const int* y, int len)     // x[0..len-1] == y[0..len-1]
{
    for (int k = 0; k < len; k++)
        if (x[k] != y[k]) return false;
    return true;
}

// L'array con una barra dopo la parte ordinata a[1..i-1]
string Split(const int* a, int n, int i)
{
    string s = Fmt(a, 0, i);
    s.pop_back();
    string t = Fmt(a, i, n);
    return s + " | " + t.substr(1);
}

// Invariante esterno: a[1..i-1] = elementi originali di a[1..i-1] ordinati, a[i..n] = originale
void PrintOuter(const int* a, const int* orig, int n, int i)
{
    int s[MAXN];
    for (int k = 0; k < i; k++) s[k] = orig[k];
    sort(s, s + i);
    bool ok = Equal(a, s, i) && Equal(a + i, orig + i, n - i);
    printf("%2d | %-26s | %-19s | %s\n", i + 1, Split(a, n, i).c_str(), Fmt(orig, 0, i).c_str(), ok ? "ok" : "FALSA");
}

// Invariante interno (I1)-(I3); b = a a inizio iterazione
void PrintInner(const int* a, const int* b, int n, int i, int j, int key)
{
    int ip = i + 1, jp = j + 1;                    // indici dello pseudocodice
    bool ok = Equal(a, b, jp) && Equal(a + jp + 1, b + jp, ip - jp - 1);
    for (int k = jp + 1; k < ip; k++)
        if (a[k] <= key) ok = false;
    string cond = "j = 0: false";
    if (j >= 0)
    {
        bool t = a[j] > key;
        cond = "a[" + to_string(jp) + "] = " + to_string(a[j]) + (t ? " > " : " <= ") + to_string(key) + ": " + (t ? "true" : "false");
    }
    printf("%d | %-15s | %-12s | %-9s | %-20s | %s\n", jp, Fmt(a, 0, n).c_str(), Fmt(a, 0, jp).c_str(), Fmt(a, jp + 1, ip).c_str(), cond.c_str(), ok ? "ok" : "FALSA");
}

// InsertionSort(a, n)
void InsertionSort(int a[], int n, bool showOuter = true, int traceInnerAt = 0)
{
    int orig[MAXN], b[MAXN];
    for (int k = 0; k < n; k++) orig[k] = a[k];    // copia dell'input: serve solo a verificare l'invariante
    int i = 1;                                     // i := 2
    if (showOuter) PrintOuter(a, orig, n, i);
    while (i < n)                                  // i <= n
    {
        for (int k = 0; k < n; k++) b[k] = a[k];   // contenuto di a a inizio iterazione (solo per l'invariante interno)
        int key = a[i];
        int j = i - 1;
        bool tr = (i + 1 == traceInnerAt);
        if (tr) PrintInner(a, b, n, i, j, key);
        while (j >= 0 && a[j] > key)               // j >= 1 and a[j] > key
        {
            a[j + 1] = a[j];
            j = j - 1;
            if (tr) PrintInner(a, b, n, i, j, key);
        }
        a[j + 1] = key;
        i = i + 1;
        if (showOuter) PrintOuter(a, orig, n, i);
    }
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Il ciclo esterno: a[1..5] = (5, 2, 4, 1, 3), esempio delle dispense
    int a[] = {5, 2, 4, 1, 3};
    int n = 5;
    printf("a = %s\n", Fmt(a, 0, n).c_str());
    printf(" i | a (ordinato | resto)       | a[1..i-1] originale | invariante\n");
    InsertionSort(a, n);
    printf("ordinato: %s\n", Fmt(a, 0, n).c_str());

    // Il ciclo interno all'iterazione i = 5, con key = 3 (a = (1, 2, 4, 5, 3) all'inizio)
    printf("\n");
    int c[] = {5, 2, 4, 1, 3};
    printf("j | a               | a[1..j]      | a[j+2..i] | condizione           | invariante\n");
    InsertionSort(c, n, false, 5);
    printf("dopo la riga 10 (a[j+1] = key): a = %s\n", Fmt(c, 0, n).c_str());
    return 0;
}
#endif
```

## 3. Ricerca binaria: BinarySearch

**Input:** un array $a[1..n]$ ordinato in modo non decrescente e un valore $m$. **Output:** un indice $k$ con $a[k] = m$ se $m$ è in $a$, $-1$ altrimenti. **Invariante:** (I1) $1 \le left$ e $right \le n$; (I2) se $m$ è in $a$, allora $m$ è in $a[left..right]$.

Pseudocodice: sezione 13.4 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/13-correttezza/).

```python
%%writefile binary_search.cpp
#include <algorithm>
#include <cstdio>
#include <string>
using namespace std;

const int MAXN = 64;

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

bool Equal(const int* x, const int* y, int len)     // x[0..len-1] == y[0..len-1]
{
    for (int k = 0; k < len; k++)
        if (x[k] != y[k]) return false;
    return true;
}

// Invariante (I1) 1 <= left e right <= n, (I2) se m e' in a allora e' in a[left..right]; mid = -1: intervallo vuoto
void PrintRow(const int* a, int n, int m, int left, int right, int mid)
{
    bool inA = false, inSub = false;
    for (int k = 0; k < n; k++)
    {
        if (a[k] == m)
        {
            inA = true;
            if (k >= left && k <= right) inSub = true;
        }
    }
    bool ok = (left + 1 >= 1 && right + 1 <= n && (!inA || inSub));
    char action[100];
    if (mid < 0)
        snprintf(action, sizeof action, "left > right => return -1");
    else if (a[mid] == m)
        snprintf(action, sizeof action, "mid = %d, a[%d] = %d = %d => return %d", mid + 1, mid + 1, a[mid], m, mid + 1);
    else if (a[mid] < m)
        snprintf(action, sizeof action, "mid = %d, a[%d] = %d < %d => left = %d", mid + 1, mid + 1, a[mid], m, mid + 2);
    else
        snprintf(action, sizeof action, "mid = %d, a[%d] = %d > %d => right = %d", mid + 1, mid + 1, a[mid], m, mid);
    printf("%4d | %5d | %-27s | %-11s | %s\n", left + 1, right + 1, Fmt(a, left, right + 1).c_str(), ok ? "ok" : "FALSA", action);
}

// BinarySearch(a, n, m)
int BinarySearch(int a[], int n, int m, bool trace = false)
{
    int left = 0;                                  // left := 1
    int right = n - 1;                             // right := n
    while (left <= right)
    {
        int mid = left + (right - left) / 2;
        if (trace) PrintRow(a, n, m, left, right, mid);   // riga dell'invariante a questa valutazione
        if (a[mid] == m)
        {
            return mid;
        }
        if (a[mid] < m)
        {
            left = mid + 1;
        }
        else
        {
            right = mid - 1;
        }
    }
    if (trace) PrintRow(a, n, m, left, right, -1);
    return -1;
}

#ifndef SOLO_LIBRERIA
int main()
{
    int a[] = {3, 9, 11, 12, 12, 16, 19};          // esempio delle dispense
    int n = 7;
    printf("a = %s, n = %d\n", Fmt(a, 0, n).c_str(), n);

    int m = 10;
    printf("\n");
    printf("m = %d\n", m);
    printf("left | right | a[left..right]              | invariante  | azione\n");
    int k = BinarySearch(a, n, m, true);
    printf("m = %d non trovato: restituisce %d\n", m, k);

    m = 12;
    printf("\n");
    printf("m = %d\n", m);
    printf("left | right | a[left..right]              | invariante  | azione\n");
    k = BinarySearch(a, n, m, true);
    printf("m = %d trovato: indice %d nel codice (posizione %d nello pseudocodice)\n", m, k, k + 1);

    // Controllo esaustivo per m = 0, ..., 20
    int errors = 0;
    for (m = 0; m <= 20; m++)
    {
        k = BinarySearch(a, n, m);
        bool inA = false;
        for (int t = 0; t < n; t++) if (a[t] == m) inA = true;
        if ((inA && (k < 0 || a[k] != m)) || (!inA && k != -1)) errors++;
    }
    printf("\n");
    printf("Controllo per m = 0, ..., 20: %d errori\n", errors);
    return 0;
}
#endif
```

## 4. Algoritmo di Euclide: EuclideanAlgorithm

**Input:** due numeri naturali $n, m \ge 1$. **Output:** $\gcd(n, m)$. **Invariante:** $n \ge 1$, $m \ge 0$ e $\gcd(n,m) = \gcd(n_0,m_0)$, dove $n_0, m_0$ sono i valori in input.

Pseudocodice: sezione 13.5 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/13-correttezza/).

```python
%%writefile euclidean_algorithm.cpp
#include <algorithm>
#include <cstdio>
#include <string>
using namespace std;

const int MAXN = 64;

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

bool Equal(const int* x, const int* y, int len)     // x[0..len-1] == y[0..len-1]
{
    for (int k = 0; k < len; k++)
        if (x[k] != y[k]) return false;
    return true;
}

int GcdCheck(int n, int m)                         // mcd per enumerazione dei divisori, indipendente dall'algoritmo
{
    if (m == 0) return n;
    int d = n < m ? n : m;
    while (n % d != 0 || m % d != 0) d = d - 1;
    return d;
}

// Invariante: n >= 1, m >= 0 e gcd(n, m) = gcd(n0, m0)
void PrintRow(int n, int m, int g0)
{
    int g = GcdCheck(n, m);
    bool ok = (n >= 1 && m >= 0 && g == g0);
    printf("%4d | %4d | %9d | %s\n", n, m, g, ok ? "ok" : "FALSA");
}

// EuclideanAlgorithm(n, m)
int EuclideanAlgorithm(int n, int m)
{
    int g0 = GcdCheck(n, m);                       // gcd(n0, m0), solo per verificare l'invariante
    PrintRow(n, m, g0);
    while (m != 0)
    {
        int r = n % m;
        n = m;
        m = r;
        PrintRow(n, m, g0);
    }
    return n;
}

#ifndef SOLO_LIBRERIA
int main()
{
    int n = 252, m = 105;      // esempio delle dispense
    printf("(n0, m0) = (%d, %d)\n", n, m);
    printf("   n |    m | gcd(n, m) | invariante\n");
    int g = EuclideanAlgorithm(n, m);
    printf("Il ciclo termina con m = 0: restituisce n = %d\n", g);

    printf("\n");
    n = 48; m = 18;
    printf("(n0, m0) = (%d, %d)\n", n, m);
    printf("   n |    m | gcd(n, m) | invariante\n");
    g = EuclideanAlgorithm(n, m);
    printf("Il ciclo termina con m = 0: restituisce n = %d\n", g);
    return 0;
}
#endif
```
