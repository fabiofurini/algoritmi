## Algoritmo di Euclide: EuclideanAlgorithm

**Input:** due numeri naturali $n, m \ge 1$. **Output:** $\gcd(n, m)$. **Invariante:** $n \ge 1$, $m \ge 0$ e $\gcd(n,m) = \gcd(n_0,m_0)$, dove $n_0, m_0$ sono i valori in input.

Pseudocodice: sezione 13.5 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/13-correttezza/).

```python
%%writefile EuclideanAlgorithm.java
public class EuclideanAlgorithm
{
// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

static boolean equal(int[] x, int xs, int[] y, int ys, int len)     // x[xs..xs+len-1] == y[ys..ys+len-1]
{
    for (int k = 0; k < len; k++)
        if (x[xs + k] != y[ys + k]) return false;
    return true;
}

    static int gcdCheck(int n, int m)              // mcd per enumerazione dei divisori, indipendente dall'algoritmo
    {
        if (m == 0) return n;
        int d = Math.min(n, m);
        while (n % d != 0 || m % d != 0) d = d - 1;
        return d;
    }

    // Invariante: n >= 1, m >= 0 e gcd(n, m) = gcd(n0, m0)
    static void printRow(int n, int m, int g0)
    {
        int g = gcdCheck(n, m);
        boolean ok = (n >= 1 && m >= 0 && g == g0);
        System.out.println(String.format("%4d | %4d | %9d | %s", n, m, g, ok ? "ok" : "FALSA"));
    }

    // EuclideanAlgorithm(n, m)
    static int euclideanAlgorithm(int n, int m)
    {
        int g0 = gcdCheck(n, m);                   // gcd(n0, m0), solo per verificare l'invariante
        printRow(n, m, g0);
        while (m != 0)
        {
            int r = n % m;
            n = m;
            m = r;
            printRow(n, m, g0);
        }
        return n;
    }

    public static void main(String[] args)
    {
        int n = 252, m = 105;      // esempio delle dispense
        System.out.printf("(n0, m0) = (%d, %d)\n", n, m);
        System.out.printf("   n |    m | gcd(n, m) | invariante\n");
        int g = euclideanAlgorithm(n, m);
        System.out.printf("Il ciclo termina con m = 0: restituisce n = %d\n", g);

        System.out.printf("\n");
        n = 48; m = 18;
        System.out.printf("(n0, m0) = (%d, %d)\n", n, m);
        System.out.printf("   n |    m | gcd(n, m) | invariante\n");
        g = euclideanAlgorithm(n, m);
        System.out.printf("Il ciclo termina con m = 0: restituisce n = %d\n", g);
    }
}
```

```python
%%writefile ProvaEuclideanAlgorithm.java
public class ProvaEuclideanAlgorithm
{
    public static void main(String[] args)
    {
        int n = 1071, m = 462;      // <-- cambia i due numeri
        System.out.printf("   n |    m | gcd(n, m) | invariante\n");
        int g = EuclideanAlgorithm.euclideanAlgorithm(n, m);
        System.out.printf("restituisce %d\n", g);
    }
}
```
