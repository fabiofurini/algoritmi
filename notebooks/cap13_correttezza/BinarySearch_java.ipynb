## Ricerca binaria: BinarySearch

**Input:** un array $a[1..n]$ ordinato in modo non decrescente e un valore $m$. **Output:** un indice $k$ con $a[k] = m$ se $m$ è in $a$, $-1$ altrimenti. **Invariante:** (I1) $1 \le left$ e $right \le n$; (I2) se $m$ è in $a$, allora $m$ è in $a[left..right]$.

Pseudocodice: sezione 13.4 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/13-correttezza/).

```python
%%writefile BinarySearch.java
public class BinarySearch
{
// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

static boolean equal(int[] x, int xs, int[] y, int ys, int len)     // x[xs..xs+len-1] == y[ys..ys+len-1]
{
    for (int k = 0; k < len; k++)
        if (x[xs + k] != y[ys + k]) return false;
    return true;
}

    // Invariante (I1) 1 <= left e right <= n, (I2) se m e' in a allora e' in a[left..right]; mid = -1: intervallo vuoto
    static void printRow(int[] a, int m, int left, int right, int mid)
    {
        int n = a.length;
        boolean inA = false, inSub = false;
        for (int k = 0; k < n; k++)
        {
            if (a[k] == m)
            {
                inA = true;
                if (k >= left && k <= right) inSub = true;
            }
        }
        boolean ok = (left + 1 >= 1 && right + 1 <= n && (!inA || inSub));
        String action;
        if (mid < 0)
            action = "left > right => return -1";
        else if (a[mid] == m)
            action = String.format("mid = %d, a[%d] = %d = %d => return %d", mid + 1, mid + 1, a[mid], m, mid + 1);
        else if (a[mid] < m)
            action = String.format("mid = %d, a[%d] = %d < %d => left = %d", mid + 1, mid + 1, a[mid], m, mid + 2);
        else
            action = String.format("mid = %d, a[%d] = %d > %d => right = %d", mid + 1, mid + 1, a[mid], m, mid);
        System.out.println(String.format("%4d | %5d | %-27s | %-11s | %s", left + 1, right + 1, fmt(a, left, right + 1), ok ? "ok" : "FALSA", action));
    }

    // BinarySearch(a, n, m)
    static int binarySearch(int[] a, int m, boolean trace)
    {
        int n = a.length;
        int left = 0;                              // left := 1
        int right = n - 1;                         // right := n
        while (left <= right)
        {
            int mid = left + (right - left) / 2;
            if (trace) printRow(a, m, left, right, mid);   // riga dell'invariante a questa valutazione
            if (a[mid] == m)
            {
                return mid;
            }
            if (a[mid] < m)
            {
                left = mid + 1;
            }
            else
            {
                right = mid - 1;
            }
        }
        if (trace) printRow(a, m, left, right, -1);
        return -1;
    }

    public static void main(String[] args)
    {
        int[] a = {3, 9, 11, 12, 12, 16, 19};      // esempio delle dispense
        int n = a.length;
        System.out.printf("a = %s, n = %d\n", fmt(a, 0, n), n);

        int m = 10;
        System.out.printf("\n");
        System.out.printf("m = %d\n", m);
        System.out.printf("left | right | a[left..right]              | invariante  | azione\n");
        int k = binarySearch(a, m, true);
        System.out.printf("m = %d non trovato: restituisce %d\n", m, k);

        m = 12;
        System.out.printf("\n");
        System.out.printf("m = %d\n", m);
        System.out.printf("left | right | a[left..right]              | invariante  | azione\n");
        k = binarySearch(a, m, true);
        System.out.printf("m = %d trovato: indice %d nel codice (posizione %d nello pseudocodice)\n", m, k, k + 1);

        // Controllo esaustivo per m = 0, ..., 20
        int errors = 0;
        for (m = 0; m <= 20; m++)
        {
            k = binarySearch(a, m, false);
            boolean inA = false;
            for (int t = 0; t < n; t++) if (a[t] == m) inA = true;
            if ((inA && (k < 0 || a[k] != m)) || (!inA && k != -1)) errors++;
        }
        System.out.printf("\n");
        System.out.printf("Controllo per m = 0, ..., 20: %d errori\n", errors);
    }
}
```
