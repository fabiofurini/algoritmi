## FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il valore massimo dell'array.

La variabile `max` parte dal primo elemento e viene aggiornata ogni volta che si incontra un elemento più grande.

```python
%%writefile FindMaximum.java
import java.util.Arrays;

public class FindMaximum {

    // Restituisce il massimo di a[0..n-1], con n >= 1
    static int findMaximum(int[] a, int n, boolean trace) {
        int max = a[0];                 // parte dal primo elemento
        if (trace)
            System.out.printf("%3s  %4s  %4d  %s\n", "--", "--", max, "--");
        for (int i = 1; i < n; i++) {   // i = 2..n nello pseudocodice
            boolean bigger = a[i] > max;    // confronto, salvato per la traccia
            if (bigger) {
                max = a[i];             // aggiorna il massimo
            }
            if (trace)
                System.out.printf("%3d  %4d  %4d  %s\n", i, a[i], max, bigger ? "Sì" : "No");
        }
        return max;
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s  %4s  %4s  %s\n", "i", "a[i]", "max", "a[i]>max?");
        int result = findMaximum(a, n, true);
        System.out.printf("Restituisce: max = %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // «Prova tu»: l'array arriva dalla riga di comando
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Prova tu", v);
            return;
        }
        run("Istanza 1", new int[]{9, 12, 3, 16, 19, 11, 12});
        run("Istanza 2", new int[]{3, 7, 2, 9, 1});
    }
}
```

## CountPositive

**Input:** un array $a[1..n]$ di $n$ numeri interi. **Output:** il numero di elementi positivi dell'array.

```python
%%writefile CountPositive.java
import java.util.Arrays;

public class CountPositive {

    // Restituisce il numero di elementi positivi di a[0..n-1]
    static int countPositive(int[] a, int n, boolean trace) {
        int count = 0;
        if (trace)
            System.out.printf("%3s  %4s  %5d  %s\n", "--", "--", count, "--");
        for (int i = 0; i < n; i++) {   // i = 1..n nello pseudocodice
            boolean positive = a[i] > 0;    // confronto, salvato per la traccia
            if (positive) {
                count = count + 1;
            }
            if (trace)
                System.out.printf("%3d  %4d  %5d  %s\n", i, a[i], count, positive ? "Sì" : "No");
        }
        return count;
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s  %4s  %5s  %s\n", "i", "a[i]", "count", "a[i]>0?");
        int result = countPositive(a, n, true);
        System.out.printf("Restituisce: count = %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // «Prova tu»: l'array arriva dalla riga di comando
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Prova tu", v);
            return;
        }
        run("Istanza 1", new int[]{-2, 5, -1, 3, 0});
        run("Istanza 2", new int[]{4, -3, 7, 2});
    }
}
```

## IsSorted

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** `true` se i valori dell'array sono in ordine non decrescente, `false` altrimenti.

L'istruzione `return` dentro il ciclo termina la funzione appena trova due elementi consecutivi nell'ordine sbagliato.

```python
%%writefile IsSorted.java
import java.util.Arrays;

public class IsSorted {

    // Restituisce true se a[0..n-1] è in ordine non decrescente
    static boolean isSorted(int[] a, int n, boolean trace) {
        for (int i = 0; i < n - 1; i++) {   // i = 1..n-1 nello pseudocodice
            boolean wrong = a[i] > a[i + 1];    // confronto, salvato per la traccia
            if (trace)
                System.out.printf("%3d  %4d  %6d  %s\n", i, a[i], a[i + 1], wrong ? "Sì" : "No");
            if (wrong) {
                return false;
            }
        }
        return true;
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s  %4s  %6s  %s\n", "i", "a[i]", "a[i+1]", "a[i]>a[i+1]?");
        boolean result = isSorted(a, n, true);
        System.out.printf("Restituisce: %s\n\n", result ? "true" : "false");
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // «Prova tu»: l'array arriva dalla riga di comando
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Prova tu", v);
            return;
        }
        run("Istanza 1", new int[]{2, 5, 7, 9});
        run("Istanza 2", new int[]{3, 1, 4, 2});
    }
}
```

## ScalarProduct

**Input:** due array $a[1..n]$ e $b[1..n]$ di dimensione $n \ge 1$. **Output:** il prodotto scalare $\sum_{i=1}^{n} a[i] \cdot b[i]$ dei due array.

```python
%%writefile ScalarProduct.java
import java.util.Arrays;

public class ScalarProduct {

    // Restituisce il prodotto scalare di a[0..n-1] e b[0..n-1], con n >= 1
    static int scalarProduct(int[] a, int[] b, int n, boolean trace) {
        int prod = a[0] * b[0];
        if (trace)
            System.out.printf("%3s  %4s  %4s  %9s  %5d\n", "--", "--", "--", "--", prod);
        for (int i = 1; i < n; i++) {   // i = 2..n nello pseudocodice
            prod = prod + a[i] * b[i];
            if (trace)
                System.out.printf("%3d  %4d  %4d  %9d  %5d\n", i, a[i], b[i], a[i] * b[i], prod);
        }
        return prod;
    }

    static void run(String title, int[] a, int[] b) {
        int n = a.length;
        System.out.printf("%s: a = %s, b = %s, n = %d\n", title, Arrays.toString(a), Arrays.toString(b), n);
        System.out.printf("%3s  %4s  %4s  %9s  %5s\n", "i", "a[i]", "b[i]", "a[i]*b[i]", "prod");
        int result = scalarProduct(a, b, n, true);
        System.out.printf("Restituisce: prod = %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 1) {          // «Prova tu»: i numeri arrivano dalla riga di comando (metà per a, metà per b)
            int n = args.length / 2;
            int[] a = new int[n], b = new int[n];
            for (int k = 0; k < n; k++) {
                a[k] = Integer.parseInt(args[k]);
                b[k] = Integer.parseInt(args[n + k]);
            }
            run("Prova tu", a, b);
            return;
        }
        run("Istanza 1", new int[]{1, 2, 3}, new int[]{4, 5, 6});
        run("Istanza 2", new int[]{2, 0, -1}, new int[]{3, 4, 5});
    }
}
```

## ReverseArray

**Input:** un array $a[1..n]$ di $n$ numeri. **Output:** lo stesso array con gli elementi in ordine inverso.

L'elemento $a[i]$ viene scambiato con l'elemento simmetrico $a[n-i+1]$; bastano $\lfloor n/2 \rfloor$ scambi (se $n$ è dispari l'elemento centrale resta al suo posto).

```python
%%writefile ReverseArray.java
import java.util.Arrays;

public class ReverseArray {

    // Inverte l'ordine degli elementi di a[0..n-1] (sul posto)
    static void reverseArray(int[] a, int n, boolean trace) {
        if (trace)
            System.out.printf("%3s  %-14s  %s\n", "--", "--", Arrays.toString(a));
        for (int i = 0; i < n / 2; i++) {   // i = 1..⌊n/2⌋ nello pseudocodice
            int tmp = a[i];
            a[i] = a[n - i - 1];            // a[n-i+1] nello pseudocodice
            a[n - i - 1] = tmp;
            if (trace)
                System.out.printf("%3d  %-14s  %s\n", i, "a[" + i + "] <-> a[" + (n - i - 1) + "]", Arrays.toString(a));
        }
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d, %d iterazioni\n", title, Arrays.toString(a), n, n / 2);
        System.out.printf("%3s  %-14s  %s\n", "i", "scambio", "array a (dopo lo scambio)");
        reverseArray(a, n, true);
        System.out.printf("Restituisce: a = %s\n\n", Arrays.toString(a));
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // «Prova tu»: l'array arriva dalla riga di comando
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Prova tu", v);
            return;
        }
        run("Istanza 1", new int[]{1, 2, 3, 4, 5});
        run("Istanza 2", new int[]{10, 20, 30, 40});
    }
}
```
