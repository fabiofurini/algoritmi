## FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il valore massimo dell'array.

La variabile `max` parte dal primo elemento e viene aggiornata ogni volta che si incontra un elemento più grande.

```python
def find_maximum(a, n, trace=False):
    max_val = a[0]                  # parte dal primo elemento
    if trace:
        print("%3s  %4s  %4d  %s" % ("--", "--", max_val, "--"))
    for i in range(1, n):           # i = 2..n nello pseudocodice
        bigger = a[i] > max_val     # confronto, salvato per la traccia
        if bigger:
            max_val = a[i]          # aggiorna il massimo
        if trace:
            print("%3d  %4d  %4d  %s" % (i, a[i], max_val, "Sì" if bigger else "No"))
    return max_val
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s  %4s  %4s  %s" % ("i", "a[i]", "max", "a[i]>max?"))
    result = find_maximum(a, n, trace=True)
    print("Restituisce: max = %d" % result)
    print()


run("Istanza 1", [9, 12, 3, 16, 19, 11, 12])
run("Istanza 2", [3, 7, 2, 9, 1])
```
