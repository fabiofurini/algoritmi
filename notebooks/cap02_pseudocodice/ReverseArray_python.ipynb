## ReverseArray

**Input:** un array $a[1..n]$ di $n$ numeri. **Output:** lo stesso array con gli elementi in ordine inverso.

L'elemento $a[i]$ viene scambiato con l'elemento simmetrico $a[n-i+1]$; bastano $\lfloor n/2 \rfloor$ scambi (se $n$ è dispari l'elemento centrale resta al suo posto).

```python
def reverse_array(a, n, trace=False):
    if trace:
        print("%3s  %-14s  %s" % ("--", "--", a))
    for i in range(n // 2):         # i = 1..⌊n/2⌋ nello pseudocodice
        tmp = a[i]
        a[i] = a[n - i - 1]         # a[n-i+1] nello pseudocodice
        a[n - i - 1] = tmp
        if trace:
            print("%3d  %-14s  %s" % (i, "a[%d] <-> a[%d]" % (i, n - i - 1), a))
    return a
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d, %d iterazioni" % (title, a, n, n // 2))
    print("%3s  %-14s  %s" % ("i", "scambio", "array a (dopo lo scambio)"))
    result = reverse_array(a, n, trace=True)
    print("Restituisce: a = %s" % result)
    print()


run("Istanza 1", [1, 2, 3, 4, 5])
run("Istanza 2", [10, 20, 30, 40])
```
