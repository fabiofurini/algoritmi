**Indici.** Nello pseudocodice gli array partono da 1, qui da 0: l'elemento $a[i]$ dello pseudocodice è `a[i-1]` nel codice. Le tracce stampate usano gli indici del codice, quindi sono quelle dello pseudocodice con tutti gli indici diminuiti di 1. I nomi delle funzioni sono quelli dello pseudocodice, nella convenzione di Python (`snake_case`: `find_maximum` per `FindMaximum`).

## CountPositive

**Input:** un array $a[1..n]$ di $n$ numeri interi. **Output:** il numero di elementi positivi dell'array.

```python
def count_positive(a, n, trace=False):
    count = 0
    if trace:
        print("%3s  %4s  %5s  %s" % ("--", "--", count, "--"))
    for i in range(n):              # i = 1..n nello pseudocodice
        positive = a[i] > 0         # confronto, salvato per la traccia
        if positive:
            count = count + 1
        if trace:
            print("%3d  %4d  %5d  %s" % (i, a[i], count, "Sì" if positive else "No"))
    return count
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s  %4s  %5s  %s" % ("i", "a[i]", "count", "a[i]>0?"))
    result = count_positive(a, n, trace=True)
    print("Restituisce: count = %d" % result)
    print()


run("Istanza 1", [-2, 5, -1, 3, 0])
run("Istanza 2", [4, -3, 7, 2])
```
