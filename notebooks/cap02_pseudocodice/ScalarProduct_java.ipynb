## ScalarProduct

**Input:** due array $a[1..n]$ e $b[1..n]$ di dimensione $n \ge 1$. **Output:** il prodotto scalare $\sum_{i=1}^{n} a[i] \cdot b[i]$ dei due array.

```python
%%writefile ScalarProduct.java
import java.util.Arrays;

public class ScalarProduct {

    // Restituisce il prodotto scalare di a[0..n-1] e b[0..n-1], con n >= 1
    static int scalarProduct(int[] a, int[] b, int n, boolean trace) {
        int prod = a[0] * b[0];
        if (trace)
            System.out.printf("%3s  %4s  %4s  %9s  %5d\n", "--", "--", "--", "--", prod);
        for (int i = 1; i < n; i++) {   // i = 2..n nello pseudocodice
            prod = prod + a[i] * b[i];
            if (trace)
                System.out.printf("%3d  %4d  %4d  %9d  %5d\n", i, a[i], b[i], a[i] * b[i], prod);
        }
        return prod;
    }

    static void run(String title, int[] a, int[] b) {
        int n = a.length;
        System.out.printf("%s: a = %s, b = %s, n = %d\n", title, Arrays.toString(a), Arrays.toString(b), n);
        System.out.printf("%3s  %4s  %4s  %9s  %5s\n", "i", "a[i]", "b[i]", "a[i]*b[i]", "prod");
        int result = scalarProduct(a, b, n, true);
        System.out.printf("Restituisce: prod = %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 1) {          // «Prova tu»: i numeri arrivano dalla riga di comando (metà per a, metà per b)
            int n = args.length / 2;
            int[] a = new int[n], b = new int[n];
            for (int k = 0; k < n; k++) {
                a[k] = Integer.parseInt(args[k]);
                b[k] = Integer.parseInt(args[n + k]);
            }
            run("Prova tu", a, b);
            return;
        }
        run("Istanza 1", new int[]{1, 2, 3}, new int[]{4, 5, 6});
        run("Istanza 2", new int[]{2, 0, -1}, new int[]{3, 4, 5});
    }
}
```
