## IsSorted

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** `true` se i valori dell'array sono in ordine non decrescente, `false` altrimenti.

L'istruzione `return` dentro il ciclo termina la funzione appena trova due elementi consecutivi nell'ordine sbagliato.

```python
def is_sorted(a, n, trace=False):
    for i in range(n - 1):          # i = 1..n-1 nello pseudocodice
        wrong = a[i] > a[i + 1]     # confronto, salvato per la traccia
        if trace:
            print("%3d  %4d  %6d  %s" % (i, a[i], a[i + 1], "Sì" if wrong else "No"))
        if wrong:
            return False
    return True
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s  %4s  %6s  %s" % ("i", "a[i]", "a[i+1]", "a[i]>a[i+1]?"))
    result = is_sorted(a, n, trace=True)
    print("Restituisce: %s" % str(result).lower())
    print()


run("Istanza 1", [2, 5, 7, 9])
run("Istanza 2", [3, 1, 4, 2])
```
