## CountPositive

**Input:** un array $a[1..n]$ di $n$ numeri interi. **Output:** il numero di elementi positivi dell'array.

```python
%%writefile CountPositive.java
import java.util.Arrays;

public class CountPositive {

    // Restituisce il numero di elementi positivi di a[0..n-1]
    static int countPositive(int[] a, int n, boolean trace) {
        int count = 0;
        if (trace)
            System.out.printf("%3s  %4s  %5d  %s\n", "--", "--", count, "--");
        for (int i = 0; i < n; i++) {   // i = 1..n nello pseudocodice
            boolean positive = a[i] > 0;    // confronto, salvato per la traccia
            if (positive) {
                count = count + 1;
            }
            if (trace)
                System.out.printf("%3d  %4d  %5d  %s\n", i, a[i], count, positive ? "Sì" : "No");
        }
        return count;
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s  %4s  %5s  %s\n", "i", "a[i]", "count", "a[i]>0?");
        int result = countPositive(a, n, true);
        System.out.printf("Restituisce: count = %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // «Prova tu»: l'array arriva dalla riga di comando
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Prova tu", v);
            return;
        }
        run("Istanza 1", new int[]{-2, 5, -1, 3, 0});
        run("Istanza 2", new int[]{4, -3, 7, 2});
    }
}
```
