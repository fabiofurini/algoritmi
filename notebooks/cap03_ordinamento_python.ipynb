## InsertionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in ordine non decrescente.

Al passo $i$ il sotto-array $a[1..i-1]$ è ordinato: la chiave $a[i]$ viene inserita al suo posto, dopo aver spostato di una posizione a destra le chiavi maggiori. La funzione restituisce anche i contatori usati nella dispensa: il numero di confronti tra chiavi e il numero di spostamenti (nessuno scambio: uno spostamento è una sola assegnazione).

```python
def row(i, j, key, a, action):          # una riga della traccia
    print("%3s %3s %4s  %-20s  %s" % (i, j, key, a, action))


def insertion_sort(a, n, trace=False):
    comparisons = 0                     # confronti tra chiavi
    shifts = 0                          # spostamenti
    if trace:
        row("-", "-", "-", a, "stato iniziale")
    i = 1                               # i = 2 nello pseudocodice
    while i < n:                        # i <= n nello pseudocodice
        key = a[i]                      # elemento da inserire
        j = i - 1
        if trace:
            row(i, j, key, a, "key <- a[%d]=%d; j <- %d" % (i, key, j))
        while j >= 0 and a[j] > key:    # j >= 1 nello pseudocodice
            comparisons += 1
            shifts += 1
            a[j + 1] = a[j]             # sposta l'elemento a destra
            j = j - 1
            if trace:
                row("", j, key, a, "a[%d]=%d>%d: sposta a[%d] <- %d; j <- %d" % (j + 1, a[j + 1], key, j + 2, a[j + 1], j))
        if j >= 0:
            comparisons += 1            # il confronto che ha fermato il ciclo (a[j] <= key)
        a[j + 1] = key                  # inserisce key nella sua posizione
        if trace:
            if j < 0:
                row("", "-", key, a, "j=%d: esce; inserisce a[%d] <- %d" % (j, j + 1, key))
            else:
                row("", "-", key, a, "a[%d]=%d<=%d: esce; inserisce a[%d] <- %d" % (j, a[j], key, j + 1, key))
        i = i + 1
    return comparisons, shifts
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s %3s %4s  %-20s  %s" % ("i", "j", "key", "a", "azione"))
    comparisons, shifts = insertion_sort(a, n, trace=True)
    print("Array ordinato: %s" % a)
    print("Confronti tra chiavi: %d, spostamenti: %d" % (comparisons, shifts))
    print()


run("Istanza 1", [5, 2, 4, 1, 3])
run("Istanza 2", [12, 11, 13, 5, 6])
```

## ExchangeSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in ordine non decrescente.

Al passo $i$ l'elemento $a[i]$ viene confrontato con ogni $a[j]$ con $j > i$ e i due elementi vengono scambiati ogni volta che $a[i] > a[j]$: alla fine del passo $a[i]$ è il minimo di $a[i..n]$. La funzione restituisce il numero di confronti tra chiavi e il numero di scambi.

```python
def row(i, j, a, action):                # una riga della traccia
    print("%3s %3s  %-20s  %s" % (i, j, a, action))


def exchange_sort(a, n, trace=False):
    comparisons = 0                     # confronti tra chiavi
    swaps = 0                           # scambi
    if trace:
        row("-", "-", a, "stato iniziale")
    for i in range(n - 1):              # i = 1..n-1 nello pseudocodice
        for j in range(i + 1, n):       # j = i+1..n nello pseudocodice
            comparisons += 1
            ai, aj = a[i], a[j]         # valori prima dello scambio, per la traccia
            if a[i] > a[j]:
                tmp = a[i]              # scambia a[i] e a[j]
                a[i] = a[j]
                a[j] = tmp
                swaps += 1
                if trace:
                    row(i if j == i + 1 else "", j, a, "a[%d]=%d>a[%d]=%d: scambio a[%d] <-> a[%d]" % (i, ai, j, aj, i, j))
            elif trace:
                row(i if j == i + 1 else "", j, a, "a[%d]=%d<=a[%d]=%d: nessuno scambio" % (i, ai, j, aj))
    return comparisons, swaps
```

## SelectionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in ordine non decrescente.

Al passo $i$ il ciclo interno cerca l'indice $\mathit{minIdx}$ del minimo di $a[i..n]$ (il confronto è stretto: in caso di parità vince il primo minimo) e poi, se $\mathit{minIdx} \ne i$, scambia $a[i]$ con $a[\mathit{minIdx}]$: al più uno scambio per passo. La funzione restituisce il numero di confronti tra chiavi e il numero di scambi.

```python
def row(i, j, min_idx, a, action):      # una riga della traccia
    print("%3s %3s %6s  %-20s  %s" % (i, j, min_idx, a, action))


def selection_sort(a, n, trace=False):
    comparisons = 0                     # confronti tra chiavi
    swaps = 0                           # scambi
    if trace:
        row("-", "-", "-", a, "stato iniziale")
    for i in range(n - 1):              # i = 1..n-1 nello pseudocodice
        min_idx = i                     # indice del minimo in a[i..n]
        if trace:
            row(i, "-", min_idx, a, "minIdx := %d (a[%d]=%d)" % (i, i, a[i]))
        for j in range(i + 1, n):       # j = i+1..n nello pseudocodice
            comparisons += 1
            if a[j] < a[min_idx]:
                old = min_idx           # solo per la traccia
                min_idx = j             # trovato un nuovo minimo
                if trace:
                    row("", j, min_idx, a, "a[%d]=%d<a[%d]=%d: minIdx <- %d" % (j, a[j], old, a[old], j))
            elif trace:
                row("", j, min_idx, a, "a[%d]=%d>=a[%d]=%d: nessun aggiornamento" % (j, a[j], min_idx, a[min_idx]))
        if min_idx != i:
            tmp = a[i]                  # scambia a[i] e a[minIdx]
            a[i] = a[min_idx]
            a[min_idx] = tmp
            swaps += 1
            if trace:
                row("", "-", "-", a, "minIdx=%d != %d: scambio a[%d] <-> a[%d]" % (min_idx, i, i, min_idx))
        elif trace:
            row("", "-", "-", a, "minIdx=%d: nessuno scambio" % min_idx)
    return comparisons, swaps
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s %3s %6s  %-20s  %s" % ("i", "j", "minIdx", "a", "azione"))
    comparisons, swaps = selection_sort(a, n, trace=True)
    print("Array ordinato: %s" % a)
    print("Confronti tra chiavi: %d, scambi: %d" % (comparisons, swaps))
    print()


run("Istanza 1", [5, 2, 4, 1, 3])
run("Istanza 2", [12, 11, 13, 5, 6])
```

### Confronto dei tre algoritmi

Sugli stessi due array, i tre algoritmi eseguono (come nell'esempio di confronto della dispensa):

| algoritmo | $[5, 2, 4, 1, 3]$: confronti | movimenti | $[12, 11, 13, 5, 6]$: confronti | movimenti |
|---|---|---|---|---|
| Insertion Sort | 9 | 7 spostamenti | 9 | 7 spostamenti |
| Exchange Sort | 10 | 7 scambi | 10 | 7 scambi |
| Selection Sort | 10 | 3 scambi | 10 | 3 scambi |

Selection Sort esegue sempre $n(n-1)/2$ confronti ma al più $n-1$ scambi; Insertion Sort si adatta all'input (su un array già ordinato esegue solo $n-1$ confronti e nessuno spostamento) ed è l'unico dei tre ad essere stabile.
