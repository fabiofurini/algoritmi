## 1. Ricerca del massimo: FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il massimo valore di $a$. Nel caso peggiore (array crescente) $S(n) = 4n = \Theta(n)$.

Pseudocodice: sezione 12.2 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile find_maximum.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Riga | Esecuzioni\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Passi totali S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// FindMaximum(a, n)
int FindMaximum(int a[], int n)
{
    int max = a[0]; cnt[2]++;          // max := a[1]
    int i = 1; cnt[3]++;               // i := 2 nello pseudocodice
    cnt[4]++;                          // la condizione del while si conta a ogni valutazione
    while (i < n)                      // i <= n nello pseudocodice
    {
        cnt[5]++;
        if (a[i] > max)
        {
            max = a[i]; cnt[6]++;
        }
        i = i + 1; cnt[8]++;
        cnt[4]++;
    }
    cnt[10]++;
    return max;
}

long long StepsOf(int a[], int n)
{
    Reset();
    FindMaximum(a, n);
    return Total();
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Esempio delle dispense: a[1..4] = (3, 1, 4, 5)
    int a[] = {3, 1, 4, 5};
    int n = 4;
    Reset();
    int m = FindMaximum(a, n);
    printf("a = %s -> max = %d\n", Fmt(a, 0, n).c_str(), m);
    PrintSteps({2, 3, 4, 5, 6, 8, 10});
    long long t = cnt[6];
    printf("sum t(i) = %lld, S(n) = 3n + 1 + sum t(i) = %lld\n", t, 3 * n + 1 + t);

    // Caso peggiore (array crescente) e caso migliore (array decrescente)
    printf("\n");
    int inc[] = {1, 2, 3, 4};
    printf("a = %s (crescente): S = %lld = 4n = %d\n", Fmt(inc, 0, n).c_str(), StepsOf(inc, n), 4 * n);
    int dec[] = {4, 3, 2, 1};
    printf("a = %s (decrescente): S = %lld = 3n + 1 = %d\n", Fmt(dec, 0, n).c_str(), StepsOf(dec, n), 3 * n + 1);

    printf("\n");
    printf("Array crescente (1, 2, ..., n):\n");
    printf("  n | S(n) | 4n\n");
    for (int k = 1; k <= 8; k++)
    {
        int b[8];
        for (int i = 0; i < k; i++) b[i] = i + 1;
        printf("%3d | %4lld | %d\n", k, StepsOf(b, k), 4 * k);
    }
    return 0;
}
#endif
```

## 2. Ordinamento per inserimento: InsertionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in modo non decrescente. Se $t(i)$ è il numero di spostamenti all'iterazione $i$, allora $S(n) = 6n - 4 + 3\sum_{i=2}^{n} t(i)$; nel caso peggiore (array decrescente) $S(n) = \frac{3n^2+9n-8}{2} = \Theta(n^2)$.

Pseudocodice: sezione 12.3 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile insertion_sort.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Riga | Esecuzioni\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Passi totali S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// InsertionSort(a, n)
void InsertionSort(int a[], int n, bool trace = false)
{
    long long prev = 0;                    // solo per la traccia
    int i = 1; cnt[2]++;                   // i := 2 nello pseudocodice
    cnt[3]++;
    while (i < n)                          // i <= n nello pseudocodice
    {
        int key = a[i]; cnt[4]++;
        int j = i - 1; cnt[5]++;
        cnt[6]++;                          // la condizione del while si conta a ogni valutazione
        while (j >= 0 && a[j] > key)       // j >= 1 and a[j] > key nello pseudocodice
        {
            a[j + 1] = a[j]; cnt[7]++;
            j = j - 1; cnt[8]++;
            cnt[6]++;
        }
        a[j + 1] = key; cnt[10]++;
        i = i + 1; cnt[11]++;
        cnt[3]++;
        if (trace)                         // traccia: dopo l'incremento, i e' l'indice dello pseudocodice dell'iterazione appena finita
        {
            printf("%2d | %3d | %4lld | %s\n", i, key, cnt[7] - prev, Fmt(a, 0, n).c_str());
            prev = cnt[7];
        }
    }
}

long long StepsOf(const int a[], int n)    // lavora su una copia di a
{
    int b[64];
    for (int i = 0; i < n; i++) b[i] = a[i];
    Reset();
    InsertionSort(b, n);
    return Total();
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Esempio delle dispense: a[1..6] = (5, 2, 4, 6, 1, 3)
    int a[] = {5, 2, 4, 6, 1, 3};
    int n = 6;
    printf("a = %s\n", Fmt(a, 0, n).c_str());
    Reset();
    printf(" i | key | t(i) | array dopo l'iterazione\n");
    InsertionSort(a, n, true);
    printf("ordinato: %s\n", Fmt(a, 0, n).c_str());
    PrintSteps({2, 3, 4, 5, 6, 7, 8, 10, 11});
    long long t = cnt[7];
    printf("sum t(i) = %lld, S(n) = 6n - 4 + 3 sum t(i) = %lld\n", t, 6 * n - 4 + 3 * t);

    // Caso peggiore: array decrescente, t(i) = i - 1
    printf("\n");
    int d[] = {6, 5, 4, 3, 2, 1};
    printf("a = %s (decrescente): S = %lld = (3n^2 + 9n - 8)/2 = %d\n", Fmt(d, 0, n).c_str(), StepsOf(d, n), (3 * n * n + 9 * n - 8) / 2);
    printf("sum t(i) = %lld = n(n-1)/2 = %d\n", cnt[7], n * (n - 1) / 2);

    printf("\n");
    printf("Array decrescente (n, ..., 2, 1):\n");
    printf("  n | S(n) | (3n^2 + 9n - 8)/2\n");
    for (int k = 1; k <= 8; k++)
    {
        int b[8];
        for (int i = 0; i < k; i++) b[i] = k - i;
        printf("%3d | %4lld | %d\n", k, StepsOf(b, k), (3 * k * k + 9 * k - 8) / 2);
    }
    return 0;
}
#endif
```

```python
!g++ -std=c++17 -O2 -o insertion_sort insertion_sort.cpp && ./insertion_sort
```

**Prova tu.** Cambia l'array e controlla la formula $6n - 4 + 3\sum t(i)$.

```python
%%writefile prova_insertion_sort.cpp
#define SOLO_LIBRERIA   // non eseguire il main di insertion_sort.cpp
#include "insertion_sort.cpp"

int main()
{
    int a[] = {8, 3, 5, 1, 9, 2, 7};      // <-- cambia l'array
    int n = 7;
    Reset();
    printf(" i | key | t(i) | array dopo l'iterazione\n");
    InsertionSort(a, n, true);
    PrintSteps({2, 3, 4, 5, 6, 7, 8, 10, 11});
    printf("6n - 4 + 3 sum t(i) = %lld\n", 6 * n - 4 + 3 * cnt[7]);
    return 0;
}
```

## 3. Cambio di base: ChangeToBase2

**Input:** un numero naturale $n \ge 1$ in base 10. **Output:** l'array $d[0..nd-1]$ delle cifre binarie e il numero di cifre $nd$. Per ogni $n$: $S(n) = 6\lfloor\log_2 n\rfloor + 11 = \Theta(\log_2 n)$.

Pseudocodice: sezione 12.4 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile change_to_base2.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Riga | Esecuzioni\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Passi totali S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

int FloorLog2(int n)                       // floor(log2 n) calcolato con un ciclo (non conta come passo)
{
    int k = 0;
    while (n > 1)
    {
        n = n / 2;
        k = k + 1;
    }
    return k;
}

// ChangeToBase2(n): restituisce l'array d[0..nd-1] (da liberare con delete[]) e nd
int* ChangeToBase2(int n, int& nd, bool trace = false)
{
    nd = FloorLog2(n) + 1; cnt[2]++;
    int* d = new int[nd]; cnt[3]++;        // array d[0..nd-1], indicizzato da 0 come nello pseudocodice
    int k = 0; cnt[4]++;
    cnt[5]++;                              // la condizione del while si conta a ogni valutazione
    while (n > 0)
    {
        int q = n / 2; cnt[6]++;
        int r = n % 2; cnt[7]++;
        d[k] = r; cnt[8]++;
        if (trace)                         // traccia
        {
            printf("%10d | %3d | %3d | %3d | d[%d] = %d\n", k + 1, n, q, r, k, r);
        }
        n = q; cnt[9]++;
        k = k + 1; cnt[10]++;
        cnt[5]++;
    }
    cnt[12]++;
    return d;
}

string Digits(const int* d, int nd)        // d[nd-1..0] (dalla cifra piu' significativa)
{
    string s = "";
    for (int k = nd - 1; k >= 0; k--) s += to_string(d[k]);
    return s;
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Esempio delle dispense: n = 13
    int n = 13, nd;
    Reset();
    printf("iterazione |   n |   q |   r | assegnamento\n");
    int* d = ChangeToBase2(n, nd, true);
    printf("n = %d = (%s)_2: nd = %d, cifre d[nd-1..0] = %s\n", n, Digits(d, nd).c_str(), nd, Digits(d, nd).c_str());
    delete[] d;
    PrintSteps({2, 3, 4, 5, 6, 7, 8, 9, 10, 12});
    printf("S(n) = 6 floor(log2 n) + 11 = %d\n", 6 * FloorLog2(n) + 11);

    // Il numero di passi cresce come log2 n
    printf("\n");
    printf("       n | nd | S(n) | 6 floor(log2 n) + 11\n");
    int values[] = {1, 2, 13, 100, 1000, 1000000};
    for (int v : values)
    {
        Reset();
        d = ChangeToBase2(v, nd);
        printf("%8d | %2d | %4lld | %d\n", v, nd, Total(), 6 * FloorLog2(v) + 11);
        delete[] d;
    }
    return 0;
}
#endif
```

```python
!g++ -std=c++17 -O2 -o change_to_base2 change_to_base2.cpp && ./change_to_base2
```

**Prova tu.** Cambia $n$ e verifica che $S(n) = 6\lfloor\log_2 n\rfloor + 11$.

```python
%%writefile prova_change_to_base2.cpp
#define SOLO_LIBRERIA   // non eseguire il main di change_to_base2.cpp
#include "change_to_base2.cpp"

int main()
{
    int n = 2026, nd;        // <-- cambia il numero
    Reset();
    int* d = ChangeToBase2(n, nd);
    printf("n = %d = (%s)_2, nd = %d, S = %lld\n", n, Digits(d, nd).c_str(), nd, Total());
    delete[] d;
    return 0;
}
```

## 4. Test di primalità: PrimalityTest

**Input:** un numero naturale $n$. **Output:** `true` se $n$ è primo, `false` altrimenti. Nel caso peggiore ($n$ primo) $S(n) = 3\lfloor\sqrt{n}\rfloor + 1 = \Theta(\sqrt{n})$, cioè $2^{\Theta(|I|)}$ in funzione della dimensione dell'input.

Pseudocodice: sezione 12.5 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile primality_test.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Riga | Esecuzioni\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Passi totali S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

long long FloorSqrt(long long n)           // floor(sqrt(n)) in aritmetica intera
{
    long long r = (long long) sqrtl((long double) n);
    while (r * r > n) r--;
    while ((r + 1) * (r + 1) <= n) r++;
    return r;
}

// PrimalityTest(n)
bool PrimalityTest(long long n)
{
    cnt[2]++;
    if (n < 2)
    {
        cnt[3]++;
        return false;
    }
    long long d = 2; cnt[5]++;
    cnt[6]++;                              // la condizione del while si conta a ogni valutazione
    while (d <= FloorSqrt(n))
    {
        cnt[7]++;
        if (n % d == 0)
        {
            cnt[8]++;
            return false;
        }
        d = d + 1; cnt[10]++;
        cnt[6]++;
    }
    cnt[12]++;
    return true;
}

int Bits(long long n)                      // numero di bit di n >= 1
{
    int k = 0;
    while (n > 0)
    {
        n = n / 2;
        k = k + 1;
    }
    return k;
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Esempio delle dispense: un numero primo e uno composto
    long long n = 97;
    Reset();
    bool r = PrimalityTest(n);
    printf("n = %lld: %s\n", n, r ? "true" : "false");
    PrintSteps({2, 3, 5, 6, 7, 8, 10, 12});
    printf("S(n) = 3 floor(sqrt(n)) + 1 = %lld\n", 3 * FloorSqrt(n) + 1);

    printf("\n");
    n = 91;
    Reset();
    r = PrimalityTest(n);
    long long p = 2;
    while (n % p != 0) p = p + 1;
    printf("n = %lld: %s, p = %lld\n", n, r ? "true" : "false", p);
    PrintSteps({2, 3, 5, 6, 7, 8, 10, 12});
    printf("S(n) = 3p - 1 = %lld, mentre floor(sqrt(n)) = %lld\n", 3 * p - 1, FloorSqrt(n));

    printf("\n");
    n = 1;
    Reset();
    r = PrimalityTest(n);
    printf("n = %lld: %s, S = %lld\n", n, r ? "true" : "false", Total());

    // Il caso peggiore sono i numeri primi: S cresce come sqrt(n), cioe' raddoppia ogni 2 bit in piu'
    printf("\n");
    printf("|I| |        n |  S(n) | 3 floor(sqrt(n)) + 1\n");
    long long primes[] = {1021, 4093, 16381, 65521, 262139, 1048573, 4194301, 16777213};
    for (long long q : primes)
    {
        Reset();
        PrimalityTest(q);
        printf("%3d | %8lld | %5lld | %lld\n", Bits(q), q, Total(), 3 * FloorSqrt(q) + 1);
    }
    return 0;
}
#endif
```

```python
!g++ -std=c++17 -O2 -o primality_test primality_test.cpp && ./primality_test
```

**Prova tu.** Cambia $n$: per un primo $S = 3\lfloor\sqrt{n}\rfloor + 1$, per un composto $S = 3p - 1$.

```python
%%writefile prova_primality_test.cpp
#define SOLO_LIBRERIA   // non eseguire il main di primality_test.cpp
#include "primality_test.cpp"

int main()
{
    long long n = 7919;        // <-- cambia il numero
    Reset();
    bool r = PrimalityTest(n);
    printf("n = %lld: %s, S = %lld, 3 floor(sqrt(n)) + 1 = %lld\n", n, r ? "true" : "false", Total(), 3 * FloorSqrt(n) + 1);
    return 0;
}
```
