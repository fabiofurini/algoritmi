## SelectionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in ordine non decrescente.

Al passo $i$ il ciclo interno cerca l'indice $\mathit{minIdx}$ del minimo di $a[i..n]$ (il confronto è stretto: in caso di parità vince il primo minimo) e poi, se $\mathit{minIdx} \ne i$, scambia $a[i]$ con $a[\mathit{minIdx}]$: al più uno scambio per passo. La funzione restituisce il numero di confronti tra chiavi e il numero di scambi.

```python
%%writefile SelectionSort.java
import java.util.Arrays;

public class SelectionSort {

    // Una riga della traccia
    static void row(String i, String j, String minIdx, int[] a, String action) {
        System.out.printf("%3s %3s %6s  %-20s  %s\n", i, j, minIdx, Arrays.toString(a), action);
    }

    // Ordina a[0..n-1] sul posto; restituisce {confronti tra chiavi, scambi}
    static int[] selectionSort(int[] a, int n, boolean trace) {
        int comparisons = 0;                // confronti tra chiavi
        int swaps = 0;                      // scambi
        if (trace)
            row("-", "-", "-", a, "stato iniziale");

        for (int i = 0; i < n - 1; i++) {   // i = 1..n-1 nello pseudocodice
            int minIdx = i;                 // indice del minimo in a[i..n]
            if (trace)
                row("" + i, "-", "" + minIdx, a, String.format("minIdx := %d (a[%d]=%d)", i, i, a[i]));
            for (int j = i + 1; j < n; j++) {   // j = i+1..n nello pseudocodice
                comparisons++;
                if (a[j] < a[minIdx]) {
                    int old = minIdx;       // solo per la traccia
                    minIdx = j;             // trovato un nuovo minimo
                    if (trace)
                        row("", "" + j, "" + minIdx, a, String.format("a[%d]=%d<a[%d]=%d: minIdx <- %d", j, a[j], old, a[old], j));
                } else if (trace)
                    row("", "" + j, "" + minIdx, a, String.format("a[%d]=%d>=a[%d]=%d: nessun aggiornamento", j, a[j], minIdx, a[minIdx]));
            }
            if (minIdx != i) {
                int tmp = a[i];             // scambia a[i] e a[minIdx]
                a[i] = a[minIdx];
                a[minIdx] = tmp;
                swaps++;
                if (trace)
                    row("", "-", "-", a, String.format("minIdx=%d != %d: scambio a[%d] <-> a[%d]", minIdx, i, i, minIdx));
            } else if (trace)
                row("", "-", "-", a, String.format("minIdx=%d: nessuno scambio", minIdx));
        }
        return new int[]{comparisons, swaps};
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s %3s %6s  %-20s  %s\n", "i", "j", "minIdx", "a", "azione");
        int[] c = selectionSort(a, n, true);
        System.out.printf("Array ordinato: %s\n", Arrays.toString(a));
        System.out.printf("Confronti tra chiavi: %d, scambi: %d\n\n", c[0], c[1]);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // «Prova tu»: l'array arriva dalla riga di comando
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Prova tu", v);
            return;
        }
        run("Istanza 1", new int[]{5, 2, 4, 1, 3});
        run("Istanza 2", new int[]{12, 11, 13, 5, 6});
    }
}
```
