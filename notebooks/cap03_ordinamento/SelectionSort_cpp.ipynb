## SelectionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in ordine non decrescente.

Al passo $i$ il ciclo interno cerca l'indice $\mathit{minIdx}$ del minimo di $a[i..n]$ (il confronto è stretto: in caso di parità vince il primo minimo) e poi, se $\mathit{minIdx} \ne i$, scambia $a[i]$ con $a[\mathit{minIdx}]$: al più uno scambio per passo. La funzione restituisce il numero di confronti tra chiavi e il numero di scambi.

```python
%%writefile selection_sort.cpp
#include <cstdarg>
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

// Funzioni di servizio per stampare la traccia
string S(int x) { return to_string(x); }

string Str(int n, const int *a)     // l'array come "[5, 2, 4, 1, 3]"
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

string F(const char *fmt, ...)      // come printf, ma restituisce una string
{
    char buf[200];
    va_list ap;
    va_start(ap, fmt);
    vsnprintf(buf, sizeof buf, fmt, ap);
    va_end(ap);
    return buf;
}

void Row(string i, string j, string minIdx, int n, const int *a, string action)    // una riga della traccia
{
    printf("%3s %3s %6s  %-20s  %s\n", i.c_str(), j.c_str(), minIdx.c_str(), Str(n, a).c_str(), action.c_str());
}

// Ordina a[0..n-1] sul posto; restituisce confronti tra chiavi e scambi
void SelectionSort(int n, int *a, int &comparisons, int &swaps, bool trace = false)
{
    comparisons = 0;                    // confronti tra chiavi
    swaps = 0;                          // scambi
    if (trace)
        Row("-", "-", "-", n, a, "stato iniziale");

    for (int i = 0; i < n - 1; i++)     // i = 1..n-1 nello pseudocodice
    {
        int minIdx = i;                 // indice del minimo in a[i..n]
        if (trace)
            Row(S(i), "-", S(minIdx), n, a, F("minIdx := %d (a[%d]=%d)", i, i, a[i]));
        for (int j = i + 1; j < n; j++) // j = i+1..n nello pseudocodice
        {
            comparisons++;
            if (a[j] < a[minIdx])
            {
                int old = minIdx;       // solo per la traccia
                minIdx = j;             // trovato un nuovo minimo
                if (trace)
                    Row("", S(j), S(minIdx), n, a, F("a[%d]=%d<a[%d]=%d: minIdx <- %d", j, a[j], old, a[old], j));
            }
            else if (trace)
                Row("", S(j), S(minIdx), n, a, F("a[%d]=%d>=a[%d]=%d: nessun aggiornamento", j, a[j], minIdx, a[minIdx]));
        }
        if (minIdx != i)
        {
            int tmp = a[i];             // scambia a[i] e a[minIdx]
            a[i] = a[minIdx];
            a[minIdx] = tmp;
            swaps++;
            if (trace)
                Row("", "-", "-", n, a, F("minIdx=%d != %d: scambio a[%d] <-> a[%d]", minIdx, i, i, minIdx));
        }
        else if (trace)
            Row("", "-", "-", n, a, F("minIdx=%d: nessuno scambio", minIdx));
    }
}

void Run(string title, vector<int> v)
{
    int n = v.size(), comparisons, swaps;
    printf("%s: a = %s, n = %d\n", title.c_str(), Str(n, v.data()).c_str(), n);
    printf("%3s %3s %6s  %-20s  %s\n", "i", "j", "minIdx", "a", "azione");
    SelectionSort(n, v.data(), comparisons, swaps, true);
    printf("Array ordinato: %s\n", Str(n, v.data()).c_str());
    printf("Confronti tra chiavi: %d, scambi: %d\n\n", comparisons, swaps);
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // «Prova tu»: l'array arriva dalla riga di comando
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        Run("Prova tu", v);
        return 0;
    }
    Run("Istanza 1", {5, 2, 4, 1, 3});
    Run("Istanza 2", {12, 11, 13, 5, 6});
    return 0;
}
```
