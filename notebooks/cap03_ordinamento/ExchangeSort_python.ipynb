## ExchangeSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in ordine non decrescente.

Al passo $i$ l'elemento $a[i]$ viene confrontato con ogni $a[j]$ con $j > i$ e i due elementi vengono scambiati ogni volta che $a[i] > a[j]$: alla fine del passo $a[i]$ è il minimo di $a[i..n]$. La funzione restituisce il numero di confronti tra chiavi e il numero di scambi.

```python
def row(i, j, a, action):                # una riga della traccia
    print("%3s %3s  %-20s  %s" % (i, j, a, action))


def exchange_sort(a, n, trace=False):
    comparisons = 0                     # confronti tra chiavi
    swaps = 0                           # scambi
    if trace:
        row("-", "-", a, "stato iniziale")
    for i in range(n - 1):              # i = 1..n-1 nello pseudocodice
        for j in range(i + 1, n):       # j = i+1..n nello pseudocodice
            comparisons += 1
            ai, aj = a[i], a[j]         # valori prima dello scambio, per la traccia
            if a[i] > a[j]:
                tmp = a[i]              # scambia a[i] e a[j]
                a[i] = a[j]
                a[j] = tmp
                swaps += 1
                if trace:
                    row(i if j == i + 1 else "", j, a, "a[%d]=%d>a[%d]=%d: scambio a[%d] <-> a[%d]" % (i, ai, j, aj, i, j))
            elif trace:
                row(i if j == i + 1 else "", j, a, "a[%d]=%d<=a[%d]=%d: nessuno scambio" % (i, ai, j, aj))
    return comparisons, swaps
```

### Esecuzione sugli esempi del capitolo

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s %3s  %-20s  %s" % ("i", "j", "a", "azione"))
    comparisons, swaps = exchange_sort(a, n, trace=True)
    print("Array ordinato: %s" % a)
    print("Confronti tra chiavi: %d, scambi: %d" % (comparisons, swaps))
    print()


run("Istanza 1", [5, 2, 4, 1, 3])
run("Istanza 2", [12, 11, 13, 5, 6])
```
