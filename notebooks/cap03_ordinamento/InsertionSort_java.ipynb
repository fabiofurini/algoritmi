## InsertionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in ordine non decrescente.

Al passo $i$ il sotto-array $a[1..i-1]$ è ordinato: la chiave $a[i]$ viene inserita al suo posto, dopo aver spostato di una posizione a destra le chiavi maggiori. La funzione restituisce anche i contatori usati nella dispensa: il numero di confronti tra chiavi e il numero di spostamenti (nessuno scambio: uno spostamento è una sola assegnazione).

```python
%%writefile InsertionSort.java
import java.util.Arrays;

public class InsertionSort {

    // Una riga della traccia
    static void row(String i, String j, String key, int[] a, String action) {
        System.out.printf("%3s %3s %4s  %-20s  %s\n", i, j, key, Arrays.toString(a), action);
    }

    // Ordina a[0..n-1] sul posto; restituisce {confronti tra chiavi, spostamenti}
    static int[] insertionSort(int[] a, int n, boolean trace) {
        int comparisons = 0;                // confronti tra chiavi
        int shifts = 0;                     // spostamenti
        if (trace)
            row("-", "-", "-", a, "stato iniziale");

        for (int i = 1; i < n; i++) {       // i = 2..n nello pseudocodice
            int key = a[i];                 // elemento da inserire
            int j = i - 1;
            if (trace)
                row("" + i, "" + j, "" + key, a, String.format("key <- a[%d]=%d; j <- %d", i, key, j));

            // sposta a destra le chiavi maggiori di key
            while (j >= 0 && a[j] > key) {
                comparisons++;
                shifts++;
                a[j + 1] = a[j];
                j = j - 1;
                if (trace)
                    row("", "" + j, "" + key, a, String.format("a[%d]=%d>%d: sposta a[%d] <- %d; j <- %d", j + 1, a[j + 1], key, j + 2, a[j + 1], j));
            }
            if (j >= 0)
                comparisons++;              // il confronto che ha fermato il ciclo (a[j] <= key)

            // inserisce key nella posizione giusta
            a[j + 1] = key;
            if (trace) {
                if (j < 0)
                    row("", "-", "" + key, a, String.format("j=%d: esce; inserisce a[%d] <- %d", j, j + 1, key));
                else
                    row("", "-", "" + key, a, String.format("a[%d]=%d<=%d: esce; inserisce a[%d] <- %d", j, a[j], key, j + 1, key));
            }
        }
        return new int[]{comparisons, shifts};
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s %3s %4s  %-20s  %s\n", "i", "j", "key", "a", "azione");
        int[] c = insertionSort(a, n, true);
        System.out.printf("Array ordinato: %s\n", Arrays.toString(a));
        System.out.printf("Confronti tra chiavi: %d, spostamenti: %d\n\n", c[0], c[1]);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // «Prova tu»: l'array arriva dalla riga di comando
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Prova tu", v);
            return;
        }
        run("Istanza 1", new int[]{5, 2, 4, 1, 3});
        run("Istanza 2", new int[]{12, 11, 13, 5, 6});
    }
}
```
