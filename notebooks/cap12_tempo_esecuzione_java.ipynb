## 1. Ricerca del massimo: FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il massimo valore di $a$. Nel caso peggiore (array crescente) $S(n) = 4n = \Theta(n)$.

Pseudocodice: sezione 12.2 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile FindMaximum.java
public class FindMaximum
{
static long[] cnt = new long[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Riga | Esecuzioni");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Passi totali S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    // FindMaximum(a, n)
    static int findMaximum(int[] a)
    {
        int n = a.length;
        int max = a[0]; cnt[2]++;          // max := a[1]
        int i = 1; cnt[3]++;               // i := 2 nello pseudocodice
        cnt[4]++;                          // la condizione del while si conta a ogni valutazione
        while (i < n)                      // i <= n nello pseudocodice
        {
            cnt[5]++;
            if (a[i] > max)
            {
                max = a[i]; cnt[6]++;
            }
            i = i + 1; cnt[8]++;
            cnt[4]++;
        }
        cnt[10]++;
        return max;
    }

    static long stepsOf(int[] a)
    {
        reset();
        findMaximum(a);
        return total();
    }

    public static void main(String[] args)
    {
        // Esempio delle dispense: a[1..4] = (3, 1, 4, 5)
        int[] a = {3, 1, 4, 5};
        int n = a.length;
        reset();
        int m = findMaximum(a);
        System.out.printf("a = %s -> max = %d\n", fmt(a, 0, n), m);
        printSteps(2, 3, 4, 5, 6, 8, 10);
        long t = cnt[6];
        System.out.printf("sum t(i) = %d, S(n) = 3n + 1 + sum t(i) = %d\n", t, 3 * n + 1 + t);

        // Caso peggiore (array crescente) e caso migliore (array decrescente)
        System.out.printf("\n");
        int[] inc = {1, 2, 3, 4};
        System.out.printf("a = %s (crescente): S = %d = 4n = %d\n", fmt(inc, 0, n), stepsOf(inc), 4 * n);
        int[] dec = {4, 3, 2, 1};
        System.out.printf("a = %s (decrescente): S = %d = 3n + 1 = %d\n", fmt(dec, 0, n), stepsOf(dec), 3 * n + 1);

        System.out.printf("\n");
        System.out.printf("Array crescente (1, 2, ..., n):\n");
        System.out.printf("  n | S(n) | 4n\n");
        for (int k = 1; k <= 8; k++)
        {
            int[] b = new int[k];
            for (int i = 0; i < k; i++) b[i] = i + 1;
            System.out.printf("%3d | %4d | %d\n", k, stepsOf(b), 4 * k);
        }
    }
}
```

## 2. Ordinamento per inserimento: InsertionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in modo non decrescente. Se $t(i)$ è il numero di spostamenti all'iterazione $i$, allora $S(n) = 6n - 4 + 3\sum_{i=2}^{n} t(i)$; nel caso peggiore (array decrescente) $S(n) = \frac{3n^2+9n-8}{2} = \Theta(n^2)$.

Pseudocodice: sezione 12.3 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile InsertionSort.java
public class InsertionSort
{
static long[] cnt = new long[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Riga | Esecuzioni");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Passi totali S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    // InsertionSort(a, n)
    static void insertionSort(int[] a, boolean trace)
    {
        int n = a.length;
        long prev = 0;                     // solo per la traccia
        int i = 1; cnt[2]++;               // i := 2 nello pseudocodice
        cnt[3]++;
        while (i < n)                      // i <= n nello pseudocodice
        {
            int key = a[i]; cnt[4]++;
            int j = i - 1; cnt[5]++;
            cnt[6]++;                      // la condizione del while si conta a ogni valutazione
            while (j >= 0 && a[j] > key)   // j >= 1 and a[j] > key nello pseudocodice
            {
                a[j + 1] = a[j]; cnt[7]++;
                j = j - 1; cnt[8]++;
                cnt[6]++;
            }
            a[j + 1] = key; cnt[10]++;
            i = i + 1; cnt[11]++;
            cnt[3]++;
            if (trace)                     // traccia: dopo l'incremento, i e' l'indice dello pseudocodice dell'iterazione appena finita
            {
                System.out.println(String.format("%2d | %3d | %4d | %s", i, key, cnt[7] - prev, fmt(a, 0, n)));
                prev = cnt[7];
            }
        }
    }

    static long stepsOf(int[] a)           // lavora su una copia di a
    {
        reset();
        insertionSort(a.clone(), false);
        return total();
    }

    public static void main(String[] args)
    {
        // Esempio delle dispense: a[1..6] = (5, 2, 4, 6, 1, 3)
        int[] a = {5, 2, 4, 6, 1, 3};
        int n = a.length;
        System.out.printf("a = %s\n", fmt(a, 0, n));
        reset();
        System.out.printf(" i | key | t(i) | array dopo l'iterazione\n");
        insertionSort(a, true);
        System.out.printf("ordinato: %s\n", fmt(a, 0, n));
        printSteps(2, 3, 4, 5, 6, 7, 8, 10, 11);
        long t = cnt[7];
        System.out.printf("sum t(i) = %d, S(n) = 6n - 4 + 3 sum t(i) = %d\n", t, 6 * n - 4 + 3 * t);

        // Caso peggiore: array decrescente, t(i) = i - 1
        System.out.printf("\n");
        int[] d = {6, 5, 4, 3, 2, 1};
        System.out.printf("a = %s (decrescente): S = %d = (3n^2 + 9n - 8)/2 = %d\n", fmt(d, 0, n), stepsOf(d), (3 * n * n + 9 * n - 8) / 2);
        System.out.printf("sum t(i) = %d = n(n-1)/2 = %d\n", cnt[7], n * (n - 1) / 2);

        System.out.printf("\n");
        System.out.printf("Array decrescente (n, ..., 2, 1):\n");
        System.out.printf("  n | S(n) | (3n^2 + 9n - 8)/2\n");
        for (int k = 1; k <= 8; k++)
        {
            int[] b = new int[k];
            for (int i = 0; i < k; i++) b[i] = k - i;
            System.out.printf("%3d | %4d | %d\n", k, stepsOf(b), (3 * k * k + 9 * k - 8) / 2);
        }
    }
}
```

**Prova tu.** Cambia l'array e controlla la formula $6n - 4 + 3\sum t(i)$.

```python
%%writefile ProvaInsertionSort.java
public class ProvaInsertionSort
{
    public static void main(String[] args)
    {
        int[] a = {8, 3, 5, 1, 9, 2, 7};      // <-- cambia l'array
        int n = a.length;
        InsertionSort.reset();
        System.out.printf(" i | key | t(i) | array dopo l'iterazione\n");
        InsertionSort.insertionSort(a, true);
        InsertionSort.printSteps(2, 3, 4, 5, 6, 7, 8, 10, 11);
        System.out.printf("6n - 4 + 3 sum t(i) = %d\n", 6 * n - 4 + 3 * InsertionSort.cnt[7]);
    }
}
```

## 3. Cambio di base: ChangeToBase2

**Input:** un numero naturale $n \ge 1$ in base 10. **Output:** l'array $d[0..nd-1]$ delle cifre binarie e il numero di cifre $nd$. Per ogni $n$: $S(n) = 6\lfloor\log_2 n\rfloor + 11 = \Theta(\log_2 n)$.

Pseudocodice: sezione 12.4 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile ChangeToBase2.java
public class ChangeToBase2
{
static long[] cnt = new long[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Riga | Esecuzioni");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Passi totali S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    static int floorLog2(int n)            // floor(log2 n) calcolato con un ciclo (non conta come passo)
    {
        int k = 0;
        while (n > 1)
        {
            n = n / 2;
            k = k + 1;
        }
        return k;
    }

    // ChangeToBase2(n): restituisce l'array d[0..nd-1] (nd = d.length)
    static int[] changeToBase2(int n, boolean trace)
    {
        int nd = floorLog2(n) + 1; cnt[2]++;
        int[] d = new int[nd]; cnt[3]++;   // array d[0..nd-1], indicizzato da 0 come nello pseudocodice
        int k = 0; cnt[4]++;
        cnt[5]++;                          // la condizione del while si conta a ogni valutazione
        while (n > 0)
        {
            int q = n / 2; cnt[6]++;
            int r = n % 2; cnt[7]++;
            d[k] = r; cnt[8]++;
            if (trace)                     // traccia
            {
                System.out.println(String.format("%10d | %3d | %3d | %3d | d[%d] = %d", k + 1, n, q, r, k, r));
            }
            n = q; cnt[9]++;
            k = k + 1; cnt[10]++;
            cnt[5]++;
        }
        cnt[12]++;
        return d;
    }

    static String digits(int[] d)          // d[nd-1..0] (dalla cifra piu' significativa)
    {
        StringBuilder s = new StringBuilder();
        for (int k = d.length - 1; k >= 0; k--) s.append(d[k]);
        return s.toString();
    }

    public static void main(String[] args)
    {
        // Esempio delle dispense: n = 13
        int n = 13;
        reset();
        System.out.printf("iterazione |   n |   q |   r | assegnamento\n");
        int[] d = changeToBase2(n, true);
        System.out.printf("n = %d = (%s)_2: nd = %d, cifre d[nd-1..0] = %s\n", n, digits(d), d.length, digits(d));
        printSteps(2, 3, 4, 5, 6, 7, 8, 9, 10, 12);
        System.out.printf("S(n) = 6 floor(log2 n) + 11 = %d\n", 6 * floorLog2(n) + 11);

        // Il numero di passi cresce come log2 n
        System.out.printf("\n");
        System.out.printf("       n | nd | S(n) | 6 floor(log2 n) + 11\n");
        int[] values = {1, 2, 13, 100, 1000, 1000000};
        for (int v : values)
        {
            reset();
            d = changeToBase2(v, false);
            System.out.printf("%8d | %2d | %4d | %d\n", v, d.length, total(), 6 * floorLog2(v) + 11);
        }
    }
}
```

**Prova tu.** Cambia $n$ e verifica che $S(n) = 6\lfloor\log_2 n\rfloor + 11$.

```python
%%writefile ProvaChangeToBase2.java
public class ProvaChangeToBase2
{
    public static void main(String[] args)
    {
        int n = 2026;        // <-- cambia il numero
        ChangeToBase2.reset();
        int[] d = ChangeToBase2.changeToBase2(n, false);
        System.out.printf("n = %d = (%s)_2, nd = %d, S = %d\n", n, ChangeToBase2.digits(d), d.length, ChangeToBase2.total());
    }
}
```

## 4. Test di primalità: PrimalityTest

**Input:** un numero naturale $n$. **Output:** `true` se $n$ è primo, `false` altrimenti. Nel caso peggiore ($n$ primo) $S(n) = 3\lfloor\sqrt{n}\rfloor + 1 = \Theta(\sqrt{n})$, cioè $2^{\Theta(|I|)}$ in funzione della dimensione dell'input.

Pseudocodice: sezione 12.5 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile PrimalityTest.java
public class PrimalityTest
{
static long[] cnt = new long[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Riga | Esecuzioni");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Passi totali S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    static long floorSqrt(long n)          // floor(sqrt(n)) in aritmetica intera
    {
        long r = (long) Math.sqrt((double) n);
        while (r * r > n) r--;
        while ((r + 1) * (r + 1) <= n) r++;
        return r;
    }

    // PrimalityTest(n)
    static boolean primalityTest(long n)
    {
        cnt[2]++;
        if (n < 2)
        {
            cnt[3]++;
            return false;
        }
        long d = 2; cnt[5]++;
        cnt[6]++;                          // la condizione del while si conta a ogni valutazione
        while (d <= floorSqrt(n))
        {
            cnt[7]++;
            if (n % d == 0)
            {
                cnt[8]++;
                return false;
            }
            d = d + 1; cnt[10]++;
            cnt[6]++;
        }
        cnt[12]++;
        return true;
    }

    static int bits(long n)                // numero di bit di n >= 1
    {
        int k = 0;
        while (n > 0)
        {
            n = n / 2;
            k = k + 1;
        }
        return k;
    }

    public static void main(String[] args)
    {
        // Esempio delle dispense: un numero primo e uno composto
        long n = 97;
        reset();
        boolean r = primalityTest(n);
        System.out.printf("n = %d: %s\n", n, r ? "true" : "false");
        printSteps(2, 3, 5, 6, 7, 8, 10, 12);
        System.out.printf("S(n) = 3 floor(sqrt(n)) + 1 = %d\n", 3 * floorSqrt(n) + 1);

        System.out.printf("\n");
        n = 91;
        reset();
        r = primalityTest(n);
        long p = 2;
        while (n % p != 0) p = p + 1;
        System.out.printf("n = %d: %s, p = %d\n", n, r ? "true" : "false", p);
        printSteps(2, 3, 5, 6, 7, 8, 10, 12);
        System.out.printf("S(n) = 3p - 1 = %d, mentre floor(sqrt(n)) = %d\n", 3 * p - 1, floorSqrt(n));

        System.out.printf("\n");
        n = 1;
        reset();
        r = primalityTest(n);
        System.out.printf("n = %d: %s, S = %d\n", n, r ? "true" : "false", total());

        // Il caso peggiore sono i numeri primi: S cresce come sqrt(n), cioe' raddoppia ogni 2 bit in piu'
        System.out.printf("\n");
        System.out.printf("|I| |        n |  S(n) | 3 floor(sqrt(n)) + 1\n");
        long[] primes = {1021, 4093, 16381, 65521, 262139, 1048573, 4194301, 16777213};
        for (long q : primes)
        {
            reset();
            primalityTest(q);
            System.out.printf("%3d | %8d | %5d | %d\n", bits(q), q, total(), 3 * floorSqrt(q) + 1);
        }
    }
}
```

```python
!javac PrimalityTest.java && java PrimalityTest
```

**Prova tu.** Cambia $n$: per un primo $S = 3\lfloor\sqrt{n}\rfloor + 1$, per un composto $S = 3p - 1$.

```python
%%writefile ProvaPrimalityTest.java
public class ProvaPrimalityTest
{
    public static void main(String[] args)
    {
        long n = 7919;        // <-- cambia il numero
        PrimalityTest.reset();
        boolean r = PrimalityTest.primalityTest(n);
        System.out.printf("n = %d: %s, S = %d, 3 floor(sqrt(n)) + 1 = %d\n", n, r ? "true" : "false", PrimalityTest.total(), 3 * PrimalityTest.floorSqrt(n) + 1);
    }
}
```
