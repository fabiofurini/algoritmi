## 1. Ricerca del massimo: FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il massimo di $a$. **Invariante:** ogni volta che si valuta la condizione, $max = \max\{a[1],\dots,a[i-1]\}$.

Pseudocodice: sezione 13.2 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/13-correttezza/).

```python
%%writefile FindMaximum.java
public class FindMaximum
{
// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

static boolean equal(int[] x, int xs, int[] y, int ys, int len)     // x[xs..xs+len-1] == y[ys..ys+len-1]
{
    for (int k = 0; k < len; k++)
        if (x[xs + k] != y[ys + k]) return false;
    return true;
}

    static int maxOf(int[] a, int from, int to)     // massimo di a[from..to-1]
    {
        int m = a[from];
        for (int k = from + 1; k < to; k++)
            if (a[k] > m) m = a[k];
        return m;
    }

    // Invariante: max = massimo di a[1..i-1]
    static void printInvariant(int[] a, int i, int max)
    {
        int m = maxOf(a, 0, i);                     // a[1..i-1] dello pseudocodice
        System.out.println(String.format("%2d | %3d | %-26s | %14d | %s", i + 1, max, fmt(a, 0, i), m, max == m ? "ok" : "FALSA"));
    }

    // FindMaximum(a, n)
    static int findMaximum(int[] a)
    {
        int n = a.length;
        int max = a[0];
        int i = 1;                                  // i := 2
        printInvariant(a, i, max);                  // prima della prima valutazione della condizione
        while (i < n)                               // i <= n
        {
            if (a[i] > max)
            {
                max = a[i];
            }
            i = i + 1;
            printInvariant(a, i, max);              // dopo ogni iterazione, cioe' prima della valutazione successiva
        }
        return max;
    }

    public static void main(String[] args)
    {
        int[] a = {9, 12, 3, 16, 19, 11, 12};       // esempio delle dispense
        int n = a.length;
        System.out.printf("a = %s, n = %d\n", fmt(a, 0, n), n);
        System.out.printf(" i | max | a[1..i-1]                  | max{a[1..i-1]} | invariante\n");
        int m = findMaximum(a);
        System.out.printf("Il ciclo termina con i = n + 1 = %d: FindMaximum restituisce %d\n", n + 1, m);
    }
}
```

## 2. Ordinamento per inserimento: InsertionSort

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri naturali. **Output:** l'array ordinato in modo non decrescente. **Invariante esterno:** $a[1..i-1]$ contiene gli elementi originali di $a[1..i-1]$ ordinati e $a[i..n]$ è uguale all'originale. **Invariante interno** (I1)–(I3): $a[1..j] = b[1..j]$, $a[j+2..i] = b[j+1..i-1]$ e questi elementi sono $> key$.

Pseudocodice: sezione 13.3 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/13-correttezza/).

```python
%%writefile InsertionSort.java
public class InsertionSort
{
// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

static boolean equal(int[] x, int xs, int[] y, int ys, int len)     // x[xs..xs+len-1] == y[ys..ys+len-1]
{
    for (int k = 0; k < len; k++)
        if (x[xs + k] != y[ys + k]) return false;
    return true;
}

    // L'array con una barra dopo la parte ordinata a[1..i-1]
    static String split(int[] a, int i)
    {
        String s = fmt(a, 0, i);
        String t = fmt(a, i, a.length);
        return s.substring(0, s.length() - 1) + " | " + t.substring(1);
    }

    // Invariante esterno: a[1..i-1] = elementi originali di a[1..i-1] ordinati, a[i..n] = originale
    static void printOuter(int[] a, int[] orig, int i)
    {
        int n = a.length;
        int[] s = java.util.Arrays.copyOf(orig, i);
        java.util.Arrays.sort(s);
        boolean ok = equal(a, 0, s, 0, i) && equal(a, i, orig, i, n - i);
        System.out.println(String.format("%2d | %-26s | %-19s | %s", i + 1, split(a, i), fmt(orig, 0, i), ok ? "ok" : "FALSA"));
    }

    // Invariante interno (I1)-(I3); b = a a inizio iterazione
    static void printInner(int[] a, int[] b, int i, int j, int key)
    {
        int ip = i + 1, jp = j + 1;                // indici dello pseudocodice
        boolean ok = equal(a, 0, b, 0, jp) && equal(a, jp + 1, b, jp, ip - jp - 1);
        for (int k = jp + 1; k < ip; k++)
            if (a[k] <= key) ok = false;
        String cond = "j = 0: false";
        if (j >= 0)
        {
            boolean t = a[j] > key;
            cond = "a[" + jp + "] = " + a[j] + (t ? " > " : " <= ") + key + ": " + (t ? "true" : "false");
        }
        System.out.println(String.format("%d | %-15s | %-12s | %-9s | %-20s | %s", jp, fmt(a, 0, a.length), fmt(a, 0, jp), fmt(a, jp + 1, ip), cond, ok ? "ok" : "FALSA"));
    }

    // InsertionSort(a, n)
    static void insertionSort(int[] a, boolean showOuter, int traceInnerAt)
    {
        int n = a.length;
        int[] orig = a.clone();                    // copia dell'input: serve solo a verificare l'invariante
        int i = 1;                                 // i := 2
        if (showOuter) printOuter(a, orig, i);
        while (i < n)                              // i <= n
        {
            int[] b = a.clone();                   // contenuto di a a inizio iterazione (solo per l'invariante interno)
            int key = a[i];
            int j = i - 1;
            boolean tr = (i + 1 == traceInnerAt);
            if (tr) printInner(a, b, i, j, key);
            while (j >= 0 && a[j] > key)           // j >= 1 and a[j] > key
            {
                a[j + 1] = a[j];
                j = j - 1;
                if (tr) printInner(a, b, i, j, key);
            }
            a[j + 1] = key;
            i = i + 1;
            if (showOuter) printOuter(a, orig, i);
        }
    }

    public static void main(String[] args)
    {
        // Il ciclo esterno: a[1..5] = (5, 2, 4, 1, 3), esempio delle dispense
        int[] a = {5, 2, 4, 1, 3};
        int n = a.length;
        System.out.printf("a = %s\n", fmt(a, 0, n));
        System.out.printf(" i | a (ordinato | resto)       | a[1..i-1] originale | invariante\n");
        insertionSort(a, true, 0);
        System.out.printf("ordinato: %s\n", fmt(a, 0, n));

        // Il ciclo interno all'iterazione i = 5, con key = 3 (a = (1, 2, 4, 5, 3) all'inizio)
        System.out.printf("\n");
        int[] c = {5, 2, 4, 1, 3};
        System.out.printf("j | a               | a[1..j]      | a[j+2..i] | condizione           | invariante\n");
        insertionSort(c, false, 5);
        System.out.printf("dopo la riga 10 (a[j+1] = key): a = %s\n", fmt(c, 0, n));
    }
}
```

## 3. Ricerca binaria: BinarySearch

**Input:** un array $a[1..n]$ ordinato in modo non decrescente e un valore $m$. **Output:** un indice $k$ con $a[k] = m$ se $m$ è in $a$, $-1$ altrimenti. **Invariante:** (I1) $1 \le left$ e $right \le n$; (I2) se $m$ è in $a$, allora $m$ è in $a[left..right]$.

Pseudocodice: sezione 13.4 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/13-correttezza/).

```python
%%writefile BinarySearch.java
public class BinarySearch
{
// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

static boolean equal(int[] x, int xs, int[] y, int ys, int len)     // x[xs..xs+len-1] == y[ys..ys+len-1]
{
    for (int k = 0; k < len; k++)
        if (x[xs + k] != y[ys + k]) return false;
    return true;
}

    // Invariante (I1) 1 <= left e right <= n, (I2) se m e' in a allora e' in a[left..right]; mid = -1: intervallo vuoto
    static void printRow(int[] a, int m, int left, int right, int mid)
    {
        int n = a.length;
        boolean inA = false, inSub = false;
        for (int k = 0; k < n; k++)
        {
            if (a[k] == m)
            {
                inA = true;
                if (k >= left && k <= right) inSub = true;
            }
        }
        boolean ok = (left + 1 >= 1 && right + 1 <= n && (!inA || inSub));
        String action;
        if (mid < 0)
            action = "left > right => return -1";
        else if (a[mid] == m)
            action = String.format("mid = %d, a[%d] = %d = %d => return %d", mid + 1, mid + 1, a[mid], m, mid + 1);
        else if (a[mid] < m)
            action = String.format("mid = %d, a[%d] = %d < %d => left = %d", mid + 1, mid + 1, a[mid], m, mid + 2);
        else
            action = String.format("mid = %d, a[%d] = %d > %d => right = %d", mid + 1, mid + 1, a[mid], m, mid);
        System.out.println(String.format("%4d | %5d | %-27s | %-11s | %s", left + 1, right + 1, fmt(a, left, right + 1), ok ? "ok" : "FALSA", action));
    }

    // BinarySearch(a, n, m)
    static int binarySearch(int[] a, int m, boolean trace)
    {
        int n = a.length;
        int left = 0;                              // left := 1
        int right = n - 1;                         // right := n
        while (left <= right)
        {
            int mid = left + (right - left) / 2;
            if (trace) printRow(a, m, left, right, mid);   // riga dell'invariante a questa valutazione
            if (a[mid] == m)
            {
                return mid;
            }
            if (a[mid] < m)
            {
                left = mid + 1;
            }
            else
            {
                right = mid - 1;
            }
        }
        if (trace) printRow(a, m, left, right, -1);
        return -1;
    }

    public static void main(String[] args)
    {
        int[] a = {3, 9, 11, 12, 12, 16, 19};      // esempio delle dispense
        int n = a.length;
        System.out.printf("a = %s, n = %d\n", fmt(a, 0, n), n);

        int m = 10;
        System.out.printf("\n");
        System.out.printf("m = %d\n", m);
        System.out.printf("left | right | a[left..right]              | invariante  | azione\n");
        int k = binarySearch(a, m, true);
        System.out.printf("m = %d non trovato: restituisce %d\n", m, k);

        m = 12;
        System.out.printf("\n");
        System.out.printf("m = %d\n", m);
        System.out.printf("left | right | a[left..right]              | invariante  | azione\n");
        k = binarySearch(a, m, true);
        System.out.printf("m = %d trovato: indice %d nel codice (posizione %d nello pseudocodice)\n", m, k, k + 1);

        // Controllo esaustivo per m = 0, ..., 20
        int errors = 0;
        for (m = 0; m <= 20; m++)
        {
            k = binarySearch(a, m, false);
            boolean inA = false;
            for (int t = 0; t < n; t++) if (a[t] == m) inA = true;
            if ((inA && (k < 0 || a[k] != m)) || (!inA && k != -1)) errors++;
        }
        System.out.printf("\n");
        System.out.printf("Controllo per m = 0, ..., 20: %d errori\n", errors);
    }
}
```

## 4. Algoritmo di Euclide: EuclideanAlgorithm

**Input:** due numeri naturali $n, m \ge 1$. **Output:** $\gcd(n, m)$. **Invariante:** $n \ge 1$, $m \ge 0$ e $\gcd(n,m) = \gcd(n_0,m_0)$, dove $n_0, m_0$ sono i valori in input.

Pseudocodice: sezione 13.5 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/13-correttezza/).

```python
%%writefile EuclideanAlgorithm.java
public class EuclideanAlgorithm
{
// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

static boolean equal(int[] x, int xs, int[] y, int ys, int len)     // x[xs..xs+len-1] == y[ys..ys+len-1]
{
    for (int k = 0; k < len; k++)
        if (x[xs + k] != y[ys + k]) return false;
    return true;
}

    static int gcdCheck(int n, int m)              // mcd per enumerazione dei divisori, indipendente dall'algoritmo
    {
        if (m == 0) return n;
        int d = Math.min(n, m);
        while (n % d != 0 || m % d != 0) d = d - 1;
        return d;
    }

    // Invariante: n >= 1, m >= 0 e gcd(n, m) = gcd(n0, m0)
    static void printRow(int n, int m, int g0)
    {
        int g = gcdCheck(n, m);
        boolean ok = (n >= 1 && m >= 0 && g == g0);
        System.out.println(String.format("%4d | %4d | %9d | %s", n, m, g, ok ? "ok" : "FALSA"));
    }

    // EuclideanAlgorithm(n, m)
    static int euclideanAlgorithm(int n, int m)
    {
        int g0 = gcdCheck(n, m);                   // gcd(n0, m0), solo per verificare l'invariante
        printRow(n, m, g0);
        while (m != 0)
        {
            int r = n % m;
            n = m;
            m = r;
            printRow(n, m, g0);
        }
        return n;
    }

    public static void main(String[] args)
    {
        int n = 252, m = 105;      // esempio delle dispense
        System.out.printf("(n0, m0) = (%d, %d)\n", n, m);
        System.out.printf("   n |    m | gcd(n, m) | invariante\n");
        int g = euclideanAlgorithm(n, m);
        System.out.printf("Il ciclo termina con m = 0: restituisce n = %d\n", g);

        System.out.printf("\n");
        n = 48; m = 18;
        System.out.printf("(n0, m0) = (%d, %d)\n", n, m);
        System.out.printf("   n |    m | gcd(n, m) | invariante\n");
        g = euclideanAlgorithm(n, m);
        System.out.printf("Il ciclo termina con m = 0: restituisce n = %d\n", g);
    }
}
```
