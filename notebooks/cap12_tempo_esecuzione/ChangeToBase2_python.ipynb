**Indici.** Nello pseudocodice gli array sono indicizzati da 1; nel codice Python da 0: l'elemento `a[i]` dello pseudocodice è `a[i-1]` nel codice (e un ciclo `while i <= n` da 1 diventa `while i < n` da 0). Le tabelle stampate usano gli indici dello pseudocodice.

## Strumenti: il contatore dei passi

La cella definisce l'array `cnt` (una casella per riga dello pseudocodice), `reset()` per azzerarlo e `print_steps()` per stampare la tabella delle esecuzioni.

```python
cnt = [0] * 20      # cnt[r] = numero di esecuzioni della riga r (righe numerate come nelle dispense)


def reset():
    for r in range(len(cnt)):
        cnt[r] = 0


def print_steps(*lines):
    print("Riga | Esecuzioni")
    for r in lines:
        print("%4d | %10d" % (r, cnt[r]))
    print("Passi totali S = %d" % sum(cnt))
```

## Cambio di base: ChangeToBase2

**Input:** un numero naturale $n \ge 1$ in base 10. **Output:** l'array $d[0..nd-1]$ delle cifre binarie e il numero di cifre $nd$. Per ogni $n$: $S(n) = 6\lfloor\log_2 n\rfloor + 11 = \Theta(\log_2 n)$.

Pseudocodice: sezione 12.4 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
def floor_log2(n):                         # floor(log2 n) calcolato con un ciclo (non conta come passo)
    k = 0
    while n > 1:
        n = n // 2
        k = k + 1
    return k


def change_to_base2(n, trace=False):
    nd = floor_log2(n) + 1; cnt[2] += 1
    d = [0] * nd; cnt[3] += 1              # array d[0..nd-1], indicizzato da 0 come nello pseudocodice
    k = 0; cnt[4] += 1
    cnt[5] += 1                            # la condizione del while si conta a ogni valutazione
    while n > 0:
        q = n // 2; cnt[6] += 1
        r = n % 2; cnt[7] += 1
        d[k] = r; cnt[8] += 1
        if trace:                          # traccia
            print("%10d | %3d | %3d | %3d | d[%d] = %d" % (k + 1, n, q, r, k, r))
        n = q; cnt[9] += 1
        k = k + 1; cnt[10] += 1
        cnt[5] += 1
    cnt[12] += 1
    return d, nd
```

```python
# Esempio delle dispense: n = 13
n = 13
reset()
print("iterazione |   n |   q |   r | assegnamento")
d, nd = change_to_base2(n, True)
print("n = %d = (%s)_2: nd = %d, cifre d[nd-1..0] = %s" % (n, "".join(str(x) for x in reversed(d)), nd, "".join(str(x) for x in reversed(d))))
print_steps(2, 3, 4, 5, 6, 7, 8, 9, 10, 12)
print("S(n) = 6 floor(log2 n) + 11 = %d" % (6 * floor_log2(n) + 11))

# Il numero di passi cresce come log2 n
print()
print("       n | nd | S(n) | 6 floor(log2 n) + 11")
for n in [1, 2, 13, 100, 1000, 1000000]:
    reset()
    d, nd = change_to_base2(n)
    print("%8d | %2d | %4d | %d" % (n, nd, sum(cnt), 6 * floor_log2(n) + 11))
```

**Prova tu.** Cambia $n$ e verifica che $S(n) = 6\lfloor\log_2 n\rfloor + 11$.

```python
n = 2026        # <-- cambia il numero
reset()
d, nd = change_to_base2(n)
print("n = %d = (%s)_2, nd = %d, S = %d" % (n, "".join(str(x) for x in reversed(d)), nd, sum(cnt)))
```
