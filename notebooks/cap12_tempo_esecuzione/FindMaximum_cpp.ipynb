## Ricerca del massimo: FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il massimo valore di $a$. Nel caso peggiore (array crescente) $S(n) = 4n = \Theta(n)$.

Pseudocodice: sezione 12.2 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile find_maximum.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Riga | Esecuzioni\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Passi totali S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// FindMaximum(a, n)
int FindMaximum(int a[], int n)
{
    int max = a[0]; cnt[2]++;          // max := a[1]
    int i = 1; cnt[3]++;               // i := 2 nello pseudocodice
    cnt[4]++;                          // la condizione del while si conta a ogni valutazione
    while (i < n)                      // i <= n nello pseudocodice
    {
        cnt[5]++;
        if (a[i] > max)
        {
            max = a[i]; cnt[6]++;
        }
        i = i + 1; cnt[8]++;
        cnt[4]++;
    }
    cnt[10]++;
    return max;
}

long long StepsOf(int a[], int n)
{
    Reset();
    FindMaximum(a, n);
    return Total();
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Esempio delle dispense: a[1..4] = (3, 1, 4, 5)
    int a[] = {3, 1, 4, 5};
    int n = 4;
    Reset();
    int m = FindMaximum(a, n);
    printf("a = %s -> max = %d\n", Fmt(a, 0, n).c_str(), m);
    PrintSteps({2, 3, 4, 5, 6, 8, 10});
    long long t = cnt[6];
    printf("sum t(i) = %lld, S(n) = 3n + 1 + sum t(i) = %lld\n", t, 3 * n + 1 + t);

    // Caso peggiore (array crescente) e caso migliore (array decrescente)
    printf("\n");
    int inc[] = {1, 2, 3, 4};
    printf("a = %s (crescente): S = %lld = 4n = %d\n", Fmt(inc, 0, n).c_str(), StepsOf(inc, n), 4 * n);
    int dec[] = {4, 3, 2, 1};
    printf("a = %s (decrescente): S = %lld = 3n + 1 = %d\n", Fmt(dec, 0, n).c_str(), StepsOf(dec, n), 3 * n + 1);

    printf("\n");
    printf("Array crescente (1, 2, ..., n):\n");
    printf("  n | S(n) | 4n\n");
    for (int k = 1; k <= 8; k++)
    {
        int b[8];
        for (int i = 0; i < k; i++) b[i] = i + 1;
        printf("%3d | %4lld | %d\n", k, StepsOf(b, k), 4 * k);
    }
    return 0;
}
#endif
```

**Prova tu.** Cambia l'array: i passi stanno sempre fra $3n + 1$ e $4n$.

```python
%%writefile prova_find_maximum.cpp
#define SOLO_LIBRERIA   // non eseguire il main di find_maximum.cpp
#include "find_maximum.cpp"

int main()
{
    int a[] = {7, 2, 9, 4, 9, 1};      // <-- cambia l'array
    int n = 6;
    Reset();
    int m = FindMaximum(a, n);
    printf("a = %s -> max = %d\n", Fmt(a, 0, n).c_str(), m);
    printf("S = %lld, tra 3n + 1 = %d e 4n = %d\n", Total(), 3 * n + 1, 4 * n);
    return 0;
}
```
