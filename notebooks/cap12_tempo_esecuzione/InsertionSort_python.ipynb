**Indici.** Nello pseudocodice gli array sono indicizzati da 1; nel codice Python da 0: l'elemento `a[i]` dello pseudocodice è `a[i-1]` nel codice (e un ciclo `while i <= n` da 1 diventa `while i < n` da 0). Le tabelle stampate usano gli indici dello pseudocodice.

## Strumenti: il contatore dei passi

La cella definisce l'array `cnt` (una casella per riga dello pseudocodice), `reset()` per azzerarlo e `print_steps()` per stampare la tabella delle esecuzioni.

```python
cnt = [0] * 20      # cnt[r] = numero di esecuzioni della riga r (righe numerate come nelle dispense)


def reset():
    for r in range(len(cnt)):
        cnt[r] = 0


def print_steps(*lines):
    print("Riga | Esecuzioni")
    for r in lines:
        print("%4d | %10d" % (r, cnt[r]))
    print("Passi totali S = %d" % sum(cnt))
```

## Ordinamento per inserimento: InsertionSort

**Input:** un array $a[1..n]$ di $n$ numeri naturali. **Output:** l'array $a$ ordinato in modo non decrescente. Se $t(i)$ è il numero di spostamenti all'iterazione $i$, allora $S(n) = 6n - 4 + 3\sum_{i=2}^{n} t(i)$; nel caso peggiore (array decrescente) $S(n) = \frac{3n^2+9n-8}{2} = \Theta(n^2)$.

Pseudocodice: sezione 12.3 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
def insertion_sort(a, trace=False):
    n = len(a)
    prev = 0                               # solo per la traccia
    i = 1; cnt[2] += 1                     # i := 2 nello pseudocodice
    cnt[3] += 1
    while i < n:                           # i <= n nello pseudocodice
        key = a[i]; cnt[4] += 1
        j = i - 1; cnt[5] += 1
        cnt[6] += 1                        # la condizione del while si conta a ogni valutazione
        while j >= 0 and a[j] > key:       # j >= 1 and a[j] > key nello pseudocodice
            a[j + 1] = a[j]; cnt[7] += 1
            j = j - 1; cnt[8] += 1
            cnt[6] += 1
        a[j + 1] = key; cnt[10] += 1
        i = i + 1; cnt[11] += 1
        cnt[3] += 1
        if trace:                          # traccia: dopo l'incremento, i e' l'indice dello pseudocodice dell'iterazione appena finita
            print("%2d | %3d | %4d | %s" % (i, key, cnt[7] - prev, a))
            prev = cnt[7]
```

```python
def steps_of(a):
    reset()
    insertion_sort(a)
    return sum(cnt)


# Esempio delle dispense: a[1..6] = (5, 2, 4, 6, 1, 3)
a = [5, 2, 4, 6, 1, 3]
n = len(a)
print("a = %s" % a)
reset()
print(" i | key | t(i) | array dopo l'iterazione")
insertion_sort(a, True)
print("ordinato: %s" % a)
print_steps(2, 3, 4, 5, 6, 7, 8, 10, 11)
t = cnt[7]
print("sum t(i) = %d, S(n) = 6n - 4 + 3 sum t(i) = %d" % (t, 6 * n - 4 + 3 * t))

# Caso peggiore: array decrescente, t(i) = i - 1
print()
a = [6, 5, 4, 3, 2, 1]
print("a = %s (decrescente): S = %d = (3n^2 + 9n - 8)/2 = %d" % (a, steps_of(a[:]), (3 * n * n + 9 * n - 8) // 2))
print("sum t(i) = %d = n(n-1)/2 = %d" % (cnt[7], n * (n - 1) // 2))

print()
print("Array decrescente (n, ..., 2, 1):")
print("  n | S(n) | (3n^2 + 9n - 8)/2")
for n in range(1, 9):
    print("%3d | %4d | %d" % (n, steps_of(list(range(n, 0, -1))), (3 * n * n + 9 * n - 8) // 2))
```

**Prova tu.** Cambia l'array e controlla la formula $6n - 4 + 3\sum t(i)$.

```python
a = [8, 3, 5, 1, 9, 2, 7]      # <-- cambia l'array
n = len(a)
reset()
print(" i | key | t(i) | array dopo l'iterazione")
insertion_sort(a, True)
print_steps(2, 3, 4, 5, 6, 7, 8, 10, 11)
print("6n - 4 + 3 sum t(i) = %d" % (6 * n - 4 + 3 * cnt[7]))
```
