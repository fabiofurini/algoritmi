**Indici.** Nello pseudocodice gli array sono indicizzati da 1; nel codice Python da 0: l'elemento `a[i]` dello pseudocodice è `a[i-1]` nel codice (e un ciclo `while i <= n` da 1 diventa `while i < n` da 0). Le tabelle stampate usano gli indici dello pseudocodice.

## Strumenti: il contatore dei passi

La cella definisce l'array `cnt` (una casella per riga dello pseudocodice), `reset()` per azzerarlo e `print_steps()` per stampare la tabella delle esecuzioni.

```python
cnt = [0] * 20      # cnt[r] = numero di esecuzioni della riga r (righe numerate come nelle dispense)


def reset():
    for r in range(len(cnt)):
        cnt[r] = 0


def print_steps(*lines):
    print("Riga | Esecuzioni")
    for r in lines:
        print("%4d | %10d" % (r, cnt[r]))
    print("Passi totali S = %d" % sum(cnt))
```

## Ricerca del massimo: FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il massimo valore di $a$. Nel caso peggiore (array crescente) $S(n) = 4n = \Theta(n)$.

Pseudocodice: sezione 12.2 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
def find_maximum(a):
    n = len(a)
    max_val = a[0]; cnt[2] += 1        # max := a[1]
    i = 1; cnt[3] += 1                 # i := 2 nello pseudocodice
    cnt[4] += 1                        # la condizione del while si conta a ogni valutazione
    while i < n:                       # i <= n nello pseudocodice
        cnt[5] += 1
        if a[i] > max_val:
            max_val = a[i]; cnt[6] += 1
        i = i + 1; cnt[8] += 1
        cnt[4] += 1
    cnt[10] += 1
    return max_val
```

```python
def steps_of(a):
    reset()
    find_maximum(a)
    return sum(cnt)


# Esempio delle dispense: a[1..4] = (3, 1, 4, 5)
a = [3, 1, 4, 5]
n = len(a)
reset()
m = find_maximum(a)
print("a = %s -> max = %d" % (a, m))
print_steps(2, 3, 4, 5, 6, 8, 10)
t = cnt[6]
print("sum t(i) = %d, S(n) = 3n + 1 + sum t(i) = %d" % (t, 3 * n + 1 + t))

# Caso peggiore (array crescente) e caso migliore (array decrescente)
print()
a = [1, 2, 3, 4]
print("a = %s (crescente): S = %d = 4n = %d" % (a, steps_of(a), 4 * n))
a = [4, 3, 2, 1]
print("a = %s (decrescente): S = %d = 3n + 1 = %d" % (a, steps_of(a), 3 * n + 1))

print()
print("Array crescente (1, 2, ..., n):")
print("  n | S(n) | 4n")
for n in range(1, 9):
    print("%3d | %4d | %d" % (n, steps_of(list(range(1, n + 1))), 4 * n))
```

**Prova tu.** Cambia l'array: i passi stanno sempre fra $3n + 1$ e $4n$.

```python
a = [7, 2, 9, 4, 9, 1]      # <-- cambia l'array
n = len(a)
reset()
m = find_maximum(a)
print("a = %s -> max = %d" % (a, m))
print("S = %d, tra 3n + 1 = %d e 4n = %d" % (sum(cnt), 3 * n + 1, 4 * n))
```
