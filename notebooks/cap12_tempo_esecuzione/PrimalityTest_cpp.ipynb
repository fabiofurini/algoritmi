## Test di primalità: PrimalityTest

**Input:** un numero naturale $n$. **Output:** `true` se $n$ è primo, `false` altrimenti. Nel caso peggiore ($n$ primo) $S(n) = 3\lfloor\sqrt{n}\rfloor + 1 = \Theta(\sqrt{n})$, cioè $2^{\Theta(|I|)}$ in funzione della dimensione dell'input.

Pseudocodice: sezione 12.5 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile primality_test.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Riga | Esecuzioni\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Passi totali S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

long long FloorSqrt(long long n)           // floor(sqrt(n)) in aritmetica intera
{
    long long r = (long long) sqrtl((long double) n);
    while (r * r > n) r--;
    while ((r + 1) * (r + 1) <= n) r++;
    return r;
}

// PrimalityTest(n)
bool PrimalityTest(long long n)
{
    cnt[2]++;
    if (n < 2)
    {
        cnt[3]++;
        return false;
    }
    long long d = 2; cnt[5]++;
    cnt[6]++;                              // la condizione del while si conta a ogni valutazione
    while (d <= FloorSqrt(n))
    {
        cnt[7]++;
        if (n % d == 0)
        {
            cnt[8]++;
            return false;
        }
        d = d + 1; cnt[10]++;
        cnt[6]++;
    }
    cnt[12]++;
    return true;
}

int Bits(long long n)                      // numero di bit di n >= 1
{
    int k = 0;
    while (n > 0)
    {
        n = n / 2;
        k = k + 1;
    }
    return k;
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Esempio delle dispense: un numero primo e uno composto
    long long n = 97;
    Reset();
    bool r = PrimalityTest(n);
    printf("n = %lld: %s\n", n, r ? "true" : "false");
    PrintSteps({2, 3, 5, 6, 7, 8, 10, 12});
    printf("S(n) = 3 floor(sqrt(n)) + 1 = %lld\n", 3 * FloorSqrt(n) + 1);

    printf("\n");
    n = 91;
    Reset();
    r = PrimalityTest(n);
    long long p = 2;
    while (n % p != 0) p = p + 1;
    printf("n = %lld: %s, p = %lld\n", n, r ? "true" : "false", p);
    PrintSteps({2, 3, 5, 6, 7, 8, 10, 12});
    printf("S(n) = 3p - 1 = %lld, mentre floor(sqrt(n)) = %lld\n", 3 * p - 1, FloorSqrt(n));

    printf("\n");
    n = 1;
    Reset();
    r = PrimalityTest(n);
    printf("n = %lld: %s, S = %lld\n", n, r ? "true" : "false", Total());

    // Il caso peggiore sono i numeri primi: S cresce come sqrt(n), cioe' raddoppia ogni 2 bit in piu'
    printf("\n");
    printf("|I| |        n |  S(n) | 3 floor(sqrt(n)) + 1\n");
    long long primes[] = {1021, 4093, 16381, 65521, 262139, 1048573, 4194301, 16777213};
    for (long long q : primes)
    {
        Reset();
        PrimalityTest(q);
        printf("%3d | %8lld | %5lld | %lld\n", Bits(q), q, Total(), 3 * FloorSqrt(q) + 1);
    }
    return 0;
}
#endif
```

```python
!g++ -std=c++17 -O2 -o primality_test primality_test.cpp && ./primality_test
```

**Prova tu.** Cambia $n$: per un primo $S = 3\lfloor\sqrt{n}\rfloor + 1$, per un composto $S = 3p - 1$.

```python
%%writefile prova_primality_test.cpp
#define SOLO_LIBRERIA   // non eseguire il main di primality_test.cpp
#include "primality_test.cpp"

int main()
{
    long long n = 7919;        // <-- cambia il numero
    Reset();
    bool r = PrimalityTest(n);
    printf("n = %lld: %s, S = %lld, 3 floor(sqrt(n)) + 1 = %lld\n", n, r ? "true" : "false", Total(), 3 * FloorSqrt(n) + 1);
    return 0;
}
```
