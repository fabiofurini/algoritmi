## Ricerca del massimo: FindMaximum

**Input:** un array $a[1..n]$ di $n \ge 1$ numeri interi. **Output:** il massimo valore di $a$. Nel caso peggiore (array crescente) $S(n) = 4n = \Theta(n)$.

Pseudocodice: sezione 12.2 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
%%writefile FindMaximum.java
public class FindMaximum
{
static long[] cnt = new long[20];   // cnt[r] = numero di esecuzioni della riga r (numerazione delle dispense)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Riga | Esecuzioni");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Passi totali S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    // FindMaximum(a, n)
    static int findMaximum(int[] a)
    {
        int n = a.length;
        int max = a[0]; cnt[2]++;          // max := a[1]
        int i = 1; cnt[3]++;               // i := 2 nello pseudocodice
        cnt[4]++;                          // la condizione del while si conta a ogni valutazione
        while (i < n)                      // i <= n nello pseudocodice
        {
            cnt[5]++;
            if (a[i] > max)
            {
                max = a[i]; cnt[6]++;
            }
            i = i + 1; cnt[8]++;
            cnt[4]++;
        }
        cnt[10]++;
        return max;
    }

    static long stepsOf(int[] a)
    {
        reset();
        findMaximum(a);
        return total();
    }

    public static void main(String[] args)
    {
        // Esempio delle dispense: a[1..4] = (3, 1, 4, 5)
        int[] a = {3, 1, 4, 5};
        int n = a.length;
        reset();
        int m = findMaximum(a);
        System.out.printf("a = %s -> max = %d\n", fmt(a, 0, n), m);
        printSteps(2, 3, 4, 5, 6, 8, 10);
        long t = cnt[6];
        System.out.printf("sum t(i) = %d, S(n) = 3n + 1 + sum t(i) = %d\n", t, 3 * n + 1 + t);

        // Caso peggiore (array crescente) e caso migliore (array decrescente)
        System.out.printf("\n");
        int[] inc = {1, 2, 3, 4};
        System.out.printf("a = %s (crescente): S = %d = 4n = %d\n", fmt(inc, 0, n), stepsOf(inc), 4 * n);
        int[] dec = {4, 3, 2, 1};
        System.out.printf("a = %s (decrescente): S = %d = 3n + 1 = %d\n", fmt(dec, 0, n), stepsOf(dec), 3 * n + 1);

        System.out.printf("\n");
        System.out.printf("Array crescente (1, 2, ..., n):\n");
        System.out.printf("  n | S(n) | 4n\n");
        for (int k = 1; k <= 8; k++)
        {
            int[] b = new int[k];
            for (int i = 0; i < k; i++) b[i] = i + 1;
            System.out.printf("%3d | %4d | %d\n", k, stepsOf(b), 4 * k);
        }
    }
}
```

**Prova tu.** Cambia l'array: i passi stanno sempre fra $3n + 1$ e $4n$.

```python
%%writefile ProvaFindMaximum.java
public class ProvaFindMaximum
{
    public static void main(String[] args)
    {
        int[] a = {7, 2, 9, 4, 9, 1};      // <-- cambia l'array
        int n = a.length;
        FindMaximum.reset();
        int m = FindMaximum.findMaximum(a);
        System.out.printf("a = %s -> max = %d\n", FindMaximum.fmt(a, 0, n), m);
        System.out.printf("S = %d, tra 3n + 1 = %d e 4n = %d\n", FindMaximum.total(), 3 * n + 1, 4 * n);
    }
}
```
