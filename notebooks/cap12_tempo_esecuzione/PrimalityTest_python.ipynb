## Strumenti: il contatore dei passi

La cella definisce l'array `cnt` (una casella per riga dello pseudocodice), `reset()` per azzerarlo e `print_steps()` per stampare la tabella delle esecuzioni.

```python
cnt = [0] * 20      # cnt[r] = numero di esecuzioni della riga r (righe numerate come nelle dispense)


def reset():
    for r in range(len(cnt)):
        cnt[r] = 0


def print_steps(*lines):
    print("Riga | Esecuzioni")
    for r in lines:
        print("%4d | %10d" % (r, cnt[r]))
    print("Passi totali S = %d" % sum(cnt))
```

## Test di primalità: PrimalityTest

**Input:** un numero naturale $n$. **Output:** `true` se $n$ è primo, `false` altrimenti. Nel caso peggiore ($n$ primo) $S(n) = 3\lfloor\sqrt{n}\rfloor + 1 = \Theta(\sqrt{n})$, cioè $2^{\Theta(|I|)}$ in funzione della dimensione dell'input.

Pseudocodice: sezione 12.5 della [pagina del capitolo](https://fabiofurini.github.io/algoritmi/analisi/12-tempo-esecuzione/).

```python
from math import isqrt             # floor(sqrt(n)) in aritmetica intera


def primality_test(n):
    cnt[2] += 1
    if n < 2:
        cnt[3] += 1
        return False
    d = 2; cnt[5] += 1
    cnt[6] += 1                        # la condizione del while si conta a ogni valutazione
    while d <= isqrt(n):
        cnt[7] += 1
        if n % d == 0:
            cnt[8] += 1
            return False
        d = d + 1; cnt[10] += 1
        cnt[6] += 1
    cnt[12] += 1
    return True
```

```python
def risposta(r):
    return "true" if r else "false"


# Esempio delle dispense: un numero primo e uno composto
n = 97
reset()
r = primality_test(n)
print("n = %d: %s" % (n, risposta(r)))
print_steps(2, 3, 5, 6, 7, 8, 10, 12)
print("S(n) = 3 floor(sqrt(n)) + 1 = %d" % (3 * isqrt(n) + 1))

print()
n = 91
reset()
r = primality_test(n)
p = 2
while n % p != 0:
    p = p + 1
print("n = %d: %s, p = %d" % (n, risposta(r), p))
print_steps(2, 3, 5, 6, 7, 8, 10, 12)
print("S(n) = 3p - 1 = %d, mentre floor(sqrt(n)) = %d" % (3 * p - 1, isqrt(n)))

print()
n = 1
reset()
r = primality_test(n)
print("n = %d: %s, S = %d" % (n, risposta(r), sum(cnt)))

# Il caso peggiore sono i numeri primi: S cresce come sqrt(n), cioe' raddoppia ogni 2 bit in piu'
print()
print("|I| |        n |  S(n) | 3 floor(sqrt(n)) + 1")
for n in [1021, 4093, 16381, 65521, 262139, 1048573, 4194301, 16777213]:
    reset()
    r = primality_test(n)
    print("%3d | %8d | %5d | %d" % (n.bit_length(), n, sum(cnt), 3 * isqrt(n) + 1))
```

**Prova tu.** Cambia $n$: per un primo $S = 3\lfloor\sqrt{n}\rfloor + 1$, per un composto $S = 3p - 1$.

```python
n = 7919        # <-- cambia il numero
reset()
r = primality_test(n)
print("n = %d: %s, S = %d, 3 floor(sqrt(n)) + 1 = %d" % (n, "true" if r else "false", sum(cnt), 3 * isqrt(n) + 1))
```
